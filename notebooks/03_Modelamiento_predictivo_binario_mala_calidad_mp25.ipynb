```python
# =============================================================================
# 1. CONFIGURACIÓN, DEPENDENCIAS Y FUNCIONES HTML PREMIUM
# =============================================================================
import os
import sys
import time
import math
import html
import re
import unicodedata
import warnings
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
from scipy.special import expit
import statsmodels.api as sm
from statsmodels.stats.diagnostic import het_breuschpagan
from statsmodels.stats.outliers_influence import variance_inflation_factor
from IPython.display import display, HTML
from sklearn.experimental import enable_iterative_imputer  # noqa: F401
from sklearn.impute import SimpleImputer, IterativeImputer
from sklearn.linear_model import LogisticRegression, BayesianRidge
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    confusion_matrix, precision_score, recall_score, f1_score, fbeta_score,
    roc_auc_score, average_precision_score, balanced_accuracy_score,
    accuracy_score, brier_score_loss, roc_curve, precision_recall_curve
)
from sklearn.calibration import calibration_curve
from sklearn.exceptions import ConvergenceWarning
from sklearn.base import BaseEstimator, TransformerMixin
# Se silencian únicamente avisos de compatibilidad futura. Los avisos numéricos
# de statsmodels se capturan dentro del ajuste seguro, se registran y se validan;
# no quedan mensajes sin control en stderr.
warnings.filterwarnings("ignore", category=FutureWarning)
warnings.filterwarnings(
    "ignore",
    message=r"Found unknown categories in columns.*",
    category=UserWarning,
)
# IterativeImputer puede agotar el máximo sin afectar la ejecución. La calidad
# no se presume: se valida externamente con enmascaramiento, NRMSE y encogimiento.
warnings.filterwarnings(
    "ignore",
    message=r"\[IterativeImputer\] Early stopping criterion not reached\.",
    category=ConvergenceWarning,
)
NOTEBOOK_VERSION = "S3-FINAL-100-R6"
#print("VERSIÓN ACTIVA: S3-FINAL-100-R6 · versión oficial unificada")
#print("No continuar si esta leyenda no aparece exactamente al ejecutar.")
AUDITORIA_AJUSTES_GLM = []
SEED = 2026
ALPHA = 0.05
MIN_RECALL = 0.70
COSTO_FN = 5.0
COSTO_FP = 1.0
# Márgenes de equivalencia definidos antes de observar el conjunto de prueba.
PR_AUC_EQUIV = 0.01
RECALL_EQUIV = 0.02
BRIER_EQUIV = 0.005
DELTA_BIC_RELEVANTE = 2.0
MIN_RETENCION_COMPLETOS = 0.99
COBERTURA_MINIMA = 0.75
HORAS_VALIDAS_MINIMAS = 18
COBERTURA_INSUFICIENTE_REFERENCIA_S2 = 0.278
COBERTURA_MERGE_MINIMA = 0.95
COBERTURA_ESTRICTA = os.getenv("S3_COBERTURA_ESTRICTA", "1") == "1"
FAST_MODE = os.getenv("S3_FAST", "0") == "1"
N_BOOT_MODEL = int(os.getenv("S3_BOOTSTRAP", "500" if FAST_MODE else "10000"))
N_MASK_REPEATS = 1 if FAST_MODE else 5
N_CV_SPLITS = 3 if FAST_MODE else 5
rng = np.random.default_rng(SEED)
CWD = Path.cwd().resolve()
if CWD.name.lower() == "notebooks":
    PROJECT_ROOT = CWD.parent
elif (CWD / "data").exists():
    PROJECT_ROOT = CWD
else:
    PROJECT_ROOT = next((p for p in [CWD, *CWD.parents] if (p / "data").exists()), CWD)
DATA_DIR = PROJECT_ROOT / "data"
CSV_DIR = DATA_DIR
FIGURES_DIR = PROJECT_ROOT / "figures" / "S3"
RESULTS_DIR = PROJECT_ROOT / "resultados" / "S3"

CSV_DIR.mkdir(parents=True, exist_ok=True)
FIGURES_DIR.mkdir(parents=True, exist_ok=True)
RESULTS_DIR.mkdir(parents=True, exist_ok=True)
plt.rcParams.update({
    "figure.figsize": (11.5, 6.2), "figure.dpi": 130, "savefig.dpi": 300,
    "axes.grid": True, "grid.alpha": 0.25, "axes.titleweight": "bold",
    "axes.titlesize": 14, "axes.labelsize": 11, "font.size": 10,
    "legend.frameon": True, "legend.framealpha": 0.95,
    "axes.spines.top": False, "axes.spines.right": False,
})
def guardar_figura(nombre):
    ruta = FIGURES_DIR / nombre
    plt.savefig(ruta, bbox_inches="tight", pad_inches=0.25, facecolor="white")
    return ruta
def exp_segura(valor):
    """Exponencial numéricamente estable para odds ratios e intervalos."""
    arr = np.asarray(valor, dtype=float)
    salida = np.exp(np.clip(arr, -700, 700))
    if np.ndim(valor) == 0:
        return float(salida)
    return salida
def validar_ajuste_glm(ajuste, X, y, contexto="GLM"):
    """
    Valida numéricamente un GLM binomial sin depender del cálculo interno de
    log-verosimilitud de statsmodels, que puede emitir avisos de overflow ante
    cuasi-separación o predicciones prácticamente 0/1.
    Un ajuste solo se acepta si:
    - convergió;
    - todos sus coeficientes son finitos;
    - las probabilidades calculadas son finitas;
    - la log-verosimilitud estabilizada es finita.
    """
    parametros = np.asarray(ajuste.params, dtype=float)
    X_arr = np.asarray(X, dtype=float)
    y_arr = np.asarray(y, dtype=float)

    if not bool(getattr(ajuste, "converged", True)):
        raise RuntimeError(f"{contexto}: el GLM no convergió.")
    if not np.all(np.isfinite(parametros)):
        raise FloatingPointError(
            f"{contexto}: se obtuvieron coeficientes no finitos."
        )
    eta = X_arr @ parametros
    probabilidades = expit(np.clip(eta, -35.0, 35.0))
    probabilidades = np.clip(probabilidades, 1e-12, 1 - 1e-12)

    if not np.all(np.isfinite(probabilidades)):
        raise FloatingPointError(
            f"{contexto}: se obtuvieron probabilidades no finitas."
        )
    log_likelihood = float(
        np.sum(
            y_arr * np.log(probabilidades)
            + (1.0 - y_arr) * np.log1p(-probabilidades)
        )
    )
    if not np.isfinite(log_likelihood):
        raise FloatingPointError(
            f"{contexto}: la log-verosimilitud estabilizada no es finita."
        )
    k = len(parametros)
    n = len(y_arr)
    pearson = float(
        np.sum(
            (y_arr - probabilidades) ** 2
            / np.clip(
                probabilidades * (1.0 - probabilidades),
                1e-12,
                None,
            )
        )
    )
    return {
        "log_likelihood": log_likelihood,
        "aic": -2.0 * log_likelihood + 2.0 * k,
        "bic": -2.0 * log_likelihood + k * np.log(max(n, 1)),
        "deviance": -2.0 * log_likelihood,
        "pearson_chi2": pearson,
        "prob_min": float(probabilidades.min()),
        "prob_max": float(probabilidades.max()),
        "n": n,
        "k": k,
    }
def ajustar_glm_binomial_seguro(
    y,
    X,
    *,
    maxiter=300,
    contexto="GLM",
    cov_type=None,
    cov_kwds=None,
    nivel_auditoria="critico",
):
    """
    Ajusta un GLM binomial capturando todos los avisos emitidos durante el fit.
    Los avisos no se imprimen en stderr: quedan registrados en
    ``AUDITORIA_AJUSTES_GLM`` y el ajuste se valida con cálculos estabilizados.
    ``nivel_auditoria`` distingue:
    - ``critico``: ajustes que deben ser válidos para cerrar la entrega;
    - ``exploratorio``: candidatos de selección o pruebas diagnósticas que
      pueden ser rechazados de manera controlada sin invalidar el modelo final.
    El ajuste que llama a esta función siempre recibe una excepción si falla;
    el flujo exploratorio decide si corresponde continuar con otros candidatos.
    """
    nivel_auditoria = str(nivel_auditoria).strip().lower()
    if nivel_auditoria not in {"critico", "exploratorio"}:
        raise ValueError(
            "nivel_auditoria debe ser 'critico' o 'exploratorio'."
        )
    modelo = sm.GLM(
        y,
        X,
        family=sm.families.Binomial(),
    )
    argumentos = {
        "maxiter": maxiter,
        "disp": 0,
    }
    if cov_type is not None:
        argumentos["cov_type"] = cov_type
    if cov_kwds is not None:
        argumentos["cov_kwds"] = cov_kwds

    avisos_capturados = []
    try:
        with warnings.catch_warnings(record=True) as avisos:
            warnings.simplefilter("always")
            with np.errstate(
                over="ignore",
                divide="ignore",
                invalid="ignore",
                under="ignore",
            ):
                ajuste = modelo.fit(**argumentos) # type: ignore

        avisos_capturados = [
            {
                "categoria": aviso.category.__name__,
                "mensaje": str(aviso.message),
            }
            for aviso in avisos
        ]
        diagnostico = validar_ajuste_glm(
            ajuste,
            X,
            y,
            contexto=contexto,
        )
    except Exception as exc:
        AUDITORIA_AJUSTES_GLM.append({
            "contexto": contexto,
            "nivel_auditoria": nivel_auditoria,
            "ajuste_validado": False,
            "estado_auditoria": (
                "FALLO_CRITICO"
                if nivel_auditoria == "critico"
                else "RECHAZO_EXPLORATORIO_CONTROLADO"
            ),
            "n_avisos_capturados": len(avisos_capturados),
            "categorias": ", ".join(sorted({
                a["categoria"] for a in avisos_capturados
            })),
            "mensajes": " | ".join(sorted({
                a["mensaje"] for a in avisos_capturados
            }))[:1000],
            "error": str(exc),
        })
        raise
    AUDITORIA_AJUSTES_GLM.append({
        "contexto": contexto,
        "nivel_auditoria": nivel_auditoria,
        "ajuste_validado": True,
        "estado_auditoria": (
            "VALIDADO_CRITICO"
            if nivel_auditoria == "critico"
            else "VALIDADO_EXPLORATORIO"
        ),
        "n_avisos_capturados": len(avisos_capturados),
        "categorias": ", ".join(sorted({
            a["categoria"] for a in avisos_capturados
        })),
        "mensajes": " | ".join(sorted({
            a["mensaje"] for a in avisos_capturados
        }))[:1000],
        "error": "",
    })
    return modelo, ajuste, diagnostico
def formatear_or(valor):
    """Presenta OR extremos sin redondearlos engañosamente a 0,000."""
    if pd.isna(valor):
        return "—"
    valor = float(valor)
    if valor == 0:
        return "0"
    if abs(valor) >= 1000 or abs(valor) < 0.001:
        return f"{valor:.3e}"
    return f"{valor:.3f}"
def _fmt_html(valor, decimales=3):
    """Formatea valores sin permitir saltos dentro de números."""
    if pd.isna(valor):
        return "—"
    if isinstance(valor, (pd.Timestamp, np.datetime64)):
        return pd.Timestamp(valor).strftime("%Y-%m-%d")
    if isinstance(valor, (bool, np.bool_)):
        return "Sí" if bool(valor) else "No"
    if isinstance(valor, (float, np.floating)):
        return f"{float(valor):,.{decimales}f}"
    if isinstance(valor, (int, np.integer)):
        return f"{int(valor):,}"
    return html.escape(str(valor))
def _etiqueta_columna(columna):
    """Convierte nombres técnicos en encabezados legibles."""
    return html.escape(str(columna).replace("_", " ").strip())
def _configurar_columna(serie, nombre):
    """
    Calcula un ancho mínimo por columna.
    La tabla puede desplazarse horizontalmente, pero no comprime números,
    métricas ni nombres hasta volverlos ilegibles.
    """
    nombre_limpio = str(nombre).replace("_", " ")
    nombre_lower = str(nombre).lower()
    columnas_texto_largo = {
        "lectura", "advertencia", "predictores", "regla", "criterio",
        "evidencia", "implicancia", "hallazgo", "uso", "uso_s3",
        "motivo_exclusion", "contenido", "resultado_validado",
    }
    if pd.api.types.is_bool_dtype(serie):
        return 92, "nowrap", "center"
    if pd.api.types.is_numeric_dtype(serie):
        ancho = max(96, min(138, 18 + len(nombre_limpio) * 7))
        return ancho, "nowrap", "right"
    muestra = serie.dropna().astype(str).head(100)
    max_chars = max([len(nombre_limpio)] + [len(x) for x in muestra.tolist()])

    if nombre_lower in columnas_texto_largo or max_chars > 44:
        ancho = max(230, min(390, 60 + min(max_chars, 48) * 6))
        return ancho, "normal", "left"
    ancho = max(125, min(245, 28 + max_chars * 7))
    return ancho, "nowrap", "left"

def tabla_html(df, titulo=None, max_filas=30, decimales=3):
    """
    Genera una tabla legible en GitHub, JupyterLab, VS Code y HTML exportado.
    - Cada columna recibe un ancho mínimo calculado.
    - Los valores numéricos nunca se dividen entre líneas.
    - Las columnas narrativas sí pueden envolver texto.
    - Las tablas anchas usan scroll horizontal sin comprimir su contenido.
    - La primera columna queda visualmente destacada durante el desplazamiento.
    """
    tabla = df.copy()
    recortada = len(tabla) > max_filas
    if recortada:
        tabla = tabla.head(max_filas)

    configuracion = {
        columna: _configurar_columna(tabla[columna], columna)
        for columna in tabla.columns
    }
    anchos = {columna: valores[0] for columna, valores in configuracion.items()}
    ancho_total = sum(anchos.values())
    ancho_tabla = max(760, ancho_total)

    colgroup = "".join(
        f'<col style="width:{anchos[col]}px;min-width:{anchos[col]}px;">'
        for col in tabla.columns
    )
    encabezados = []
    for indice, columna in enumerate(tabla.columns):
        ancho, _, alineacion = configuracion[columna]
        sticky = (
            "position:sticky;left:0;z-index:4;"
            "box-shadow:5px 0 8px rgba(15,23,42,.08);"
            if indice == 0 else ""
        )
        encabezados.append(
            f"""<th style="box-sizing:border-box;width:{ancho}px;min-width:{ancho}px;
                         padding:11px 13px;text-align:{alineacion};
                         background:#17375E !important;color:#FFFFFF !important;
                         border-bottom:3px solid #3B82F6;font-weight:900;
                         white-space:normal;word-break:normal;overflow-wrap:normal;
                         vertical-align:middle;line-height:1.25;{sticky}">
                  {_etiqueta_columna(columna)}
                </th>"""
        )
    filas_html = []
    for indice_fila, (_, fila) in enumerate(tabla.iterrows()):
        fondo = "#FFFFFF" if indice_fila % 2 == 0 else "#F3F6F9"
        celdas = []
        for indice_columna, columna in enumerate(tabla.columns):
            ancho, white_space, alineacion = configuracion[columna]
            sticky = (
                f"position:sticky;left:0;z-index:2;background:{fondo} !important;"
                "box-shadow:5px 0 8px rgba(15,23,42,.06);font-weight:700;"
                if indice_columna == 0 else f"background:{fondo} !important;"
            )
            celdas.append(
                f"""<td style="box-sizing:border-box;width:{ancho}px;min-width:{ancho}px;
                             padding:10px 13px;{sticky}color:#172033 !important;
                             border-bottom:1px solid #DDE5EE;vertical-align:middle;
                             line-height:1.45;white-space:{white_space};
                             word-break:normal;overflow-wrap:normal;text-align:{alineacion};
                             font-variant-numeric:tabular-nums;">
                      {_fmt_html(fila[columna], decimales)}
                    </td>"""
            )
        filas_html.append(f"<tr>{''.join(celdas)}</tr>")
    titulo_html = ""
    if titulo:
        titulo_html = f"""
        <div style="padding:14px 17px;background:#EAF2FA !important;
                    color:#17375E !important;border-bottom:1px solid #CBD5E1;
                    font-size:17px;font-weight:900;line-height:1.3;">
          {html.escape(titulo)}
        </div>"""
    nota = ""
    if recortada:
        nota = f"""
        <div style="padding:10px 14px;background:#FFF7ED !important;
                    color:#7C2D12 !important;border-top:1px solid #FED7AA;
                    font-size:12.5px;">
          Se muestran {max_filas} de {len(df):,} filas. La tabla completa se exporta a CSV.
        </div>"""
    ayuda_scroll = ""
    if ancho_total > 1250:
        ayuda_scroll = """
        <div style="padding:8px 14px;background:#F8FAFC !important;
                    color:#475569 !important;border-bottom:1px solid #E2E8F0;
                    font-size:12px;">
          ↔ Tabla extendida: desplácese horizontalmente. Los valores se mantienen completos.
        </div>"""
    return f"""
    <div style="box-sizing:border-box;width:100%;max-width:100%;margin:16px 0 20px 0;
                background:#FFFFFF !important;color:#172033 !important;color-scheme:light;
                border:1px solid #CBD5E1;border-radius:14px;overflow:hidden;
                box-shadow:0 7px 20px rgba(15,23,42,.08);
                font-family:Arial,Helvetica,sans-serif;">
      {titulo_html}
      {ayuda_scroll}
      <div style="box-sizing:border-box;width:100%;max-width:100%;overflow-x:auto;
                  overflow-y:hidden;scrollbar-gutter:stable;padding-bottom:3px;">
        <table style="width:{ancho_tabla}px;min-width:{ancho_tabla}px;
                      table-layout:fixed;border-collapse:separate;border-spacing:0;
                      background:#FFFFFF !important;color:#172033 !important;
                      font-size:13.5px;">
          <colgroup>{colgroup}</colgroup>
          <thead><tr>{''.join(encabezados)}</tr></thead>
          <tbody>{''.join(filas_html)}</tbody>
        </table>
      </div>
      {nota}
    </div>
    """
def mostrar_tabla(df, titulo=None, max_filas=30, decimales=3):
    display(HTML(tabla_html(df, titulo, max_filas, decimales)))
def mostrar_interpretacion(titulo, parrafos, nivel="info"):
    colores = {
        "info": ("#EAF4FF", "#1D6FB8", "#17375E"),
        "ok": ("#ECFDF3", "#15803D", "#14532D"),
        "alerta": ("#FFF7ED", "#B45309", "#7C2D12"),
        "riesgo": ("#FFF1F2", "#B42335", "#7F1D1D"),
    }
    fondo, borde, texto = colores.get(nivel, colores["info"])
    if isinstance(parrafos, str):
        parrafos = [parrafos]
    cuerpo = "".join(
        f"<p style='margin:{0 if i == 0 else 9}px 0 0 0;color:#26364D !important;'>{p}</p>"
        for i, p in enumerate(parrafos)
    )
    display(HTML(f"""
    <div style="box-sizing:border-box;width:100%;max-width:100%;overflow-wrap:anywhere;
                background:{fondo} !important;color:{texto} !important;color-scheme:light;
                border:1px solid #CBD5E1;border-left:7px solid {borde};border-radius:14px;
                padding:18px 21px;margin:16px 0;font-family:Arial,Helvetica,sans-serif;
                line-height:1.62;">
      <div style="color:{texto} !important;font-size:17px;font-weight:900;margin-bottom:7px;">
        {titulo}
      </div>
      {cuerpo}
    </div>
    """))
def mostrar_kpis(titulo, items):
    tarjetas = []
    for etiqueta, valor, detalle in items:
        tarjetas.append(f"""
        <div style="box-sizing:border-box;background:#FFFFFF !important;color:#172033 !important;
                    border:1px solid #CBD5E1;border-top:5px solid #2563EB;border-radius:12px;
                    padding:14px 15px;min-width:180px;">
          <div style="color:#40536B !important;font-size:12px;font-weight:900;text-transform:uppercase;
                      letter-spacing:.5px;">{html.escape(str(etiqueta))}</div>
          <div style="color:#17375E !important;font-size:22px;font-weight:900;margin-top:5px;">
            {html.escape(str(valor))}
          </div>
          <div style="color:#40536B !important;font-size:12.5px;margin-top:5px;line-height:1.4;">
            {html.escape(str(detalle))}
          </div>
        </div>""")
    display(HTML(f"""
    <div style="box-sizing:border-box;width:100%;max-width:100%;background:#F7FAFC !important;
                color:#172033 !important;color-scheme:light;border:1px solid #CBD5E1;
                border-radius:15px;padding:18px;margin:16px 0;font-family:Arial,Helvetica,sans-serif;">
      <div style="color:#17375E !important;font-size:18px;font-weight:900;margin-bottom:13px;">
        {html.escape(titulo)}
      </div>
      <div style="display:grid;grid-template-columns:repeat(auto-fit,minmax(180px,1fr));gap:12px;">
        {''.join(tarjetas)}
      </div>
    </div>
    """))
mostrar_kpis(
    "Entorno técnico",
    [
        ("Python", sys.version.split()[0], "Versión activa"),
        ("Semilla", SEED, "Reproducibilidad"),
        ("Bootstrap", f"{N_BOOT_MODEL:,}", "Remuestras del modelo"),
        ("CV temporal", N_CV_SPLITS, "Bloques expansivos"),
        ("Modo", "RÁPIDO" if FAST_MODE else "COMPLETO", "Control mediante S3_FAST"),
    ],
)
```

```python
# =============================================================================
# 2.1 CARGA ROBUSTA Y RECONSTRUCCIÓN DE LA MATRIZ ARMONIZADA S1/S2
# =============================================================================
def leer_csv_robusto(ruta: str | Path) -> pd.DataFrame:
    """
    Lee un archivo CSV probando codificaciones y separadores habituales.
    Las llamadas a ``pd.read_csv`` se realizan de forma explícita para que
    Pylance pueda validar correctamente los tipos de cada argumento. Esto
    evita las advertencias falsas generadas por un diccionario ``**kwargs``
    construido dinámicamente.
    """
    errores: list[str] = []
    codificaciones: tuple[str, ...] = ("utf-8-sig", "utf-8", "latin-1")
    separadores: tuple[str | None, ...] = (None, ";", ",")
    for codificacion in codificaciones:
        for separador in separadores:
            try:
                if separador is None:
                    # Detección automática del separador mediante el motor Python.
                    temporal = pd.read_csv(
                        ruta,
                        encoding=codificacion,
                        sep=None,
                        engine="python",
                    )
                else:
                    temporal = pd.read_csv(
                        ruta,
                        encoding=codificacion,
                        sep=separador,
                    )
                if temporal.shape[1] > 1:
                    return temporal

                errores.append(
                    f"{codificacion}/{separador}: solo se detectó una columna"
                )
            except (UnicodeDecodeError, pd.errors.ParserError, OSError, ValueError) as exc:
                errores.append(f"{codificacion}/{separador}: {exc}")

    detalle = " | ".join(errores[-8:])
    raise RuntimeError(
        "No fue posible leer el archivo CSV con las combinaciones "
        f"de codificación y separador disponibles. Detalle: {detalle}"
    )
CANDIDATOS_DATA = [
    DATA_DIR / "calidad_aire_diario.csv",
    PROJECT_ROOT / "calidad_aire_diario.csv",
    CWD / "data" / "calidad_aire_diario.csv",
    CWD.parent / "data" / "calidad_aire_diario.csv",
]
ruta_dataset = next((p for p in CANDIDATOS_DATA if p.exists()), None)
if ruta_dataset is None:
    raise FileNotFoundError(
        "No se encontró calidad_aire_diario.csv.\nRutas revisadas:\n- "
        + "\n- ".join(str(p) for p in CANDIDATOS_DATA)
    )
# ---------------------------------------------------------------------------
# A. Estructura física original: 16.071 filas × 36 columnas.
# ---------------------------------------------------------------------------
df_base = leer_csv_robusto(ruta_dataset)
df_base.columns = (
    df_base.columns.astype(str).str.strip().str.replace("\ufeff", "", regex=False)
)
col_fecha = next((c for c in ["fecha", "date", "Fecha"] if c in df_base.columns), None)
if col_fecha is None:
    raise ValueError("Se requiere una columna de fecha para realizar validación temporal.")
df_base[col_fecha] = pd.to_datetime(df_base[col_fecha], errors="coerce")
n_fechas_invalidas = int(df_base[col_fecha].isna().sum())
df_base = (
    df_base.loc[df_base[col_fecha].notna()]
    .sort_values(col_fecha)
    .reset_index(drop=True)
)
shape_base_original = df_base.shape
columnas_base_originales = df_base.columns.tolist()
# ---------------------------------------------------------------------------
# B. Enriquecimiento idéntico al utilizado en S1 y reconstruido en S2.
#    Cinco variables medias generan diez variables físicas min/max:
#    temperatura, presión, humedad, viento y radiación.
# ---------------------------------------------------------------------------
DELTAS_FISICOS_S1_S2 = {
    "temperatura_mean": {"min_sub": 6.5, "max_add": 8.0},
    "presion_mean": {"min_sub": 2.1, "max_add": 2.3},
    "humedad_mean": {"min_sub": 20.0, "max_add": 22.0},
    "viento_mean": {"min_sub": 1.2, "max_add": 1.8},
    "radiacion_mean": {"min_sub": 150.0, "max_add": 300.0},
}
df_raw = df_base.copy()
orden_enriquecido = []
def agregar_columna_unica(lista, columna):
    if columna in df_raw.columns and columna not in lista:
        lista.append(columna)
for columna in columnas_base_originales:
    agregar_columna_unica(orden_enriquecido, columna)
    if columna not in DELTAS_FISICOS_S1_S2:
        continue
    configuracion = DELTAS_FISICOS_S1_S2[columna]
    columna_min = columna.replace("_mean", "_min")
    columna_max = columna.replace("_mean", "_max")
    if columna_min not in df_raw.columns:
        df_raw[columna_min] = (
            pd.to_numeric(df_raw[columna], errors="coerce")
            - configuracion["min_sub"]
        )
    if columna_max not in df_raw.columns:
        df_raw[columna_max] = (
            pd.to_numeric(df_raw[columna], errors="coerce")
            + configuracion["max_add"]
        )
    if "humedad" in columna:
        df_raw[columna_min] = df_raw[columna_min].clip(lower=0, upper=100)
        df_raw[columna_max] = df_raw[columna_max].clip(lower=0, upper=100)
    if "viento" in columna or "radiacion" in columna:
        df_raw[columna_min] = df_raw[columna_min].clip(lower=0)
        df_raw[columna_max] = df_raw[columna_max].clip(lower=0)
    agregar_columna_unica(orden_enriquecido, columna_min)
    agregar_columna_unica(orden_enriquecido, columna_max)
# Si el archivo ya contiene alguna derivada, se preserva una sola vez.
for columna in df_raw.columns:
    agregar_columna_unica(orden_enriquecido, columna)
df_raw = df_raw.loc[:, orden_enriquecido]
shape_enriquecida = df_raw.shape
# Categorías ordenadas, consistentes con S1/S2.
orden_calidad = ["Bueno", "Regular", "Alerta", "Preemergencia", "Emergencia"]
orden_nse = [
    "Q1 (bajo)", "Q2 (medio-bajo)", "Q3 (medio)",
    "Q4 (medio-alto)", "Q5 (alto)",
]
if "calidad_aire_mp25" in df_raw.columns:
    df_raw["calidad_aire_mp25"] = pd.Categorical(
        df_raw["calidad_aire_mp25"],
        categories=orden_calidad,
        ordered=True,
    )
if "calidad_aire_mp10" in df_raw.columns:
    df_raw["calidad_aire_mp10"] = pd.Categorical(
        df_raw["calidad_aire_mp10"],
        categories=orden_calidad,
        ordered=True,
    )
if "nivel_socioeconomico" in df_raw.columns:
    df_raw["nivel_socioeconomico"] = pd.Categorical(
        df_raw["nivel_socioeconomico"],
        categories=orden_nse,
        ordered=True,
    )
# Línea base histórica del proyecto.
BASELINE_PROYECTO = {
    "filas_base": 16071,
    "columnas_base": 36,
    "filas_inferenciales": 11603,
    "columnas_enriquecidas": 46,
}
resumen_carga = pd.DataFrame([
    {
        "estructura": "CSV físico original",
        "filas": shape_base_original[0],
        "columnas": shape_base_original[1],
        "esperado": "16.071 × 36",
        "estado": (
            "Consistente"
            if shape_base_original == (
                BASELINE_PROYECTO["filas_base"],
                BASELINE_PROYECTO["columnas_base"],
            )
            else "Revisar versión del archivo"
        ),
    },
    {
        "estructura": "Matriz armonizada S1/S2",
        "filas": shape_enriquecida[0],
        "columnas": shape_enriquecida[1],
        "esperado": "16.071 × 46",
        "estado": (
            "Consistente"
            if shape_enriquecida == (
                BASELINE_PROYECTO["filas_base"],
                BASELINE_PROYECTO["columnas_enriquecidas"],
            )
            else "Revisar enriquecimiento"
        ),
    },
])
resumen_archivo = pd.DataFrame([{
    "archivo": ruta_dataset.name,
    "fecha mínima": df_raw[col_fecha].min(),
    "fecha máxima": df_raw[col_fecha].max(),
    "fechas inválidas excluidas": n_fechas_invalidas,
    "duplicados exactos": int(df_raw.duplicated().sum()),
    "variables físicas incorporadas": shape_enriquecida[1] - shape_base_original[1],
}])
resumen_carga.to_csv(
    CSV_DIR / "s3_consistencia_estructura_s1_s2.csv",
    index=False,
    encoding="utf-8-sig",
)
resumen_archivo.to_csv(
    CSV_DIR / "s3_resumen_carga.csv",
    index=False,
    encoding="utf-8-sig",
)
mostrar_tabla(resumen_carga, "Continuidad estructural: archivo físico y matriz enriquecida")
mostrar_tabla(resumen_archivo, "Control técnico del archivo")
mostrar_tabla(
    df_raw.head(5),
    "Vista de la matriz armonizada de 46 variables",
    max_filas=5,
)
mostrar_interpretacion(
    "Carga y estructura de datos",
    [
        f"El archivo físico contiene <b>{shape_base_original[0]:,}</b> observaciones y "
        f"<b>{shape_base_original[1]}</b> columnas originales.",
        f"Al replicar la preparación de S1/S2 se recupera una matriz de "
        f"<b>{shape_enriquecida[0]:,} × {shape_enriquecida[1]}</b>.",
        "Las diez columnas adicionales corresponden a mínimos y máximos físicos derivados "
        "de temperatura, presión, humedad, viento y radiación; no son diez fuentes nuevas.",
        "Esta distinción evita confundir el inventario total de variables con el subconjunto "
        "de predictores que finalmente utiliza cada modelo.",
    ],
    "ok" if shape_enriquecida[1] == 46 else "alerta",
)
```

```python
# =============================================================================
# 2.2 CARGA Y AUDITORÍA DEL REPORTE PUENTE DE SUMATIVA 2
# =============================================================================
CANDIDATOS_S2 = [
    PROJECT_ROOT / "resultados" / "s2_reporte_resultados_validados_para_s3.csv",
    PROJECT_ROOT / "s2_reporte_resultados_validados_para_s3.csv",
    CWD / "resultados" / "s2_reporte_resultados_validados_para_s3.csv",
    CWD / "s2_reporte_resultados_validados_para_s3.csv",
]
ruta_reporte_s2 = next((p for p in CANDIDATOS_S2 if p.exists()), None)
if ruta_reporte_s2 is None:
    raise FileNotFoundError(
        "No se encontró s2_reporte_resultados_validados_para_s3.csv.\n"
        "Ubicación recomendada: resultados/s2_reporte_resultados_validados_para_s3.csv"
    )
reporte_s2 = leer_csv_robusto(ruta_reporte_s2)
reporte_s2.columns = (
    reporte_s2.columns.astype(str).str.strip().str.replace("\ufeff", "", regex=False)
)
COLUMNAS_S2 = {
    "categoria", "elemento", "resultado_validado", "estado", "uso_s3"
}
faltan_s2 = COLUMNAS_S2 - set(reporte_s2.columns)
if faltan_s2:
    raise ValueError("Columnas ausentes en el reporte S2: " + ", ".join(sorted(faltan_s2)))
estado_s2 = reporte_s2["estado"].astype(str).str.lower()
categoria_s2 = reporte_s2["categoria"].astype(str).str.lower()

evidencias_confirmadas_s2 = reporte_s2.loc[
    estado_s2.str.contains("robusta|validado|confiable|concordante", regex=True, na=False)
].copy()
evidencias_cautela_s2 = reporte_s2.loc[
    estado_s2.str.contains("cautela|sensible", regex=True, na=False)
].copy()
decisiones_s3 = reporte_s2.loc[
    categoria_s2.str.contains("decisión metodológica s3|decision metodologica s3", regex=True, na=False)
].copy()
resumen_s2 = pd.DataFrame([{
    "archivo": ruta_reporte_s2.name,
    "evidencias_totales": len(reporte_s2),
    "confirmadas_o_validadas": len(evidencias_confirmadas_s2),
    "sensibles_o_con_cautela": len(evidencias_cautela_s2),
    "decisiones_metodologicas_s3": len(decisiones_s3),
    "duplicados_exactos": int(reporte_s2.duplicated().sum()),
}])
resumen_s2.to_csv(
    CSV_DIR / "s3_trazabilidad_reporte_s2.csv", index=False, encoding="utf-8-sig"
)
decisiones_s3.to_csv(
    CSV_DIR / "s3_decisiones_metodologicas_importadas_desde_s2.csv",
    index=False,
    encoding="utf-8-sig",
)
mostrar_tabla(resumen_s2, "Auditoría del archivo puente S2→S3")
mostrar_tabla(
    reporte_s2[["categoria", "elemento", "resultado_validado", "estado", "uso_s3"]],
    "Evidencias metodológicas heredadas de Sumativa 2",
    max_filas=25,
)
mostrar_interpretacion(
    "Resumen de evidencias S2",
    [
        f"Se importaron <b>{len(reporte_s2)}</b> evidencias de S2; <b>{len(evidencias_confirmadas_s2)}</b> están confirmadas o validadas.",
        f"Los resultados sensibles se conservan como cautelas explícitas para S3.",
        "Este archivo guía el modelamiento, pero no forma parte de la matriz de entrenamiento porque no contiene observaciones estación-día.",
    ],
    "ok",
)
```

    <p style="margin:0 0 10px 0;">La unidad analítica es <b>estación-día</b>. El objetivo es
    <code>mala_calidad_mp25</code>; se valida únicamente en registros con medición diaria disponible de MP2.5 y, cuando no existe, se reconstruye con reglas
    documentadas y nunca se imputa.</p> <p style="margin:0 0 10px 0;"><b>Horizonte declarado:</b> clasificación contemporánea
    (<i>nowcasting</i>). Por ello MP10 del mismo día puede evaluarse como predictor adicional,
    pero siempre se reporta como proxy altamente correlacionado y no como antecedente causal. Se aplica una lista blanca de variables aprobadas y una detección por patrones para bloquear columnas derivadas del objetivo.</p>
    </div>
</div>

```python
# =============================================================================
# 3.1 OBJETIVO BINARIO Y AUDITORÍA DE FUGA
# =============================================================================
def normalizar_binaria(serie):
    s = pd.Series(serie).copy()
    if pd.api.types.is_bool_dtype(s):
        return s.astype(float)
    if pd.api.types.is_numeric_dtype(s):
        out = pd.to_numeric(s, errors="coerce")
        return out.where(out.isin([0, 1]))
    mapa = {
        "1": 1, "0": 0, "si": 1, "sí": 1, "true": 1, "verdadero": 1,
        "no": 0, "false": 0, "falso": 0, "mala": 1, "buena": 0,
    }
    return s.astype(str).str.strip().str.lower().map(mapa)
TARGET = "mala_calidad_mp25"
HORIZONTE_MODELO = "nowcasting_contemporaneo"
if TARGET in df_raw.columns:
    y_all = normalizar_binaria(df_raw[TARGET])
    fuente_target = "columna original mala_calidad_mp25, validada sobre MP2.5 disponible"
elif "calidad_aire_mp25" in df_raw.columns:
    calidad = df_raw["calidad_aire_mp25"].astype(str).str.strip().str.lower()
    y_all = calidad.map(
        lambda x: 0 if x == "bueno" else (1 if x not in ["nan", "none", ""] else np.nan)
    )
    fuente_target = "derivada de calidad_aire_mp25"
elif "mp2_5_mean" in df_raw.columns:
    mp25 = pd.to_numeric(df_raw["mp2_5_mean"], errors="coerce")
    y_all = (mp25 > 50).astype(float)
    y_all[mp25.isna()] = np.nan
    fuente_target = "fallback: MP2.5 diario > 50 µg/m³"
else:
    raise ValueError("No existe una variable válida para construir mala_calidad_mp25.")
# El target solo es válido cuando existe una medición diaria de MP2.5.
# Esta regla replica el filtro inferencial aplicado en S1 y S2.
if "mp2_5_mean" in df_raw.columns:
    mp25_disponible = pd.to_numeric(
        df_raw["mp2_5_mean"], errors="coerce"
    ).notna()
    y_all = y_all.where(mp25_disponible)
df = df_raw.copy()
df[TARGET] = y_all
filas_sin_target = int(df[TARGET].isna().sum())
df = df.loc[df[TARGET].notna()].copy()
df[TARGET] = df[TARGET].astype(int)
PATRONES_FUGA = [
    r"^mp2[_\.]?5", r"mala_calidad", r"critico.*mp25", r"calidad_aire_mp25",
    r"nivel_contaminacion", r"episodio.*mp25", r"estaciones_en_episodio",
]
columnas_fuga = []
for columna in df.columns:
    nombre = str(columna).lower()
    if columna == TARGET:
        columnas_fuga.append((columna, "variable objetivo"))
    elif any(pd.Series([nombre]).str.contains(pat, regex=True).iloc[0] for pat in PATRONES_FUGA):
        columnas_fuga.append((columna, "derivada directa o proxy prohibido de MP2.5"))
auditoria_fuga = pd.DataFrame(columnas_fuga, columns=["variable", "motivo_exclusion"]).drop_duplicates()
auditoria_fuga.to_csv(
    CSV_DIR / "s3_auditoria_fuga_variables.csv", index=False, encoding="utf-8-sig"
)
auditoria_faltantes = pd.DataFrame({
    "variable": df_raw.columns,
    "n_faltantes": df_raw.isna().sum().to_numpy(dtype=int),
    "porcentaje_faltantes": df_raw.isna().mean().to_numpy(dtype=float) * 100.0,
    "tipo": [str(t) for t in df_raw.dtypes],
}).sort_values("porcentaje_faltantes", ascending=False)
resumen_target = pd.DataFrame([{
    "fuente_target": fuente_target,
    "horizonte": HORIZONTE_MODELO,
    "filas_originales": len(df_raw),
    "filas_sin_target_excluidas": filas_sin_target,
    "filas_modelables": len(df),
    "positivos": int(df[TARGET].sum()),
    "prevalencia": df[TARGET].mean(),
    "filas esperadas S1/S2": BASELINE_PROYECTO["filas_inferenciales"],
    "estado continuidad": (
        "Consistente"
        if len(df) == BASELINE_PROYECTO["filas_inferenciales"]
        else "Revisar: no coincide con la matriz inferencial S1/S2"
    ),
}])
auditoria_faltantes.to_csv(
    CSV_DIR / "s3_auditoria_faltantes.csv", index=False, encoding="utf-8-sig"
)
resumen_target.to_csv(
    CSV_DIR / "s3_resumen_target.csv", index=False, encoding="utf-8-sig"
)
mostrar_tabla(resumen_target, "Definición y disponibilidad de la variable objetivo")
mostrar_tabla(auditoria_fuga, "Variables bloqueadas por riesgo de fuga", max_filas=30)
mostrar_tabla(auditoria_faltantes.head(20), "Variables con mayor proporción de faltantes", max_filas=20)
top_faltantes = auditoria_faltantes.head(15).sort_values("porcentaje_faltantes")
fig, ax = plt.subplots(figsize=(11, 6))
ax.barh(top_faltantes["variable"], top_faltantes["porcentaje_faltantes"])
ax.axvline(27.8, linestyle="--", linewidth=1.6, label="Referencia S1 ≈ 27,8%")
ax.set_title("Proporción de datos faltantes por variable")
ax.set_xlabel("Porcentaje de faltantes")
ax.set_ylabel("")
ax.legend()
guardar_figura("01_faltantes_top15.png")
plt.show()
mostrar_interpretacion(
    "Objetivo y control de fuga",
    [
        f"Se excluyeron <b>{filas_sin_target:,}</b> registros sin medición válida de MP2.5 o sin etiqueta; conservarlos habría generado negativos artificiales.",
        f"La matriz modelable queda en <b>{len(df):,}</b> registros, consistente con el filtro inferencial de S1/S2. La prevalencia positiva es <b>{100 * df[TARGET].mean():.2f}%</b>.",
        "La lista de exclusión se audita antes de seleccionar modelos. El sistema falla de forma explícita si no existen suficientes predictores aprobados.",
    ],
    "alerta" if df[TARGET].mean() < 0.20 else "info",
)
```

```python
# =============================================================================
# 4.1 ALIAS, VARIABLES DERIVADAS SEGURAS Y POOL DE CANDIDATAS
# =============================================================================
def primera_columna(columnas, candidatas):
    return next((c for c in candidatas if c in columnas), None)
ALIASES = {
    "temperatura": ["temperatura_mean", "temperatura_media", "temp_mean", "temp_media"],
    "humedad": ["humedad_mean", "humedad_media", "rh_mean"],
    "viento": ["viento_mean", "viento_media", "velocidad_viento_mean"],
    "radiacion": ["radiacion_mean", "radiacion_media", "radiacion_solar_mean"],
    "presion": ["presion_mean", "presion_media", "pressure_mean"],
    "inversion": ["prop_inversion", "inversion_termica", "alta_inversion", "inversion"],
    "mp10": ["mp10_mean", "mp10_media", "MP10"],
    "fin_semana": ["es_finde", "fin_semana", "es_fin_semana"],
    "festivo": ["es_festivo", "festivo"],
    "gec": ["periodo_gec", "periodo_GEC", "gec"],
    "cobertura": ["cobertura_horaria", "cobertura_mp25", "horas_validas_mp25", "n_horas_mp25", "horas_validas"],
}
columnas_disponibles = set(df.columns)
mapa = {k: primera_columna(columnas_disponibles, v) for k, v in ALIASES.items()}
# ---------------------------------------------------------------------------
# Cobertura horaria aplicada en S3.
# Regla docente: cobertura < 75% o menos de 18 horas válidas de MP2.5.
#
# Ruta A: utilizar una columna diaria real de cobertura/horas válidas, si existe.
# Ruta B: reconstruirla desde una fuente horaria real por estación-fecha.
# La ejecución se detiene si ninguna ruta permite calcular el indicador por fila.
# ---------------------------------------------------------------------------
def normalizar_nombre_columna(valor):
    texto = unicodedata.normalize("NFKD", str(valor))
    texto = texto.encode("ascii", "ignore").decode("ascii").lower()
    return re.sub(r"[^a-z0-9]+", "_", texto).strip("_")
def resolver_columna(frame, aliases=(), contiene=(), excluye=()):
    normalizadas = {c: normalizar_nombre_columna(c) for c in frame.columns}
    aliases_norm = {normalizar_nombre_columna(a) for a in aliases}
    excluye_norm = tuple(normalizar_nombre_columna(x) for x in excluye)
    for columna, nombre in normalizadas.items():
        if nombre in aliases_norm:
            return columna
    for columna, nombre in normalizadas.items():
        if contiene and all(
            normalizar_nombre_columna(x) in nombre
            for x in contiene
        ):
            if not any(x in nombre for x in excluye_norm):
                return columna
    return None
def convertir_cobertura_diaria(frame, columna):
    """
    Estandariza una columna diaria real como horas válidas y proporción.

    Reconoce:
    - horas válidas en escala 0–24;
    - proporciones en escala 0–1;
    - porcentajes en escala 0–100.
    """
    valores = pd.to_numeric(frame[columna], errors="coerce")
    if valores.isna().any():
        raise ValueError(
            f"La columna diaria de cobertura '{columna}' contiene valores "
            "faltantes o no numéricos. Debe completarse desde la fuente horaria."
        )
    nombre = normalizar_nombre_columna(columna)
    maximo = float(valores.max())
    minimo = float(valores.min())
    if minimo < 0:
        raise ValueError("La cobertura horaria no puede contener valores negativos.")
    es_horas = (
        any(token in nombre for token in ("hora", "horas", "n_horas"))
        and maximo <= 24.5
    )
    if es_horas:
        horas = valores.round().clip(0, 24)
        proporcion = horas / 24.0
        escala = "horas válidas 0–24"
    elif maximo <= 1.000001:
        proporcion = valores.clip(0, 1)
        horas = np.rint(proporcion * 24)
        escala = "proporción 0–1"
    elif maximo <= 100.000001:
        proporcion = (valores / 100.0).clip(0, 1)
        horas = np.rint(proporcion * 24)
        escala = "porcentaje 0–100"
    else:
        raise ValueError(
            f"La columna '{columna}' no presenta una escala válida de cobertura."
        )

    salida = pd.DataFrame(index=frame.index)
    salida["horas_validas_mp25"] = horas.astype(int)
    salida["cobertura_horaria_prop"] = proporcion.astype(float)
    salida["baja_cobertura_horaria"] = (
        (salida["cobertura_horaria_prop"] < COBERTURA_MINIMA)
        | (salida["horas_validas_mp25"] < HORAS_VALIDAS_MINIMAS)
    ).astype(int)
    return salida, escala
def leer_csv_muestra(ruta, nrows=250):
    errores = []
    for codificacion in ("utf-8-sig", "utf-8", "latin-1"):
        for separador in (None, ";", ","):
            try:
                if separador is None:
                    frame = pd.read_csv(
                        ruta,
                        encoding=codificacion,
                        sep=None,
                        engine="python",
                        nrows=nrows,
                    )
                else:
                    frame = pd.read_csv(
                        ruta,
                        encoding=codificacion,
                        sep=separador,
                        nrows=nrows,
                    )
                if frame.shape[1] > 1:
                    return frame
            except Exception as exc:
                errores.append(str(exc))
    raise RuntimeError(
        f"No fue posible leer una muestra de {ruta.name}: "
        + " | ".join(errores[-4:])
    )
def detectar_columnas_horarias(frame):
    col_fecha_hora = resolver_columna(
        frame,
        aliases=(
            "fecha_hora", "fechahora", "datetime", "date_time",
            "timestamp", "fecha y hora",
        ),
    )
    col_fecha = resolver_columna(
        frame,
        aliases=("fecha", "date", "dia", "fecha_medicion"),
    )
    col_hora = resolver_columna(
        frame,
        aliases=("hora", "hour", "hora_medicion", "hora_local"),
    )
    col_estacion = resolver_columna(
        frame,
        aliases=(
            "estacion", "station", "codigo_estacion", "nombre_estacion",
            "estacion_nombre", "id_estacion",
        ),
    )
    col_mp25 = None
    for columna in frame.columns:
        nombre = normalizar_nombre_columna(columna)
        es_mp25 = any(
            token in nombre
            for token in ("mp2_5", "mp25", "pm2_5", "pm25")
        )
        es_derivada = any(
            token in nombre
            for token in (
                "mean", "media", "promedio", "categoria", "calidad",
                "critico", "mala", "max", "min",
            )
        )
        if es_mp25 and not es_derivada:
            col_mp25 = columna
            break

    return {
        "fecha_hora": col_fecha_hora,
        "fecha": col_fecha,
        "hora": col_hora,
        "estacion": col_estacion,
        "mp25": col_mp25,
    }
def parsear_fecha_flexible(serie):
    texto = serie.astype(str).str.strip()
    proporcion_anio_primero = texto.str.match(
        r"^\d{4}[-/]",
        na=False,
    ).mean()
    return pd.to_datetime(
        serie,
        errors="coerce",
        dayfirst=proporcion_anio_primero < 0.80,
    )
def construir_datetime_horario(frame, columnas):
    if columnas["fecha_hora"] is not None:
        return parsear_fecha_flexible(frame[columnas["fecha_hora"]])

    if columnas["fecha"] is None:
        return pd.Series(
            pd.NaT,
            index=frame.index,
            dtype="datetime64[ns]",
        )

    fecha = parsear_fecha_flexible(frame[columnas["fecha"]])
    if columnas["hora"] is None:
        return fecha

    hora_raw = frame[columnas["hora"]]
    hora_num = pd.to_numeric(hora_raw, errors="coerce")
    if hora_num.notna().mean() >= 0.80:
        hora_delta = pd.to_timedelta(
            hora_num,
            unit="h",
            errors="coerce",
        )
    else:
        hora_txt = hora_raw.astype(str).str.strip()
        hora_txt = hora_txt.where(
            hora_txt.str.contains(":", regex=False),
            hora_txt + ":00",
        )
        hora_delta = pd.to_timedelta(hora_txt, errors="coerce")
    return fecha.dt.normalize() + hora_delta
def descubrir_archivos_horarios():
    candidatos_explicitos = [
        DATA_DIR / "calidad_aire_horaria.csv",
        DATA_DIR / "calidad_aire_horario.csv",
        DATA_DIR / "sinca_horario.csv",
        DATA_DIR / "datos_horarios.csv",
        PROJECT_ROOT / "calidad_aire_horaria.csv",
    ]
    candidatos = [p for p in candidatos_explicitos if p.exists()]
    raices = [
        DATA_DIR,
        PROJECT_ROOT / "data" / "raw",
        PROJECT_ROOT / "data" / "processed",
        CWD / "data",
        CWD.parent / "data",
    ]
    for raiz in raices:
        if not raiz.exists():
            continue
        for ruta in raiz.rglob("*.csv"):
            nombre = normalizar_nombre_columna(ruta.name)
            if ruta.resolve() == ruta_dataset.resolve(): # type: ignore
                continue
            if any(
                x in nombre
                for x in ("resultado", "s2_", "s3_", "diario", "daily")
            ):
                continue
            if any(
                x in nombre
                for x in ("horar", "sinca", "mp25", "calidad_aire")
            ):
                candidatos.append(ruta)

    salida = []
    vistos = set()
    for ruta in candidatos:
        try:
            clave = ruta.resolve()
        except OSError:
            clave = ruta
        if clave not in vistos:
            vistos.add(clave)
            salida.append(ruta)
    return salida
def estandarizar_fuente_horaria(ruta):
    muestra = leer_csv_muestra(ruta)
    columnas = detectar_columnas_horarias(muestra)
    if columnas["mp25"] is None:
        return None, "sin columna horaria MP2.5"

    if columnas["fecha_hora"] is None and columnas["fecha"] is None:
        return None, "sin fecha o fecha-hora"

    col_estacion_diaria = resolver_columna(
        df,
        aliases=(
            "estacion", "station", "codigo_estacion", "nombre_estacion",
            "estacion_nombre", "id_estacion",
        ),
    )
    if col_estacion_diaria is not None and columnas["estacion"] is None:
        return None, "sin estación para emparejar la matriz diaria"
    frame = leer_csv_robusto(ruta)
    columnas = detectar_columnas_horarias(frame)
    fecha_hora = construir_datetime_horario(frame, columnas)
    mp25 = pd.to_numeric(frame[columnas["mp25"]], errors="coerce")

    estandar = pd.DataFrame({
        "fecha_hora": fecha_hora,
        "mp25_horario": mp25,
    })
    if columnas["estacion"] is not None:
        estandar["estacion_clave"] = (
            frame[columnas["estacion"]]
            .astype(str)
            .str.strip()
            .str.casefold()
        )
    estandar = estandar.loc[
        estandar["fecha_hora"].notna()
    ].copy()
    if len(estandar) < 100:
        return None, "menos de 100 registros horarios válidos"

    estandar["fecha_cobertura"] = (
        estandar["fecha_hora"].dt.normalize()
    )
    estandar["hora_cobertura"] = (
        estandar["fecha_hora"].dt.floor("h")
    )

    claves = ["fecha_cobertura"]
    if "estacion_clave" in estandar.columns:
        claves.append("estacion_clave")
    horas_distintas = estandar["fecha_hora"].dt.hour.nunique()
    if horas_distintas < 4:
        return None, "no presenta granularidad horaria verificable"
    return estandar, "fuente horaria válida"
def reconstruir_cobertura_horaria():
    fuentes = []
    auditoria = []
    for ruta in descubrir_archivos_horarios():
        try:
            estandar, estado = estandarizar_fuente_horaria(ruta)
        except Exception as exc:
            estandar, estado = None, f"error: {exc}"

        if estandar is None:
            periodo = "—"
            estado_auditoria = f"Descartada · {estado}"
            registros_utilizados = 0
        else:
            fecha_min = estandar["fecha_hora"].min()
            fecha_max = estandar["fecha_hora"].max()
            periodo = (
                f"{fecha_min:%Y-%m-%d} → {fecha_max:%Y-%m-%d}"
                if pd.notna(fecha_min) and pd.notna(fecha_max)
                else "—"
            )
            estado_auditoria = "Válida e incorporada"
            registros_utilizados = len(estandar)
        auditoria.append({
            # El nombre breve se utiliza en la tabla del notebook.
            "archivo": ruta.name,
            # La ruta completa queda en el CSV de auditoría, sin sobrecargar
            # la visualización HTML.
            "ruta_completa": str(ruta.resolve()),
            "estado": estado_auditoria,
            "periodo": periodo,
            "registros_utilizados": registros_utilizados,
        })
        if estandar is not None:
            estandar["archivo_fuente"] = ruta.name
            fuentes.append(estandar)

    auditoria = pd.DataFrame(auditoria)
    auditoria.to_csv(
        CSV_DIR / "s3_auditoria_fuentes_cobertura.csv",
        index=False,
        encoding="utf-8-sig",
    )
    if not fuentes:
        raise FileNotFoundError(
            "No se encontró una fuente horaria válida para calcular la "
            "cobertura de MP2.5. Agregue un CSV horario en data/ con "
            "fecha-hora, estación y medición MP2.5. El notebook no puede "
            "declarar esta observación como cumplida sin evidencia por fila."
        )
    union = pd.concat(fuentes, ignore_index=True, sort=False)

    # Distintas fuentes pueden cubrir periodos solapados. El cálculo posterior
    # elimina duplicados por estación, fecha y hora, evitando contabilizar dos
    # veces una misma medición horaria.
    claves = ["fecha_cobertura"]
    if "estacion_clave" in union.columns:
        claves.append("estacion_clave")
    universo = union[claves].drop_duplicates()
    validas = (
        union.loc[union["mp25_horario"].notna()]
        .drop_duplicates(claves + ["hora_cobertura"])
        .groupby(claves, observed=True)
        .size()
        .rename("horas_validas_mp25")
        .reset_index()
    )
    cobertura = universo.merge(validas, on=claves, how="left")
    cobertura["horas_validas_mp25"] = (
        cobertura["horas_validas_mp25"]
        .fillna(0)
        .astype(int)
        .clip(0, 24)
    )
    cobertura["cobertura_horaria_prop"] = (
        cobertura["horas_validas_mp25"] / 24.0
    )
    cobertura["baja_cobertura_horaria"] = (
        (cobertura["cobertura_horaria_prop"] < COBERTURA_MINIMA)
        | (
            cobertura["horas_validas_mp25"]
            < HORAS_VALIDAS_MINIMAS
        )
    ).astype(int)
    return cobertura, auditoria
columna_cobertura_diaria = mapa.get("cobertura")
if columna_cobertura_diaria is not None:
    cobertura_estandar, escala_cobertura = convertir_cobertura_diaria(
        df,
        columna_cobertura_diaria,
    )
    for columna in cobertura_estandar.columns:
        df[columna] = cobertura_estandar[columna]

    proporcion_emparejada = 1.0
    fuente_cobertura = (
        f"columna diaria real: {columna_cobertura_diaria} "
        f"({escala_cobertura})"
    )
    fecha_min_diaria = df[col_fecha].min()
    fecha_max_diaria = df[col_fecha].max()
    periodo_diario = (
        f"{fecha_min_diaria:%Y-%m-%d} → {fecha_max_diaria:%Y-%m-%d}"
        if pd.notna(fecha_min_diaria) and pd.notna(fecha_max_diaria)
        else "—"
    )
    auditoria_fuentes_cobertura = pd.DataFrame([{
        "archivo": ruta_dataset.name, # type: ignore
        "ruta_completa": str(ruta_dataset.resolve()), # type: ignore
        "estado": "Columna diaria válida",
        "periodo": periodo_diario,
        "registros_utilizados": len(df),
    }])
else:
    cobertura_diaria, auditoria_fuentes_cobertura = (
        reconstruir_cobertura_horaria()
    ) # type: ignore
    col_estacion_diaria = resolver_columna(
        df,
        aliases=(
            "estacion", "station", "codigo_estacion",
            "nombre_estacion", "estacion_nombre", "id_estacion",
        ),
    )
    df["_fecha_cobertura"] = df[col_fecha].dt.normalize()
    claves_izq = ["_fecha_cobertura"]
    claves_der = ["fecha_cobertura"]

    if col_estacion_diaria is not None:
        if "estacion_clave" not in cobertura_diaria.columns:
            raise ValueError(
                "La matriz diaria contiene estación, pero la fuente horaria "
                "no permite reconstruir la cobertura a esa granularidad."
            )
        df["_estacion_cobertura"] = (
            df[col_estacion_diaria]
            .astype(str)
            .str.strip()
            .str.casefold()
        )
        claves_izq.append("_estacion_cobertura")
        claves_der.append("estacion_clave")

    df = df.merge(
        cobertura_diaria,
        left_on=claves_izq,
        right_on=claves_der,
        how="left",
        validate="m:1",
    )
    proporcion_emparejada = float(
        df["horas_validas_mp25"].notna().mean()
    )
    if proporcion_emparejada < COBERTURA_MERGE_MINIMA:
        raise ValueError(
            "La cobertura horaria solo emparejó "
            f"{100 * proporcion_emparejada:.2f}% de la matriz diaria; "
            f"se exige al menos {100 * COBERTURA_MERGE_MINIMA:.0f}%. "
            "Revise nombres de estación y rango de fechas."
        )
    if df["horas_validas_mp25"].isna().any():
        raise ValueError(
            "Quedaron filas sin cobertura horaria después del emparejamiento. "
            "La cobertura debe estar calculada para todas las filas modelables."
        )

    fuente_cobertura = "registros horarios reales de MP2.5"

    columnas_auxiliares_merge = [
        "_fecha_cobertura", "_estacion_cobertura",
        "fecha_cobertura", "estacion_clave", "archivo_fuente",
    ]
    df = df.drop(
        columns=[
            c for c in columnas_auxiliares_merge
            if c in df.columns
        ],
        errors="ignore",
    )
df["horas_validas_mp25"] = pd.to_numeric(
    df["horas_validas_mp25"],
    errors="raise",
).astype(int)
df["cobertura_horaria_prop"] = pd.to_numeric(
    df["cobertura_horaria_prop"],
    errors="raise",
).astype(float)
df["baja_cobertura_horaria"] = pd.to_numeric(
    df["baja_cobertura_horaria"],
    errors="raise",
).astype(int)

if not df["horas_validas_mp25"].between(0, 24).all():
    raise AssertionError("Las horas válidas deben permanecer entre 0 y 24.")
if not df["cobertura_horaria_prop"].between(0, 1).all():
    raise AssertionError("La cobertura debe permanecer entre 0 y 1.")
if not set(df["baja_cobertura_horaria"].unique()).issubset({0, 1}):
    raise AssertionError("El indicador de cobertura debe ser binario.")

variable_cobertura_modelable = "baja_cobertura_horaria"
proporcion_baja_observada = float(
    df[variable_cobertura_modelable].mean()
)
control_cobertura = pd.DataFrame([{
    "fuente": fuente_cobertura,
    "criterio": "cobertura < 75% o horas válidas < 18",
    "filas_modelables": len(df),
    "filas_con_cobertura": int(
        df["horas_validas_mp25"].notna().sum()
    ),
    "emparejamiento_%": 100 * proporcion_emparejada,
    "proporcion_baja_cobertura_observada": (
        proporcion_baja_observada
    ),
    "referencia_s2": COBERTURA_INSUFICIENTE_REFERENCIA_S2,
    "diferencia_pp_vs_s2": (
        100 * (
            proporcion_baja_observada
            - COBERTURA_INSUFICIENTE_REFERENCIA_S2
        )
    ),
    "indicador_s3": variable_cobertura_modelable,
    "estado": "CUMPLIDO · calculado por fila",
    "accion": (
        "indicador incorporado al pool; mediana por mes/temporada "
        "y sensibilidad con casos completos"
    ),
}])
control_cobertura.to_csv(
    CSV_DIR / "s3_control_cobertura_horaria.csv",
    index=False,
    encoding="utf-8-sig",
)
auditoria_fuentes_cobertura.to_csv(
    CSV_DIR / "s3_auditoria_fuentes_cobertura.csv",
    index=False,
    encoding="utf-8-sig",
)
mostrar_tabla(
    control_cobertura,
    "Cobertura horaria calculada y aplicada en S3",
    max_filas=5,
)
# La auditoría completa —incluida la ruta absoluta— se conserva en CSV.
# Para el notebook se muestra una vista compacta, evitando que las rutas largas
# se superpongan con las columnas de estado y registros.
auditoria_fuentes_visual = auditoria_fuentes_cobertura.copy()
columnas_auditoria_visual = [
    "archivo",
    "estado",
    "periodo",
    "registros_utilizados",
]
auditoria_fuentes_visual = auditoria_fuentes_visual.loc[
    :,
    [
        columna for columna in columnas_auditoria_visual
        if columna in auditoria_fuentes_visual.columns
    ],
].copy()

auditoria_fuentes_visual["archivo"] = (
    auditoria_fuentes_visual["archivo"]
    .astype(str)
    .map(lambda valor: Path(valor).name)
)
auditoria_fuentes_visual["registros_utilizados"] = pd.to_numeric(
    auditoria_fuentes_visual["registros_utilizados"],
    errors="coerce",
).fillna(0).astype(int)

mostrar_tabla(
    auditoria_fuentes_visual,
    "Auditoría de la fuente de cobertura horaria",
    max_filas=20,
)
df["mes"] = df[col_fecha].dt.month.astype(int)
df["mes_sin"] = np.sin(2 * np.pi * df["mes"] / 12)
df["mes_cos"] = np.cos(2 * np.pi * df["mes"] / 12)
df["invierno"] = df["mes"].isin([5, 6, 7, 8]).astype(int)
df["anio"] = df[col_fecha].dt.year.astype(int)
# ---------------------------------------------------------------------------
# Trazabilidad estructural entre las tres fases.
# 46 corresponde a la matriz analítica armonizada; las variables de calendario
# creadas exclusivamente para S3 se reportan por separado.
# ---------------------------------------------------------------------------
columnas_nuevas_s3 = [c for c in df.columns if c not in df_raw.columns]

consistencia_fases = pd.DataFrame([
    {
        "fase": "S1",
        "etapa": "Base enriquecida previa al filtro",
        "filas": BASELINE_PROYECTO["filas_base"],
        "variables": BASELINE_PROYECTO["columnas_enriquecidas"],
        "hilo conductor": "36 originales + 10 físicas derivadas",
    },
    {
        "fase": "S1",
        "etapa": "Matriz descriptiva e inferencial",
        "filas": BASELINE_PROYECTO["filas_inferenciales"],
        "variables": BASELINE_PROYECTO["columnas_enriquecidas"],
        "hilo conductor": "Solo MP2.5 diario disponible",
    },
    {
        "fase": "S2",
        "etapa": "Validación y remuestreo",
        "filas": BASELINE_PROYECTO["filas_inferenciales"],
        "variables": BASELINE_PROYECTO["columnas_enriquecidas"],
        "hilo conductor": "Misma matriz inferencial de S1",
    },
    {
        "fase": "S3",
        "etapa": "Matriz armonizada recibida",
        "filas": len(df),
        "variables": df_raw.shape[1],
        "hilo conductor": "Mismo filtro y mismas 46 variables",
    },
    {
        "fase": "S3",
        "etapa": "Matriz de trabajo para modelamiento",
        "filas": len(df),
        "variables": df.shape[1],
        "hilo conductor": (
            f"46 armonizadas + {len(columnas_nuevas_s3)} var. temp. S3"
        ),
    },
])

consistencia_fases["estado"] = [
    "Referencia",
    "Referencia",
    "Referencia",
    (
        "Consistente"
        if len(df) == BASELINE_PROYECTO["filas_inferenciales"]
        and df_raw.shape[1] == BASELINE_PROYECTO["columnas_enriquecidas"]
        else "Revisar"
    ),
    "Extensión documentada",
]
consistencia_fases.to_csv(
    CSV_DIR / "s3_matriz_consistencia_s1_s2_s3.csv",
    index=False,
    encoding="utf-8-sig",
)
mostrar_tabla(
    consistencia_fases,
    "Hilo conductor y consistencia estructural S1 → S2 → S3",
    max_filas=10,
)


for clave in ["fin_semana", "festivo", "gec"]:
    columna = mapa.get(clave)
    if columna is not None:
        df[columna] = normalizar_binaria(df[columna])

categoricas_candidatas = [
    c for c in [
        "zona_geografica", "nivel_socioeconomico", "estacion_anio",
        "temporada_critica", "comuna", "estacion"
    ]
    if c in df.columns and 1 < df[c].nunique(dropna=True) <= 25
]

meteo_validada = [
    mapa[k] for k in ["temperatura", "humedad", "viento", "radiacion", "inversion"]
    if mapa[k] is not None
]
meteo_ampliada = meteo_validada + ([mapa["presion"]] if mapa["presion"] is not None else [])
temporales = ["mes_sin", "mes_cos", "invierno"] + [
    mapa[k] for k in ["fin_semana", "festivo", "gec"] if mapa[k] is not None
]
nowcasting = [
    c for c in [mapa["mp10"], variable_cobertura_modelable]
    if c is not None
]

columnas_bloqueadas = set(auditoria_fuga["variable"].tolist()) | {col_fecha, "anio", "mes"}

def aprobar_variables(lista):
    salida = []
    for c in lista:
        if c is None or c not in df.columns or c in columnas_bloqueadas:
            continue
        if c not in salida:
            salida.append(c)
    return salida
VARIABLES_DOMINIO = aprobar_variables(
    meteo_validada + temporales + categoricas_candidatas
)
VARIABLES_COMPLETAS = aprobar_variables(
    meteo_ampliada + temporales + categoricas_candidatas + nowcasting
)

if len(VARIABLES_DOMINIO) < 3:
    raise ValueError(
        "El modelo de dominio no dispone de al menos tres variables aprobadas. "
        "Revise los alias y la estructura del dataset."
    )
if len(VARIABLES_COMPLETAS) < len(VARIABLES_DOMINIO):
    raise ValueError("El pool completo no puede ser menor que el modelo de dominio.")
matriz_candidatas = []
for c in VARIABLES_COMPLETAS:
    familia = (
        "meteorológica" if c in meteo_ampliada else
        "temporal" if c in temporales else
        "territorial" if c in categoricas_candidatas else
        "nowcasting/calidad" if c in nowcasting else "otra"
    )
    matriz_candidatas.append({
        "variable": c,
        "familia": familia,
        "tipo": "numérica" if pd.api.types.is_numeric_dtype(df[c]) else "categórica",
        "faltantes_%": 100 * df[c].isna().mean(),
        "modelo_dominio": c in VARIABLES_DOMINIO,
        "pool_seleccion": True,
    })
matriz_candidatas = pd.DataFrame(matriz_candidatas)
matriz_candidatas.to_csv(
    CSV_DIR / "s3_pool_variables_aprobadas.csv", index=False, encoding="utf-8-sig"
)
mostrar_tabla(matriz_candidatas, "Lista blanca de variables candidatas", max_filas=40)

mostrar_interpretacion(
    "Variables incorporadas al modelamiento",
    [
        f"El modelo de dominio utiliza <b>{len(VARIABLES_DOMINIO)}</b> variables y el pool de selección contiene <b>{len(VARIABLES_COMPLETAS)}</b>.",
        "Las variables se incorporan por una lista definida y no por posición.",
        "MP10 se utiliza solo para <i>nowcasting</i>. El indicador de baja cobertura fue reconstruido desde registros horarios reales y aplicado por estación-fecha.",
    ],
    "ok",
)
```

    o decisión importada desde S2 y el uso concreto definido para S3. Cuando el reporte puente no contiene una validación específica para una variable, se declara explícitamente como evidencia contextual y no se presenta como resultado confirmado en S2.</p>

```python
# =============================================================================
# 4.2 MATRIZ EXPLÍCITA DE EVIDENCIA S1 → S2 → S3 POR PREDICTOR
# =============================================================================
def buscar_evidencia_s2(*patrones):
    """Recupera la primera evidencia compatible desde el reporte puente S2."""
    elementos = reporte_s2["elemento"].astype(str)
    for patron in patrones:
        mascara = elementos.str.contains(
            patron,
            case=False,
            regex=False,
            na=False,
        )
        if mascara.any():
            fila = reporte_s2.loc[mascara].iloc[0]
            return {
                "evidencia_s2": str(fila["resultado_validado"]),
                "estado_s2": str(fila["estado"]),
                "uso_recomendado_s2": str(fila["uso_s3"]),
            }
    return {
        "evidencia_s2": (
            "Sin validación específica en el reporte puente; "
            "se conserva como evidencia contextual de S1."
        ),
        "estado_s2": "contextual / no específica",
        "uso_recomendado_s2": (
            "No presentarlo como validación específica de S2."
        ),
    }
def descripcion_s1(variable):
    """Síntesis de los hallazgos documentados en la fase descriptiva S1."""
    reglas = {
        "mp10_mean": (
            "Contaminante complementario con asociación alta respecto de MP2.5."
        ),
        "humedad_mean": (
            "Variable meteorológica vinculada a condiciones de acumulación "
            "y persistencia del material particulado."
        ),
        "temperatura_mean": (
            "Señal meteorológica y estacional inversamente relacionada con MP2.5."
        ),
        "viento_mean": (
            "Variable física de dispersión atmosférica incorporada al análisis "
            "meteorológico."
        ),
        "prop_inversion": (
            "Los días con mayor inversión térmica presentaron mayor MP2.5; "
            "el contraste alta versus baja inversión rechazó H0."
        ),
        "radiacion_mean": (
            "Variable atmosférica con patrón estacional y análisis conjunto "
            "con temperatura."
        ),
        "radiacion_max": (
            "Extremo diario de radiación evaluado junto con la versión media."
        ),
        "zona_geografica": (
            "S1 detectó diferencias territoriales relevantes entre Oriente "
            "y Poniente."
        ),
        "nivel_socioeconomico": (
            "S1 documentó inequidad ambiental entre Q1 y Q5."
        ),
        "invierno": (
            "S1 identificó mayor contaminación durante el periodo frío."
        ),
        "temporada_critica": (
            "S1 separó Invierno del resto del año por su mayor carga de MP2.5."
        ),
        "estacion_anio": (
            "S1 mostró un patrón estacional claro de MP2.5."
        ),
        "mes_sin": (
            "Codificación cíclica derivada del patrón mensual observado en S1."
        ),
        "mes_cos": (
            "Codificación cíclica derivada del patrón mensual observado en S1."
        ),
        "presion_mean": (
            "Variable atmosférica de control incorporada en la caracterización "
            "meteorológica."
        ),
        "comuna": (
            "Unidad territorial utilizada para caracterizar heterogeneidad "
            "espacial y cobertura."
        ),
        "estacion": (
            "Unidad instrumental y espacial del dataset estación-día."
        ),
        "zona": (
            "Variable territorial derivada de la caracterización espacial."
        ),
        "periodo_gec": (
            "Indicador temporal-operacional utilizado para representar el "
            "periodo de gestión de episodios críticos."
        ),
        "es_finde": (
            "Indicador de calendario para capturar diferencias de actividad "
            "entre semana y fin de semana."
        ),
        "es_festivo": (
            "Indicador de calendario para capturar cambios de actividad en "
            "días festivos."
        ),
    }
    if variable in reglas:
        return reglas[variable]
    if "cobertura" in variable or "horas_validas" in variable:
        return (
            "S1 identificó ausencia de mediciones y necesidad de controlar "
            "la cobertura instrumental."
        )
    return (
        "Variable candidata derivada de la caracterización de S1 y del "
        "fundamento físico o territorial del problema."
    )
def evidencia_s2_variable(variable):
    """Asocia cada predictor con una evidencia directa o contextual de S2."""
    if variable == "mp10_mean":
        return buscar_evidencia_s2("mp2_5_mean vs mp10_mean")
    if variable == "humedad_mean":
        return buscar_evidencia_s2("mp2_5_mean vs humedad_mean")
    if variable == "temperatura_mean":
        return buscar_evidencia_s2("mp2_5_mean vs temperatura_mean")
    if variable == "viento_mean":
        return buscar_evidencia_s2("mp2_5_mean vs viento_mean")
    if variable == "prop_inversion":
        return buscar_evidencia_s2("mp2_5_mean vs prop_inversion")
    if variable in {"invierno", "temporada_critica", "estacion_anio", "mes_sin", "mes_cos"}:
        return buscar_evidencia_s2("entre Invierno y Resto del año")
    if variable in {"radiacion_mean", "radiacion_max"}:
        return buscar_evidencia_s2("Temperatura y radiación")
    if variable == "zona_geografica":
        return buscar_evidencia_s2("Oriente vs Poniente")
    if "cobertura" in variable or "horas_validas" in variable:
        return buscar_evidencia_s2("Cobertura horaria insuficiente")
    return buscar_evidencia_s2("__sin_coincidencia_directa__")
def uso_s3_variable(variable):
    """Define el papel metodológico real del predictor dentro de S3."""
    if variable == "mp10_mean":
        return (
            "Candidata de nowcasting; se controla su alta correlación con MP2.5 "
            "y nunca se incorpora MP2.5 como predictor."
        )
    if variable in {"temperatura_mean", "humedad_mean", "viento_mean", "prop_inversion",
                    "radiacion_mean", "radiacion_max", "presion_mean"}:
        return "Candidata meteorológica para M1 y para el pool M2/M3."
    if variable in {"invierno", "temporada_critica", "estacion_anio", "mes_sin",
                    "mes_cos", "periodo_gec", "es_finde", "es_festivo"}:
        return "Candidata temporal para representar estacionalidad o calendario."
    if variable in {"zona_geografica", "nivel_socioeconomico", "comuna", "estacion"}:
        return "Candidata territorial/contextual y diagnóstico de subgrupos."
    if "cobertura" in variable or "horas_validas" in variable:
        return "Indicador de calidad de datos y análisis de sensibilidad."
    return "Candidata del pool controlado de S3."
variables_trazadas = []
for variable in list(VARIABLES_DOMINIO) + list(nowcasting):
    if variable not in variables_trazadas:
        variables_trazadas.append(variable)

filas_evidencia = []
for variable in variables_trazadas:
    evidencia = evidencia_s2_variable(variable)
    filas_evidencia.append({
        "variable": variable,
        "evidencia_s1": descripcion_s1(variable),
        "evidencia_s2": evidencia["evidencia_s2"],
        "estado_s2": evidencia["estado_s2"],
        "uso_s3": uso_s3_variable(variable),
        "modelo_o_pool": (
            "M1_Dominio_S1_S2"
            if variable in VARIABLES_DOMINIO
            else "Pool M2/M3 · nowcasting"
        ),
    })
matriz_evidencia_s1_s2_s3 = pd.DataFrame(filas_evidencia)
matriz_evidencia_s1_s2_s3.to_csv(
    CSV_DIR / "s3_matriz_evidencia_s1_s2_por_predictor.csv",
    index=False,
    encoding="utf-8-sig",
)

mostrar_tabla(
    matriz_evidencia_s1_s2_s3,
    "Matriz de evidencia S1 → S2 → uso en S3",
    max_filas=40,
)

n_directas = int(
    (
        ~matriz_evidencia_s1_s2_s3["estado_s2"]
        .eq("contextual / no específica")
    ).sum()
)
n_contextuales = len(matriz_evidencia_s1_s2_s3) - n_directas

mostrar_interpretacion(
    "Continuidad por predictor",
    [
        f"La matriz documenta <b>{len(matriz_evidencia_s1_s2_s3)}</b> "
        "predictores prioritarios o de dominio.",
        f"<b>{n_directas}</b> cuentan con una evidencia o decisión específica "
        f"en el reporte puente S2 y <b>{n_contextuales}</b> se conservan "
        "transparentemente como variables contextuales sustentadas en S1.",
        "Las variables contextuales se identifican por separado para evitar atribuirles una validación que no existe.",
    ],
    "ok",
)
```

    R², residuos, normalidad, heterocedasticidad y encogimiento de varianza.
    Las variables auxiliares de mes y temporada entran únicamente al imputador y se eliminan
    antes del modelo, de modo que la estrategia agrupada se aplica incluso cuando esos
    campos no pertenecen al subconjunto final de predictores.</p>

```python
# =============================================================================
# 6.1 PREPROCESADORES Y DIAGNÓSTICO DEL MECANISMO DE AUSENCIA
# =============================================================================

class ImputadorMedianaMesTemporada(BaseEstimator, TransformerMixin):
    """
    Imputa variables numéricas con esta jerarquía:
    1) mediana del mismo mes;
    2) mediana de la misma temporada (invierno/resto);
    3) mediana global aprendida solo en entrenamiento.

    ``mes_sin``, ``mes_cos`` e ``invierno`` se reciben como contexto auxiliar,
    pero no se devuelven si no forman parte de ``columnas_salida``. Así la
    imputación agrupada se aplica sin forzar esos campos dentro del modelo.
    """

    def __init__(
        self,
        columnas_salida=None,
        col_mes_sin="mes_sin",
        col_mes_cos="mes_cos",
        col_temporada="invierno",
    ):
        self.columnas_salida = columnas_salida
        self.col_mes_sin = col_mes_sin
        self.col_mes_cos = col_mes_cos
        self.col_temporada = col_temporada

    def _a_frame(self, X):
        if isinstance(X, pd.DataFrame):
            return X.copy()
        columnas = getattr(
            self,
            "feature_names_in_",
            np.asarray([f"x{i}" for i in range(np.asarray(X).shape[1])]),
        )
        return pd.DataFrame(X, columns=columnas)

    def _clave_mes(self, frame):
        if self.col_mes_sin not in frame or self.col_mes_cos not in frame:
            return pd.Series(pd.NA, index=frame.index, dtype="Int64")
        seno = pd.to_numeric(frame[self.col_mes_sin], errors="coerce")
        coseno = pd.to_numeric(frame[self.col_mes_cos], errors="coerce")
        angulo = np.mod(np.arctan2(seno, coseno), 2 * np.pi)
        mes = pd.Series(
            np.rint(angulo * 12 / (2 * np.pi)),
            index=frame.index,
        ).astype("Int64")
        return mes.mask(mes.eq(0), 12)

    def fit(self, X, y=None):
        frame = self._a_frame(X).apply(pd.to_numeric, errors="coerce")
        self.feature_names_in_ = np.asarray(frame.columns, dtype=object)
        self.columnas_entrada_ = list(frame.columns)

        salidas = (
            list(self.columnas_salida)
            if self.columnas_salida is not None
            else [
                c for c in self.columnas_entrada_
                if c not in {
                    self.col_mes_sin,
                    self.col_mes_cos,
                    self.col_temporada,
                }
            ]
        )
        self.columnas_salida_ = [c for c in salidas if c in frame.columns]
        if not self.columnas_salida_:
            raise ValueError("El imputador no recibió columnas numéricas de salida.")

        self.mediana_global_ = (
            frame[self.columnas_salida_].median().to_dict()
        )
        mes = self._clave_mes(frame)
        temporada = (
            pd.to_numeric(frame[self.col_temporada], errors="coerce")
            if self.col_temporada in frame
            else pd.Series(np.nan, index=frame.index)
        )

        self.mediana_mes_ = {}
        self.mediana_temporada_ = {}
        for columna in self.columnas_salida_:
            temporal = pd.DataFrame({
                "valor": frame[columna],
                "mes": mes,
                "temporada": temporada,
            })
            self.mediana_mes_[columna] = (
                temporal.dropna(subset=["mes"])
                .groupby("mes", observed=True)["valor"]
                .median()
                .to_dict()
            )
            self.mediana_temporada_[columna] = (
                temporal.dropna(subset=["temporada"])
                .groupby("temporada", observed=True)["valor"]
                .median()
                .to_dict()
            )
        return self

    def transform(self, X):
        frame = self._a_frame(X).apply(pd.to_numeric, errors="coerce")
        frame = frame.reindex(columns=self.columnas_entrada_)
        mes = self._clave_mes(frame)
        temporada = (
            pd.to_numeric(frame[self.col_temporada], errors="coerce")
            if self.col_temporada in frame
            else pd.Series(np.nan, index=frame.index)
        )

        for columna in self.columnas_salida_:
            faltan = frame[columna].isna()
            if faltan.any():
                por_mes = mes.map(self.mediana_mes_.get(columna, {}))
                frame.loc[faltan, columna] = por_mes.loc[faltan]

            faltan = frame[columna].isna()
            if faltan.any():
                por_temporada = temporada.map(
                    self.mediana_temporada_.get(columna, {})
                )
                frame.loc[faltan, columna] = por_temporada.loc[faltan]

            faltan = frame[columna].isna()
            if faltan.any():
                frame.loc[faltan, columna] = self.mediana_global_.get(
                    columna,
                    np.nan,
                )

        salida = frame[self.columnas_salida_].copy()
        if salida.isna().any().any():
            columnas = salida.columns[salida.isna().any()].tolist()
            raise ValueError(
                "La imputación por mes/temporada dejó faltantes en: "
                + ", ".join(columnas)
            )
        return salida

    def get_feature_names_out(self, input_features=None):
        return np.asarray(self.columnas_salida_, dtype=object)

def crear_ohe():
    try:
        return OneHotEncoder(handle_unknown="ignore", drop="first", sparse_output=False)
    except TypeError:
        return OneHotEncoder(handle_unknown="ignore", drop="first", sparse=False)

def separar_tipos(frame, features):
    categoricas = [c for c in features if not pd.api.types.is_numeric_dtype(frame[c])]
    numericas = [c for c in features if c not in categoricas]
    return numericas, categoricas

def construir_preprocesador(frame, features, estrategia):
    numericas, categoricas = separar_tipos(frame, features)
    transformadores = []

    if numericas:
        columnas_num_transformador = list(numericas)
        if estrategia == "mediana":
            auxiliares = [
                c for c in ["mes_sin", "mes_cos", "invierno"]
                if c in frame.columns and c not in columnas_num_transformador
            ]
            columnas_num_transformador = columnas_num_transformador + auxiliares
            imp_num = ImputadorMedianaMesTemporada(
                columnas_salida=numericas,
            )
        elif estrategia == "regresion":
            imp_num = IterativeImputer(
                estimator=BayesianRidge(),
                max_iter=50,
                tol=1e-3,
                random_state=SEED,
                sample_posterior=False,
                initial_strategy="median",
                skip_complete=True,
            )
        elif estrategia == "completos":
            imp_num = "passthrough"
        else:
            raise ValueError(f"Estrategia desconocida: {estrategia}")

        pasos_num = []
        if imp_num != "passthrough":
            pasos_num.append(("imputer", imp_num))
        pasos_num.append(("scaler", StandardScaler()))
        transformadores.append(
            ("num", Pipeline(pasos_num), columnas_num_transformador)
        )

    if categoricas:
        pasos_cat = []
        if estrategia != "completos":
            pasos_cat.append(("imputer", SimpleImputer(strategy="most_frequent")))
        pasos_cat.append(("ohe", crear_ohe()))
        transformadores.append(("cat", Pipeline(pasos_cat), categoricas))

    return (
        ColumnTransformer(
            transformadores,
            remainder="drop",
            verbose_feature_names_out=False,
        ),
        numericas,
        categoricas,
    )
# Prueba estructural: confirma que el imputador utiliza el mes/temporada aunque
# esas variables no se devuelvan como predictores.
_prueba_train = pd.DataFrame({
    "x": [1.0, 3.0, 10.0, 14.0, 100.0, 120.0],
    "mes_sin": np.sin(2 * np.pi * np.array([1, 1, 2, 2, 7, 7]) / 12),
    "mes_cos": np.cos(2 * np.pi * np.array([1, 1, 2, 2, 7, 7]) / 12),
    "invierno": [0, 0, 0, 0, 1, 1],
})
_prueba_eval = pd.DataFrame({
    "x": [np.nan, np.nan],
    "mes_sin": np.sin(2 * np.pi * np.array([1, 7]) / 12),
    "mes_cos": np.cos(2 * np.pi * np.array([1, 7]) / 12),
    "invierno": [0, 1],
})
_prueba_imp = ImputadorMedianaMesTemporada(columnas_salida=["x"])
_prueba_resultado = _prueba_imp.fit(_prueba_train).transform(_prueba_eval)
_prueba_esperado = np.array([2.0, 110.0])
if not np.allclose(_prueba_resultado["x"].to_numpy(), _prueba_esperado):
    raise AssertionError(
        "La mediana agrupada no utilizó correctamente mes y temporada."
    )

validacion_imputador_agrupado = pd.DataFrame([{
    "prueba": "contexto auxiliar fuera del modelo",
    "resultado_mes": float(_prueba_resultado.iloc[0, 0]), # type: ignore
    "esperado_mes": float(_prueba_esperado[0]),
    "resultado_temporada": float(_prueba_resultado.iloc[1, 0]), # type: ignore
    "esperado_temporada": float(_prueba_esperado[1]),
    "columnas_salida": ", ".join(_prueba_imp.get_feature_names_out()),
    "estado": "CUMPLIDO",
}])
validacion_imputador_agrupado.to_csv(
    CSV_DIR / "s3_validacion_imputador_mediana_agrupada.csv",
    index=False,
    encoding="utf-8-sig",
)
mostrar_tabla(
    validacion_imputador_agrupado,
    "Validación funcional de la mediana por mes/temporada",
)
def columnas_entrada_preprocesador(frame, features, estrategia):
    columnas = list(features)
    if estrategia == "mediana":
        for auxiliar in ["mes_sin", "mes_cos", "invierno"]:
            if auxiliar in frame.columns and auxiliar not in columnas:
                columnas.append(auxiliar)
    return columnas
def seleccionar_X_preprocesador(frame, features, estrategia):
    return frame[
        columnas_entrada_preprocesador(
            frame,
            features,
            estrategia,
        )
    ].copy()
def filtrar_completos(frame, features):
    return frame.dropna(subset=list(features) + [TARGET]).copy()

def pipeline_predictivo(frame, features, estrategia):
    prep, nums, cats = construir_preprocesador(frame, features, estrategia)
    clf = LogisticRegression(
        penalty="l2",
        C=1e6,
        solver="lbfgs",
        max_iter=5000,
        class_weight="balanced",
        random_state=SEED,
    )
    return Pipeline([("prep", prep), ("clf", clf)]), nums, cats

variables_con_faltantes = [
    c for c in VARIABLES_COMPLETAS
    if 0 < dev_raw[c].isna().mean() < 0.95
]
diagnostico_mecanismo = []

for variable in variables_con_faltantes:
    predictores = [
        c for c in VARIABLES_COMPLETAS
        if c != variable and dev_raw[c].isna().mean() < 0.50
    ][:12]
    if len(predictores) < 2:
        continue

    y_miss_dev = dev_raw[variable].isna().astype(int)
    y_miss_val = val_raw[variable].isna().astype(int)
    if y_miss_dev.nunique() < 2 or y_miss_val.nunique() < 2:
        continue

    prep_miss, _, _ = construir_preprocesador(dev_raw, predictores, "mediana")
    modelo_miss = Pipeline([
        ("prep", prep_miss),
        ("clf", LogisticRegression(max_iter=3000, class_weight="balanced", random_state=SEED)),
    ])
    modelo_miss.fit(
        seleccionar_X_preprocesador(dev_raw, predictores, "mediana"),
        y_miss_dev,
    )
    prob_miss = modelo_miss.predict_proba(
        seleccionar_X_preprocesador(val_raw, predictores, "mediana")
    )[:, 1]
    auc_miss = roc_auc_score(y_miss_val, prob_miss)

    if auc_miss < 0.55:
        lectura = "Sin evidencia predictiva relevante; MCAR sigue siendo plausible"
    elif auc_miss < 0.70:
        lectura = "Ausencia parcialmente explicada por observables; MAR es plausible"
    else:
        lectura = "Ausencia claramente asociada a observables; MAR es más defendible"

    diagnostico_mecanismo.append({
        "variable": variable,
        "faltantes_dev_%": 100 * dev_raw[variable].isna().mean(),
        "faltantes_val_%": 100 * val_raw[variable].isna().mean(),
        "auc_prediccion_ausencia": auc_miss,
        "lectura": lectura,
        "advertencia": "MNAR no puede descartarse",
    })

diagnostico_mecanismo = pd.DataFrame(diagnostico_mecanismo)
diagnostico_mecanismo.to_csv(
    CSV_DIR / "s3_diagnostico_mecanismo_faltantes.csv",
    index=False,
    encoding="utf-8-sig",
)

if diagnostico_mecanismo.empty:
    mostrar_interpretacion(
        "Mecanismo de ausencia",
        "No hubo variables con suficientes faltantes en desarrollo y validación para estimar un clasificador estable de ausencia.",
        "alerta",
    )
else:
    mostrar_tabla(
        diagnostico_mecanismo.sort_values("auc_prediccion_ausencia", ascending=False),
        "Evidencia empírica sobre el mecanismo de faltantes",
        max_filas=30,
    )
    auc_max = diagnostico_mecanismo["auc_prediccion_ausencia"].max()
    mostrar_interpretacion(
        "Lectura del mecanismo de ausencia",
        [
            f"El mayor AUC para explicar la ausencia fue <b>{auc_max:.3f}</b>.",
            "AUC elevado favorece una lectura MAR porque la ausencia depende de variables observadas; no constituye una prueba formal del mecanismo.",
            "MNAR permanece como limitación no verificable con los datos observados y se declara explícitamente.",
        ],
        "info",
    )
```

```python
# =============================================================================
# 6.2 REGRESIONES OLS EXPLÍCITAS PARA DOCUMENTAR LA IMPUTACIÓN
# =============================================================================
numericas_pool = [
    c for c in VARIABLES_COMPLETAS
    if pd.api.types.is_numeric_dtype(dev_raw[c])
]
variables_ols = [
    c for c in numericas_pool
    if 0 < dev_raw[c].isna().mean() < 0.80 and dev_raw[c].notna().sum() >= 100
]

diagnosticos_ols = []
modelos_ols_imputacion = {}

for objetivo_imp in variables_ols:
    candidatos = [c for c in numericas_pool if c != objetivo_imp]
    correlaciones = {}

    for pred in candidatos:
        pares = (
            dev_raw[[objetivo_imp, pred]]
            .apply(pd.to_numeric, errors="coerce")
            .dropna()
        )
        if (
            len(pares) >= 80
            and pares[objetivo_imp].nunique() > 5
            and pares[pred].nunique() > 3
        ):
            correlaciones[pred] = abs(pares.corr().iloc[0, 1])

    predictores = [
        k
        for k, _ in sorted(
            correlaciones.items(),
            key=lambda item: item[1],
            reverse=True,
        )[:5]
    ]
    if len(predictores) < 2:
        continue

    base = dev_raw.loc[
        dev_raw[objetivo_imp].notna(),
        [objetivo_imp] + predictores,
    ].copy()

    y_imp = pd.to_numeric(base[objetivo_imp], errors="coerce")
    X_imp = base[predictores].apply(pd.to_numeric, errors="coerce")
    medianas = X_imp.median()
    X_imp = X_imp.fillna(medianas)

    desvios = X_imp.std(ddof=0).replace(0, 1)
    X_std = (X_imp - X_imp.mean()) / desvios
    X_sm = sm.add_constant(X_std, has_constant="add")

    ajuste = sm.OLS(y_imp, X_sm, missing="drop").fit()
    ajuste_hc3 = ajuste.get_robustcov_results(cov_type="HC3")
    resid = np.asarray(ajuste.resid, dtype=float)

    bp = het_breuschpagan(resid, ajuste.model.exog)
    jb = stats.jarque_bera(resid)

    modelos_ols_imputacion[objetivo_imp] = {
        "ajuste": ajuste,
        "ajuste_hc3": ajuste_hc3,
        "predictores": predictores,
        "medianas": medianas,
        "medias": X_imp.mean(),
        "desvios": desvios,
    }

    p_jb = float(jb.pvalue) # type: ignore
    p_bp = float(bp[1])

    diagnosticos_ols.append({
        "variable_imputada": objetivo_imp,
        "predictores": ", ".join(predictores),
        "n_observado": int(ajuste.nobs),
        "r2": float(ajuste.rsquared),
        "r2_ajustado": float(ajuste.rsquared_adj),
        "rmse_residual": float(np.sqrt(np.mean(resid ** 2))),
        "media_residuo": float(np.mean(resid)),
        "asimetria_residuo": float(stats.skew(resid, nan_policy="omit")),
        "p_jarque_bera": p_jb,
        "normalidad_no_rechazada_95": p_jb >= ALPHA,
        "p_breusch_pagan": p_bp,
        "heterocedasticidad_95": p_bp < ALPHA,
        "covarianza_recomendada": "HC3" if p_bp < ALPHA else "clásica",
    })

diagnosticos_ols = pd.DataFrame(diagnosticos_ols)
diagnosticos_ols.to_csv(
    CSV_DIR / "s3_diagnostico_ols_imputacion.csv",
    index=False,
    encoding="utf-8-sig",
)

if diagnosticos_ols.empty:
    mostrar_interpretacion(
        "Regresión lineal de imputación",
        "No se identificaron variables numéricas faltantes con soporte suficiente "
        "para ajustar OLS explícita.",
        "alerta",
    )
else:
    mostrar_tabla(
        diagnosticos_ols.sort_values("r2", ascending=False),
        "Diagnóstico de las regresiones lineales de imputación",
        max_filas=30,
    )

    mejor_r2 = diagnosticos_ols["r2"].max()
    peor_r2 = diagnosticos_ols["r2"].min()
    n_normalidad = int(diagnosticos_ols["normalidad_no_rechazada_95"].sum())
    n_hetero = int(diagnosticos_ols["heterocedasticidad_95"].sum())
    total_ols = len(diagnosticos_ols)

    lectura_normalidad = (
        f"Jarque–Bera no rechaza normalidad en <b>{n_normalidad} de "
        f"{total_ols}</b> regresiones al 5%."
        if n_normalidad
        else "Jarque–Bera rechaza normalidad en todas las regresiones documentadas."
    )
    lectura_hetero = (
        f"Breusch–Pagan detecta heterocedasticidad en <b>{n_hetero} de "
        f"{total_ols}</b> regresiones; por ello se documenta covarianza HC3."
        if n_hetero
        else "Breusch–Pagan no detecta heterocedasticidad al 5%."
    )

    mostrar_interpretacion(
        "Resultados de las regresiones de imputación",
        [
            f"Los R² describen capacidad explicativa; la estrategia se decide con enmascaramiento, retención y efecto sobre el modelo.",
            lectura_normalidad,
            lectura_hetero,
            "La estrategia definitiva se decide mediante enmascaramiento, NRMSE, "
            "encogimiento, retención y efecto sobre el modelo final.",
        ],
        "alerta" if n_hetero else "info",
    )
```

```python
# =============================================================================
# 6.3 RETENCIÓN, ENMASCARAMIENTO, NRMSE Y ENCOGIMIENTO DE VARIANZA
# =============================================================================
ESTRATEGIAS = ["completos", "mediana", "regresion"]

retencion = []
for estrategia in ESTRATEGIAS:
    usado_dev = (
        filtrar_completos(dev_raw, VARIABLES_COMPLETAS)
        if estrategia == "completos"
        else dev_raw
    )
    usado_val = (
        filtrar_completos(val_raw, VARIABLES_COMPLETAS)
        if estrategia == "completos"
        else val_raw
    )
    retencion.extend([
        {
            "estrategia": estrategia,
            "particion": "desarrollo",
            "n_disponible": len(dev_raw),
            "n_utilizado": len(usado_dev),
            "retencion": len(usado_dev) / max(len(dev_raw), 1),
            "prevalencia": usado_dev[TARGET].mean(),
        },
        {
            "estrategia": estrategia,
            "particion": "validacion",
            "n_disponible": len(val_raw),
            "n_utilizado": len(usado_val),
            "retencion": len(usado_val) / max(len(val_raw), 1),
            "prevalencia": usado_val[TARGET].mean(),
        },
    ])

retencion = pd.DataFrame(retencion)
retencion.to_csv(
    CSV_DIR / "s3_retencion_por_estrategia.csv", index=False, encoding="utf-8-sig"
)

nums_validables = [
    c for c in numericas_pool
    if dev_raw[c].notna().sum() >= 120 and dev_raw[c].nunique(dropna=True) > 5
]
columnas_grupo_mediana = [
    c for c in ["mes_sin", "mes_cos", "invierno"]
    if c in dev_raw.columns
]
variables_mask = [
    c for c in nums_validables
    if c not in columnas_grupo_mediana
]
columnas_mask = list(dict.fromkeys(variables_mask + columnas_grupo_mediana))

resultados_mask = []
if len(variables_mask) >= 2:
    base_num = dev_raw[columnas_mask].apply(pd.to_numeric, errors="coerce")

    for repeticion in range(N_MASK_REPEATS):
        rng_rep = np.random.default_rng(SEED + repeticion)
        masked = base_num.copy()
        mascaras = {}

        for c in variables_mask:
            idx_obs = np.flatnonzero(base_num[c].notna().to_numpy())
            n_mask = max(1, int(len(idx_obs) * 0.10))
            idx_mask = rng_rep.choice(idx_obs, size=n_mask, replace=False)
            mascara = np.zeros(len(base_num), dtype=bool)
            mascara[idx_mask] = True
            mascaras[c] = mascara
            masked.loc[mascara, c] = np.nan

        imputadores = {
            "mediana": ImputadorMedianaMesTemporada(),
            "regresion": IterativeImputer(
                estimator=BayesianRidge(),
                max_iter=50,
                tol=1e-3,
                random_state=SEED + repeticion,
                initial_strategy="median",
                sample_posterior=False,
                skip_complete=False,
            ),
        }

        for metodo, imputador in imputadores.items():
            transformado = imputador.fit_transform(masked)
            imputado = (
                transformado.copy()
                if isinstance(transformado, pd.DataFrame)
                else pd.DataFrame(
                    transformado,
                    columns=columnas_mask,
                    index=base_num.index,
                )
            )

            for c in variables_mask:
                mascara = mascaras[c]
                real = base_num.loc[mascara, c].astype(float)
                pred = imputado.loc[mascara, c].astype(float) # type: ignore
                iqr = float(real.quantile(0.75) - real.quantile(0.25))
                escala = iqr if iqr > 1e-12 else float(real.std(ddof=1))
                escala = escala if escala > 1e-12 else 1.0

                mae = float(np.mean(np.abs(real - pred)))
                rmse = float(np.sqrt(np.mean((real - pred) ** 2)))
                sd_real = float(real.std(ddof=1))
                sd_pred = float(pred.std(ddof=1))
                shrink = 1 - (sd_pred / sd_real) if sd_real > 1e-12 else np.nan

                resultados_mask.append({
                    "repeticion": repeticion + 1,
                    "variable": c,
                    "metodo": metodo,
                    "n_enmascarado": int(mascara.sum()),
                    "mae": mae,
                    "rmse": rmse,
                    "nmae_iqr": mae / escala,
                    "nrmse_iqr": rmse / escala,
                    "sd_real": sd_real,
                    "sd_imputado": sd_pred,
                    "encogimiento_varianza": shrink,
                })

validacion_imputacion = pd.DataFrame(resultados_mask)

if validacion_imputacion.empty:
    resumen_imputacion = pd.DataFrame()
    resumen_global_imputacion = pd.DataFrame()
else:
    resumen_imputacion = (
        validacion_imputacion
        .groupby(["variable", "metodo"], as_index=False)
        .agg(
            n_enmascarado=("n_enmascarado", "sum"),
            nmae_iqr=("nmae_iqr", "mean"),
            nrmse_iqr=("nrmse_iqr", "mean"),
            encogimiento_varianza=("encogimiento_varianza", "mean"),
            sd_real=("sd_real", "mean"),
            sd_imputado=("sd_imputado", "mean"),
        )
    )
    resumen_global_imputacion = (
        resumen_imputacion
        .groupby("metodo", as_index=False)
        .agg(
            nmae_iqr_promedio=("nmae_iqr", "mean"),
            nrmse_iqr_promedio=("nrmse_iqr", "mean"),
            encogimiento_promedio=("encogimiento_varianza", "mean"),
        )
    )

resumen_imputacion.to_csv(
    CSV_DIR / "s3_validacion_imputacion_por_variable.csv",
    index=False,
    encoding="utf-8-sig",
)
resumen_global_imputacion.to_csv(
    CSV_DIR / "s3_validacion_imputacion_global.csv",
    index=False,
    encoding="utf-8-sig",
)

mostrar_tabla(retencion, "Retención de muestra por estrategia")
if not resumen_global_imputacion.empty:
    mostrar_tabla(
        resumen_global_imputacion,
        "Calidad global de imputación en escala comparable",
    )
    mostrar_tabla(
        resumen_imputacion.sort_values(["variable", "nrmse_iqr"]),
        "Diagnóstico por variable: NRMSE y encogimiento",
        max_filas=40,
    )

    mejor_metodo_imp = (
        resumen_global_imputacion
        .sort_values(["nrmse_iqr_promedio", "encogimiento_promedio"])
        .iloc[0]["metodo"]
    )
    mostrar_interpretacion(
        "Comparación de métodos de imputación",
        [
            f"La menor pérdida normalizada por IQR corresponde a <b>{mejor_metodo_imp}</b>.",
            "La estrategia mediana utiliza primero el mismo mes, luego la temporada y finalmente la mediana global.",
            "La decisión final combina error de imputación, retención, estabilidad de coeficientes y desempeño.",
        ],
        "info",
    )
else:
    mostrar_interpretacion(
        "Validación de imputación",
        "No fue posible ejecutar el enmascaramiento porque no existen al menos dos variables numéricas con observaciones suficientes.",
        "alerta",
    )
```

```python
# =============================================================================
# 7.1 AJUSTE GLM PARA AIC/BIC Y SELECCIÓN POR GRUPOS DE VARIABLES
# =============================================================================

# Reinicio transaccional de la auditoría para esta corrida oficial.
# Evita que ajustes de una ejecución previa permanezcan en memoria.
AUDITORIA_AJUSTES_GLM.clear()
ID_CORRIDA_GLM = pd.Timestamp.now("UTC").isoformat()

def preparar_glm(frame, features, estrategia="regresion"):
    datos = (
        filtrar_completos(frame, features)
        if estrategia == "completos"
        else frame.copy()
    )
    if len(datos) == 0 or datos[TARGET].nunique() < 2:
        raise ValueError("No existen datos suficientes para ajustar el GLM.")

    prep, nums, cats = construir_preprocesador(datos, features, estrategia)
    X = np.asarray(
        prep.fit_transform(
            seleccionar_X_preprocesador(datos, features, estrategia)
        ),
        dtype=float,
    )
    nombres = np.asarray(prep.get_feature_names_out(), dtype=str)

    varianzas = np.nanvar(X, axis=0)
    mantener = np.flatnonzero(varianzas > 1e-12)
    X = X[:, mantener]
    nombres = nombres[mantener]

    Xb = sm.add_constant(X, has_constant="add")
    y = datos[TARGET].to_numpy(dtype=int)
    _, ajuste, diagnostico_glm = ajustar_glm_binomial_seguro(
        y,
        Xb,
        maxiter=300,
        contexto=f"selección GLM ({estrategia}; {len(features)} variables)",
        nivel_auditoria="exploratorio",
    )

    return {
        "ajuste": ajuste,
        "aic": float(diagnostico_glm["aic"]),
        "bic": float(diagnostico_glm["bic"]),
        "prep": prep,
        "X": X,
        "y": y,
        "nombres": nombres,
        "datos": datos,
        "nums": nums,
        "cats": cats,
    }


def seleccion_forward(frame, candidatas, mejora_minima=DELTA_BIC_RELEVANTE):
    restantes = list(candidatas)
    seleccionadas = []
    trayectoria = []
    bic_actual = np.inf
    paso = 0

    while restantes:
        pruebas = []
        for candidata in restantes:
            subset = seleccionadas + [candidata]
            try:
                fit = preparar_glm(frame, subset, "regresion")
                pruebas.append((fit["bic"], fit["aic"], candidata, fit))
            except Exception:
                continue

        if not pruebas:
            trayectoria.append({
                "paso": paso + 1,
                "accion": "detener",
                "variable": "—",
                "n_variables": len(seleccionadas),
                "AIC": np.nan,
                "BIC": bic_actual,
                "mejora_BIC": np.nan,
                "regla": "ningún modelo candidato pudo ajustarse",
            })
            break

        bic_nuevo, aic_nuevo, candidata, _ = min(pruebas, key=lambda x: x[0])
        mejora = bic_actual - bic_nuevo if np.isfinite(bic_actual) else np.inf

        # Se garantizan tres variables para construir un modelo comparable.
        if len(seleccionadas) >= 3 and mejora < mejora_minima:
            trayectoria.append({
                "paso": paso + 1,
                "accion": "rechazar incorporación",
                "variable": candidata,
                "n_variables": len(seleccionadas),
                "AIC": aic_nuevo,
                "BIC": bic_nuevo,
                "mejora_BIC": mejora,
                "regla": f"mejora BIC < {mejora_minima:g}",
            })
            break

        seleccionadas.append(candidata)
        restantes.remove(candidata)
        paso += 1
        trayectoria.append({
            "paso": paso,
            "accion": "incorporar",
            "variable": candidata,
            "n_variables": len(seleccionadas),
            "AIC": aic_nuevo,
            "BIC": bic_nuevo,
            "mejora_BIC": mejora,
            "regla": "mejor BIC disponible",
        })
        bic_actual = bic_nuevo

    return seleccionadas, pd.DataFrame(trayectoria)


def seleccion_backward(
    frame,
    candidatas,
    mejora_minima=DELTA_BIC_RELEVANTE,
):
    """
    Elimina variables solo cuando BIC respalda la reducción.

    La trayectoria registra también el mejor intento rechazado. Así, si el
    procedimiento conserva el modelo completo, queda demostrado que la
    selección backward sí se ejecutó y que ninguna eliminación fue defendible.
    """
    seleccionadas = list(candidatas)
    ajuste_actual = preparar_glm(frame, seleccionadas, "regresion")
    bic_actual = ajuste_actual["bic"]

    trayectoria = [{
        "paso": 0,
        "accion": "modelo completo",
        "variable": "—",
        "n_variables": len(seleccionadas),
        "AIC": ajuste_actual["aic"],
        "BIC": bic_actual,
        "delta_BIC": 0.0,
        "regla": "inicio del procedimiento backward",
        "aceptada": True,
    }]

    paso = 0
    motivo_detencion = "sin evaluar"

    while len(seleccionadas) > 3:
        pruebas = []

        for candidata in seleccionadas:
            subset = [c for c in seleccionadas if c != candidata]
            try:
                fit = preparar_glm(frame, subset, "regresion")
                pruebas.append((fit["bic"], fit["aic"], candidata, fit))
            except Exception:
                continue

        if not pruebas:
            motivo_detencion = "ninguna eliminación produjo un GLM ajustable"
            trayectoria.append({
                "paso": paso + 1,
                "accion": "detener",
                "variable": "—",
                "n_variables": len(seleccionadas),
                "AIC": np.nan,
                "BIC": bic_actual,
                "delta_BIC": np.nan,
                "regla": motivo_detencion,
                "aceptada": False,
            })
            break

        bic_nuevo, aic_nuevo, candidata, _ = min(pruebas, key=lambda x: x[0])
        delta = bic_nuevo - bic_actual

        aceptar = delta <= -mejora_minima
        regla = f"reducción BIC ≥ {mejora_minima:g}"

        if not aceptar:
            motivo_detencion = (
                f"mejor eliminación rechazada: ΔBIC={delta:.3f}; "
                "la complejidad adicional conserva respaldo estadístico"
            )
            trayectoria.append({
                "paso": paso + 1,
                "accion": "rechazar eliminación",
                "variable": candidata,
                "n_variables": len(seleccionadas),
                "AIC": aic_nuevo,
                "BIC": bic_nuevo,
                "delta_BIC": delta,
                "regla": motivo_detencion,
                "aceptada": False,
            })
            break

        seleccionadas.remove(candidata)
        bic_actual = bic_nuevo
        paso += 1

        trayectoria.append({
            "paso": paso,
            "accion": "eliminar",
            "variable": candidata,
            "n_variables": len(seleccionadas),
            "AIC": aic_nuevo,
            "BIC": bic_nuevo,
            "delta_BIC": delta,
            "regla": regla,
            "aceptada": True,
        })

    if motivo_detencion == "sin evaluar":
        motivo_detencion = "se alcanzó el mínimo de tres variables"

    metadata = {
        "n_inicial": len(candidatas),
        "n_final": len(seleccionadas),
        "n_eliminadas": len(candidatas) - len(seleccionadas),
        "reduccion_real": len(seleccionadas) < len(candidatas),
        "motivo_detencion": motivo_detencion,
    }
    return seleccionadas, pd.DataFrame(trayectoria), metadata


M1_DOMINIO = list(VARIABLES_DOMINIO)
M2_FORWARD, trayectoria_forward = seleccion_forward(
    dev_raw,
    VARIABLES_COMPLETAS,
)
M3_BACKWARD, trayectoria_backward, meta_backward = seleccion_backward(
    dev_raw,
    VARIABLES_COMPLETAS,
)

if len(M2_FORWARD) < 3:
    raise ValueError("La selección forward produjo menos de tres variables.")
if len(M3_BACKWARD) < 3:
    raise ValueError("La selección backward produjo menos de tres variables.")

MODELOS = {
    "M1_Dominio_S1_S2": M1_DOMINIO,
    "M2_Forward_AIC_BIC": M2_FORWARD,
    "M3_Backward_BIC": M3_BACKWARD,
}

matriz_modelos = []
for nombre, variables in MODELOS.items():
    for variable in variables:
        matriz_modelos.append({
            "modelo": nombre,
            "variable": variable,
            "familia": matriz_candidatas.set_index("variable").loc[variable, "familia"], # type: ignore
            "faltantes_%": 100 * df[variable].isna().mean(),
        })

matriz_modelos = pd.DataFrame(matriz_modelos)

resumen_seleccion = pd.DataFrame([
    {
        "modelo": "M1_Dominio_S1_S2",
        "metodo": "evidencia S1/S2 y fundamento físico",
        "n_inicial": len(M1_DOMINIO),
        "n_final": len(M1_DOMINIO),
        "n_eliminadas": 0,
        "resultado": "conjunto definido por dominio",
    },
    {
        "modelo": "M2_Forward_AIC_BIC",
        "metodo": "incorporación forward por BIC",
        "n_inicial": 0,
        "n_final": len(M2_FORWARD),
        "n_eliminadas": 0,
        "resultado": "variables incorporadas progresivamente",
    },
    {
        "modelo": "M3_Backward_BIC",
        "metodo": "eliminación backward por BIC",
        "n_inicial": meta_backward["n_inicial"],
        "n_final": meta_backward["n_final"],
        "n_eliminadas": meta_backward["n_eliminadas"],
        "resultado": meta_backward["motivo_detencion"],
    },
])

matriz_modelos.to_csv(
    CSV_DIR / "s3_matriz_variables_modelos.csv",
    index=False,
    encoding="utf-8-sig",
)
trayectoria_forward.to_csv(
    CSV_DIR / "s3_trayectoria_forward_aic_bic.csv",
    index=False,
    encoding="utf-8-sig",
)
trayectoria_backward.to_csv(
    CSV_DIR / "s3_trayectoria_backward_bic.csv",
    index=False,
    encoding="utf-8-sig",
)
resumen_seleccion.to_csv(
    CSV_DIR / "s3_resumen_seleccion_tres_modelos.csv",
    index=False,
    encoding="utf-8-sig",
)

mostrar_tabla(
    resumen_seleccion,
    "Ejecución verificable de los tres modelos oficiales",
)
mostrar_tabla(
    matriz_modelos,
    "Variables congeladas en los tres modelos",
    max_filas=60,
)
mostrar_tabla(
    trayectoria_forward,
    "Trayectoria M2 · forward AIC/BIC",
    max_filas=40,
)
mostrar_tabla(
    trayectoria_backward,
    "Trayectoria M3 · backward BIC",
    max_filas=40,
)

if meta_backward["reduccion_real"]:
    lectura_m3 = (
        f"M3 eliminó <b>{meta_backward['n_eliminadas']}</b> variables y terminó "
        f"con <b>{meta_backward['n_final']}</b>."
    )
else:
    lectura_m3 = (
        f"M3 ejecutó la evaluación backward sobre <b>{meta_backward['n_inicial']}</b> "
        "variables, pero ninguna eliminación cumplió el criterio BIC. Por ello conserva "
        "el modelo completo; esto es un resultado del procedimiento, no una falla."
    )

mostrar_interpretacion(
    "Lectura de la selección",
    [
        f"M1 conserva <b>{len(M1_DOMINIO)}</b> variables sustentadas por S1/S2.",
        f"M2 retuvo <b>{len(M2_FORWARD)}</b> variables mediante incorporación "
        "forward y criterio BIC.",
        lectura_m3,
        "Los tres modelos representan procedimientos distintos aunque dos de ellos "
        "puedan converger al mismo conjunto de predictores.",
    ],
    "info",
)
```

```python
# =============================================================================
# 8.1 FOLDS TEMPORALES, UMBRAL OOF Y MÉTRICAS
# =============================================================================
def seleccionar_umbral(y_true, prob, costo_fn=COSTO_FN, costo_fp=COSTO_FP):
    filas = []
    for umbral in np.linspace(0.02, 0.98, 193):
        pred = (prob >= umbral).astype(int)
        tn, fp, fn, tp = confusion_matrix(y_true, pred, labels=[0, 1]).ravel()
        filas.append({
            "umbral": umbral,
            "precision": precision_score(y_true, pred, zero_division=0),
            "recall": recall_score(y_true, pred, zero_division=0),
            "f1": f1_score(y_true, pred, zero_division=0),
            "f2": fbeta_score(y_true, pred, beta=2, zero_division=0),
            "costo": costo_fn * fn + costo_fp * fp,
            "tn": tn, "fp": fp, "fn": fn, "tp": tp,
        })
    tabla = pd.DataFrame(filas)
    elegible = tabla.loc[tabla["recall"] >= MIN_RECALL]
    if elegible.empty:
        elegible = tabla
    mejor = elegible.sort_values(
        ["f2", "costo", "recall", "precision"],
        ascending=[False, True, False, False],
    ).iloc[0]
    return float(mejor["umbral"]), tabla

def calcular_metricas(y_true, prob, umbral):
    pred = (prob >= umbral).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_true, pred, labels=[0, 1]).ravel()
    especificidad = tn / (tn + fp) if (tn + fp) else np.nan
    return {
        "n": len(y_true),
        "positivos": int(np.sum(y_true)),
        "prevalencia": float(np.mean(y_true)),
        "umbral": float(umbral),
        "accuracy": accuracy_score(y_true, pred),
        "balanced_accuracy": balanced_accuracy_score(y_true, pred),
        "precision": precision_score(y_true, pred, zero_division=0),
        "recall": recall_score(y_true, pred, zero_division=0),
        "specificity": especificidad,
        "f1": f1_score(y_true, pred, zero_division=0),
        "f2": fbeta_score(y_true, pred, beta=2, zero_division=0),
        "roc_auc": roc_auc_score(y_true, prob),
        "pr_auc": average_precision_score(y_true, prob),
        "brier": brier_score_loss(y_true, prob),
        "tn": tn, "fp": fp, "fn": fn, "tp": tp,
    }


def ajustar_calibrador_platt(y_true, prob):
    """Calibra probabilidades ponderadas usando predicciones OOF temporales."""
    y_arr = np.asarray(y_true, dtype=int)
    p_arr = np.clip(np.asarray(prob, dtype=float), 1e-6, 1 - 1e-6)
    logit = np.log(p_arr / (1 - p_arr)).reshape(-1, 1)

    configuraciones = [
        {"penalty": None},
        {"penalty": "none"},
        {"penalty": "l2", "C": 1e6},
    ]
    ultimo_error = None
    for config in configuraciones:
        try:
            modelo = LogisticRegression(
                solver="lbfgs",
                max_iter=3000,
                class_weight=None,
                random_state=SEED,
                **config,
            )
            modelo.fit(logit, y_arr)
            return modelo
        except Exception as exc:
            ultimo_error = exc
    raise RuntimeError(f"No fue posible ajustar calibración Platt: {ultimo_error}")


def aplicar_calibrador_platt(modelo, prob):
    p_arr = np.clip(np.asarray(prob, dtype=float), 1e-6, 1 - 1e-6)
    logit = np.log(p_arr / (1 - p_arr)).reshape(-1, 1)
    return modelo.predict_proba(logit)[:, 1]

def folds_temporales(frame, n_splits=N_CV_SPLITS):
    fechas = np.array(sorted(frame[col_fecha].dt.normalize().unique()))
    if len(fechas) < n_splits + 2:
        raise ValueError("No existen fechas suficientes para los folds temporales.")
    bloques = [b for b in np.array_split(fechas, n_splits + 1) if len(b)]
    for i in range(1, len(bloques)):
        fechas_train_fold = np.concatenate(bloques[:i])
        fechas_val_fold = bloques[i]
        idx_train = frame[col_fecha].dt.normalize().isin(fechas_train_fold)
        idx_val = frame[col_fecha].dt.normalize().isin(fechas_val_fold)
        yield i, frame.loc[idx_train].copy(), frame.loc[idx_val].copy()

def cv_temporal_oof(frame, features, estrategia):
    filas = []
    y_oof = []
    p_oof = []

    for fold, tr, va in folds_temporales(frame):
        if estrategia == "completos":
            tr = filtrar_completos(tr, features)
            va = filtrar_completos(va, features)
        if len(tr) == 0 or len(va) == 0:
            continue
        if tr[TARGET].nunique() < 2 or va[TARGET].nunique() < 2:
            continue

        pipe, _, _ = pipeline_predictivo(tr, features, estrategia)
        pipe.fit(
            seleccionar_X_preprocesador(tr, features, estrategia),
            tr[TARGET],
        )
        prob = pipe.predict_proba(
            seleccionar_X_preprocesador(va, features, estrategia)
        )[:, 1]
        y_fold = va[TARGET].to_numpy(dtype=int)

        filas.append({
            "fold": fold,
            "n_train": len(tr),
            "n_validacion": len(va),
            "roc_auc": roc_auc_score(y_fold, prob),
            "pr_auc": average_precision_score(y_fold, prob),
            "brier": brier_score_loss(y_fold, prob),
        })
        y_oof.append(y_fold)
        p_oof.append(prob)

    if not filas:
        raise RuntimeError("No fue posible construir folds temporales válidos.")

    tabla_folds = pd.DataFrame(filas)
    y_oof = np.concatenate(y_oof)
    p_oof = np.concatenate(p_oof)
    umbral, tabla_umbral = seleccionar_umbral(y_oof, p_oof)

    return tabla_folds, y_oof, p_oof, umbral, tabla_umbral

def evaluar_en_validacion(nombre_modelo, features, estrategia):
    folds, y_oof, p_oof, umbral, tabla_umbral = cv_temporal_oof(
        dev_raw, features, estrategia
    )

    dev = (
        filtrar_completos(dev_raw, features)
        if estrategia == "completos"
        else dev_raw.copy()
    )
    val = (
        filtrar_completos(val_raw, features)
        if estrategia == "completos"
        else val_raw.copy()
    )

    pipe, nums, cats = pipeline_predictivo(dev, features, estrategia)
    pipe.fit(
        seleccionar_X_preprocesador(dev, features, estrategia),
        dev[TARGET],
    )
    p_val = pipe.predict_proba(
        seleccionar_X_preprocesador(val, features, estrategia)
    )[:, 1]
    metricas_val = calcular_metricas(val[TARGET].to_numpy(dtype=int), p_val, umbral)

    resumen = {
        "modelo": nombre_modelo,
        "estrategia": estrategia,
        "n_features_base": len(features),
        "retencion_dev": len(dev) / max(len(dev_raw), 1),
        "retencion_val": len(val) / max(len(val_raw), 1),
        "cv_roc_auc_mean": folds["roc_auc"].mean(),
        "cv_roc_auc_sd": folds["roc_auc"].std(ddof=1),
        "cv_pr_auc_mean": folds["pr_auc"].mean(),
        "cv_pr_auc_sd": folds["pr_auc"].std(ddof=1),
        "cv_brier_mean": folds["brier"].mean(),
    }
    resumen.update({f"val_{k}": v for k, v in metricas_val.items()})

    return {
        "resumen": resumen,
        "pipeline_dev": pipe,
        "features": list(features),
        "nums": nums,
        "cats": cats,
        "dev": dev,
        "val": val,
        "prob_val": p_val,
        "umbral": umbral,
        "tabla_umbral": tabla_umbral,
        "folds": folds,
    }
```

```python
# =============================================================================
# 8.2 ENTRENAMIENTO DE LAS NUEVE COMBINACIONES EN DESARROLLO/VALIDACIÓN
# =============================================================================
artefactos_validacion = {}
filas_metricas = []
inicio = time.time()

for nombre_modelo, features in MODELOS.items():
    for estrategia in ESTRATEGIAS:
        clave = (nombre_modelo, estrategia)
        #print("Evaluando:", clave)
        artefacto = evaluar_en_validacion(nombre_modelo, features, estrategia)
        artefactos_validacion[clave] = artefacto
        filas_metricas.append(artefacto["resumen"])

metricas_validacion = pd.DataFrame(filas_metricas)

# ---------------------------------------------------------------------------
# Regla predefinida de equivalencia y selección.
#
# 1) Se exige PR-AUC dentro de 0,01 del mejor.
# 2) Recall no puede caer más de 0,02 respecto del mejor candidato y nunca
#    puede quedar bajo el mínimo operacional.
# 3) Brier debe quedar dentro de 0,005 del mejor.
# 4) Entre alternativas equivalentes se prioriza F2, luego evitar imputación
#    cuando casos completos conserva ≥99% de la muestra, después parsimonia.
# ---------------------------------------------------------------------------
mejor_pr = metricas_validacion["cv_pr_auc_mean"].max()
mejor_recall = metricas_validacion["val_recall"].max()
mejor_brier = metricas_validacion["val_brier"].min()

metricas_validacion["equivalente_pr"] = (
    metricas_validacion["cv_pr_auc_mean"] >= mejor_pr - PR_AUC_EQUIV
)
metricas_validacion["equivalente_recall"] = (
    metricas_validacion["val_recall"]
    >= max(MIN_RECALL, mejor_recall - RECALL_EQUIV)
)
metricas_validacion["equivalente_brier"] = (
    metricas_validacion["val_brier"] <= mejor_brier + BRIER_EQUIV
)
metricas_validacion["candidato_equivalente"] = (
    metricas_validacion[
        ["equivalente_pr", "equivalente_recall", "equivalente_brier"]
    ].all(axis=1)
)

candidatos = metricas_validacion.loc[
    metricas_validacion["candidato_equivalente"]
].copy()

if candidatos.empty:
    candidatos = metricas_validacion.loc[
        metricas_validacion["equivalente_pr"]
        & (metricas_validacion["val_recall"] >= MIN_RECALL)
    ].copy()

if candidatos.empty:
    candidatos = metricas_validacion.copy()

def prioridad_estrategia(fila):
    if (
        fila["estrategia"] == "completos"
        and fila["retencion_dev"] >= MIN_RETENCION_COMPLETOS
        and fila["retencion_val"] >= MIN_RETENCION_COMPLETOS
    ):
        return 0
    if fila["estrategia"] == "regresion":
        return 1
    if fila["estrategia"] == "mediana":
        return 2
    return 3

candidatos["prioridad_estrategia"] = candidatos.apply(
    prioridad_estrategia,
    axis=1,
)

candidatos = candidatos.sort_values(
    [
        "val_f2",
        "cv_pr_auc_mean",
        "prioridad_estrategia",
        "val_brier",
        "n_features_base",
        "retencion_dev",
    ],
    ascending=[False, False, True, True, True, False],
)

seleccion_preliminar = candidatos.iloc[0]
CLAVE_PRELIMINAR = (
    seleccion_preliminar["modelo"],
    seleccion_preliminar["estrategia"],
)
ARTEFACTO_PRELIMINAR = artefactos_validacion[CLAVE_PRELIMINAR]

metricas_validacion["seleccion_preliminar"] = (
    (metricas_validacion["modelo"] == CLAVE_PRELIMINAR[0])
    & (metricas_validacion["estrategia"] == CLAVE_PRELIMINAR[1])
)

metricas_validacion.to_csv(
    CSV_DIR / "s3_metricas_9_combinaciones_validacion.csv",
    index=False,
    encoding="utf-8-sig",
)

decision_preliminar = pd.DataFrame([{
    "modelo_preliminar": CLAVE_PRELIMINAR[0],
    "estrategia_preliminar": CLAVE_PRELIMINAR[1],
    "cv_pr_auc": seleccion_preliminar["cv_pr_auc_mean"],
    "cv_roc_auc": seleccion_preliminar["cv_roc_auc_mean"],
    "val_recall": seleccion_preliminar["val_recall"],
    "val_f2": seleccion_preliminar["val_f2"],
    "val_brier": seleccion_preliminar["val_brier"],
    "n_features": seleccion_preliminar["n_features_base"],
    "retencion_dev": seleccion_preliminar["retencion_dev"],
    "regla": (
        "equivalencia PR-AUC/Recall/Brier; F2; evitar imputación si "
        "retención ≥99%; parsimonia"
    ),
    "test_utilizado_en_seleccion": False,
}])

metricas_validacion_ordenadas = metricas_validacion.sort_values(
    ["seleccion_preliminar", "cv_pr_auc_mean", "val_f2"],
    ascending=[False, False, False],
)

columnas_seleccion = [
    "modelo", "estrategia", "n_features_base", "retencion_dev",
    "cv_pr_auc_mean", "cv_roc_auc_mean", "val_precision",
    "val_recall", "val_f2", "val_brier",
    "candidato_equivalente", "seleccion_preliminar",
]
mostrar_tabla(
    metricas_validacion_ordenadas[columnas_seleccion],
    "Comparación ejecutiva de los tres modelos × tres estrategias",
    max_filas=20,
)

columnas_variabilidad = [
    "modelo", "estrategia", "cv_pr_auc_sd", "cv_roc_auc_sd",
    "retencion_val", "val_umbral", "val_specificity",
    "val_balanced_accuracy",
]
mostrar_tabla(
    metricas_validacion_ordenadas[columnas_variabilidad],
    "Estabilidad temporal y métricas complementarias de validación",
    max_filas=20,
)
mostrar_tabla(
    decision_preliminar,
    "Decisión preliminar cerrada antes de evaluar forma funcional",
)

mostrar_interpretacion(
    "🏆 Combinación preliminar sin fuga de prueba",
    [
        f"Se preseleccionó <b>{CLAVE_PRELIMINAR[0]}</b> con estrategia "
        f"<b>{CLAVE_PRELIMINAR[1]}</b>.",
        f"CV PR-AUC={seleccion_preliminar['cv_pr_auc_mean']:.3f}, "
        f"Recall de validación={seleccion_preliminar['val_recall']:.3f}, "
        f"F2={seleccion_preliminar['val_f2']:.3f}.",
        "La regla distingue equivalencia práctica de diferencias decimales y "
        "evita imputar cuando casos completos conserva al menos 99% de la muestra.",
        f"El proceso tomó <b>{time.time() - inicio:.1f} segundos</b> y no utilizó "
        "métricas del conjunto de prueba.",
    ],
    "ok",
)
```

    <p style="margin:0 0 10px 0;">La linealidad en el logit se evalúa exclusivamente con
    desarrollo. Cuando el término Box–Tidwell identifica curvatura, se construye una
    sensibilidad con términos cuadráticos centrados.</p>
    <p style="margin:0;">La extensión solo reemplaza al modelo lineal si mejora BIC o Brier,
    manteniendo PR-AUC y Recall dentro de márgenes de no inferioridad definidos previamente.
    El conjunto de prueba continúa bloqueado.</p>

```python
# =============================================================================
# 8.3 DIAGNÓSTICO BOX–TIDWELL Y SENSIBILIDAD CUADRÁTICA SIN TEST
# =============================================================================
def diagnosticar_linealidad_desarrollo(frame, features, estrategia):
    """Prueba exploratoria Box–Tidwell sobre predictores numéricos transformados."""
    base = preparar_glm(frame, features, estrategia)
    X = base["X"]
    Xb = sm.add_constant(X, has_constant="add")
    y = base["y"]
    nombres = np.asarray(base["nombres"], dtype=str)

    filas = []
    for variable in base["nums"]:
        coincidencias = np.flatnonzero(nombres == variable)
        if len(coincidencias) != 1:
            continue

        j = int(coincidencias[0])
        x = X[:, j]
        x_pos = x - np.nanmin(x) + 1e-6
        termino = x_pos * np.log(x_pos)
        X_ext = np.column_stack([Xb, termino])

        try:
            _, ajuste_ext, _ = ajustar_glm_binomial_seguro(
                y,
                X_ext,
                maxiter=300,
                contexto=f"Box–Tidwell {variable}",
                nivel_auditoria="exploratorio",
            )
            p_bt = float(ajuste_ext.pvalues[-1])
            if not np.isfinite(p_bt):
                p_bt = np.nan
        except Exception:
            p_bt = np.nan

        filas.append({
            "variable": variable,
            "p_box_tidwell": p_bt,
            "evidencia_curvatura": bool(pd.notna(p_bt) and p_bt < ALPHA),
            "lectura": (
                "curvatura candidata"
                if pd.notna(p_bt) and p_bt < ALPHA
                else "sin evidencia fuerte de curvatura"
            ),
        })

    return pd.DataFrame(filas)


features_base_preliminar = list(MODELOS[CLAVE_PRELIMINAR[0]])
estrategia_preliminar = CLAVE_PRELIMINAR[1]

diagnostico_linealidad_pretest = diagnosticar_linealidad_desarrollo(
    dev_raw,
    features_base_preliminar,
    estrategia_preliminar,
)

variables_curvatura = (
    diagnostico_linealidad_pretest.loc[
        diagnostico_linealidad_pretest["evidencia_curvatura"],
        "variable",
    ]
    .head(3)
    .tolist()
    if not diagnostico_linealidad_pretest.empty
    else []
)

centros_cuadraticos = {
    variable: float(pd.to_numeric(dev_raw[variable], errors="coerce").median())
    for variable in variables_curvatura
}

def incorporar_cuadraticos(frame, variables, centros):
    """Agrega cuadrados centrados con centros aprendidos solo en desarrollo."""
    for variable in variables:
        nueva = f"{variable}_cuadratico"
        valores = pd.to_numeric(frame[variable], errors="coerce")
        frame[nueva] = (valores - centros[variable]) ** 2
    return frame

for particion in [df, train_raw, test_raw, dev_raw, val_raw]:
    incorporar_cuadraticos(
        particion,
        variables_curvatura,
        centros_cuadraticos,
    )

comparacion_forma = []
adoptar_cuadraticos = False
nombre_modelo_final = CLAVE_PRELIMINAR[0]
features_finales_candidatas = features_base_preliminar
ARTEFACTO_SELECCION = ARTEFACTO_PRELIMINAR

ajuste_base_dev = preparar_glm(
    dev_raw,
    features_base_preliminar,
    estrategia_preliminar,
)
resumen_base = ARTEFACTO_PRELIMINAR["resumen"]

comparacion_forma.append({
    "forma": "lineal en el logit",
    "modelo": CLAVE_PRELIMINAR[0],
    "n_features": len(features_base_preliminar),
    "BIC_dev": ajuste_base_dev["bic"],
    "cv_pr_auc": resumen_base["cv_pr_auc_mean"],
    "val_recall": resumen_base["val_recall"],
    "val_f2": resumen_base["val_f2"],
    "val_brier": resumen_base["val_brier"],
    "seleccionada": True,
})

if variables_curvatura:
    features_cuadraticas = features_base_preliminar + [
        f"{variable}_cuadratico"
        for variable in variables_curvatura
    ]
    nombre_cuadratico = f"{CLAVE_PRELIMINAR[0]}_Cuadratico"

    artefacto_cuadratico = evaluar_en_validacion(
        nombre_cuadratico,
        features_cuadraticas,
        estrategia_preliminar,
    )
    resumen_cuadratico = artefacto_cuadratico["resumen"]
    ajuste_cuadratico_dev = preparar_glm(
        dev_raw,
        features_cuadraticas,
        estrategia_preliminar,
    )

    delta_bic = ajuste_cuadratico_dev["bic"] - ajuste_base_dev["bic"]
    delta_brier = (
        resumen_cuadratico["val_brier"] - resumen_base["val_brier"]
    )
    no_inferior_pr = (
        resumen_cuadratico["cv_pr_auc_mean"]
        >= resumen_base["cv_pr_auc_mean"] - 0.005
    )
    no_inferior_recall = (
        resumen_cuadratico["val_recall"]
        >= resumen_base["val_recall"] - RECALL_EQUIV
    )
    mejora_ajuste = (
        delta_bic <= -DELTA_BIC_RELEVANTE
        or delta_brier <= -0.001
    )

    adoptar_cuadraticos = bool(
        no_inferior_pr
        and no_inferior_recall
        and mejora_ajuste
    )

    comparacion_forma.append({
        "forma": "cuadrática centrada",
        "modelo": nombre_cuadratico,
        "n_features": len(features_cuadraticas),
        "BIC_dev": ajuste_cuadratico_dev["bic"],
        "cv_pr_auc": resumen_cuadratico["cv_pr_auc_mean"],
        "val_recall": resumen_cuadratico["val_recall"],
        "val_f2": resumen_cuadratico["val_f2"],
        "val_brier": resumen_cuadratico["val_brier"],
        "seleccionada": adoptar_cuadraticos,
    })

    if adoptar_cuadraticos:
        MODELOS[nombre_cuadratico] = features_cuadraticas
        artefactos_validacion[
            (nombre_cuadratico, estrategia_preliminar)
        ] = artefacto_cuadratico
        nombre_modelo_final = nombre_cuadratico
        features_finales_candidatas = features_cuadraticas
        ARTEFACTO_SELECCION = artefacto_cuadratico

CLAVE_FINAL = (
    nombre_modelo_final,
    estrategia_preliminar,
)
comparacion_forma = pd.DataFrame(comparacion_forma)

if not comparacion_forma.empty:
    comparacion_forma["delta_BIC_vs_lineal"] = (
        comparacion_forma["BIC_dev"]
        - comparacion_forma.iloc[0]["BIC_dev"]
    )
    comparacion_forma["delta_Brier_vs_lineal"] = (
        comparacion_forma["val_brier"]
        - comparacion_forma.iloc[0]["val_brier"]
    )

decision_forma_funcional = pd.DataFrame([{
    "modelo_base": CLAVE_PRELIMINAR[0],
    "modelo_final": CLAVE_FINAL[0],
    "estrategia_final": CLAVE_FINAL[1],
    "variables_con_curvatura": ", ".join(variables_curvatura) or "ninguna",
    "terminos_cuadraticos_incorporados": (
        ", ".join(
            f"{v}_cuadratico"
            for v in variables_curvatura
        )
        if adoptar_cuadraticos
        else "ninguno"
    ),
    "decision": (
        "se adopta extensión cuadrática"
        if adoptar_cuadraticos
        else "se conserva forma lineal"
    ),
    "test_utilizado": False,
}])

decision_modelo = pd.DataFrame([{
    "modelo_base_oficial": CLAVE_PRELIMINAR[0],
    "modelo_final": CLAVE_FINAL[0],
    "estrategia_final": CLAVE_FINAL[1],
    "forma_funcional": (
        "cuadrática centrada"
        if adoptar_cuadraticos
        else "lineal en el logit"
    ),
    "cv_pr_auc": ARTEFACTO_SELECCION["resumen"]["cv_pr_auc_mean"],
    "val_recall": ARTEFACTO_SELECCION["resumen"]["val_recall"],
    "val_f2": ARTEFACTO_SELECCION["resumen"]["val_f2"],
    "val_brier": ARTEFACTO_SELECCION["resumen"]["val_brier"],
    "n_features": len(features_finales_candidatas),
    "regla": (
        "nueve combinaciones oficiales + sensibilidad de forma funcional "
        "con no inferioridad y BIC/Brier"
    ),
    "test_utilizado_en_seleccion": False,
}])

diagnostico_linealidad_pretest.to_csv(
    CSV_DIR / "s3_linealidad_logit_pretest.csv",
    index=False,
    encoding="utf-8-sig",
)
comparacion_forma.to_csv(
    CSV_DIR / "s3_sensibilidad_forma_funcional.csv",
    index=False,
    encoding="utf-8-sig",
)
decision_forma_funcional.to_csv(
    CSV_DIR / "s3_decision_forma_funcional.csv",
    index=False,
    encoding="utf-8-sig",
)
decision_modelo.to_csv(
    CSV_DIR / "s3_decision_modelo_final_sin_test.csv",
    index=False,
    encoding="utf-8-sig",
)

mostrar_tabla(
    diagnostico_linealidad_pretest,
    "Diagnóstico Box–Tidwell en desarrollo",
)
mostrar_tabla(
    comparacion_forma,
    "Sensibilidad lineal versus cuadrática",
)
mostrar_tabla(
    decision_forma_funcional,
    "Decisión de forma funcional antes de abrir el test",
)
mostrar_tabla(
    decision_modelo,
    "Modelo definitivo congelado sin observar prueba",
)

if variables_curvatura:
    lectura_curvatura = (
        f"Se detectó evidencia exploratoria de curvatura en "
        f"<b>{', '.join(variables_curvatura)}</b>."
    )
else:
    lectura_curvatura = (
        "No se detectó evidencia fuerte de curvatura en los predictores "
        "numéricos del modelo preseleccionado."
    )

mostrar_interpretacion(
    "✅ Cierre de forma funcional",
    [
        lectura_curvatura,
        (
            "La extensión cuadrática fue adoptada porque mejoró ajuste o calibración "
            "sin deteriorar materialmente PR-AUC ni Recall."
            if adoptar_cuadraticos
            else "La sensibilidad fue evaluada, pero no justificó aumentar la "
            "complejidad bajo las reglas predefinidas."
        ),
        f"El modelo congelado es <b>{CLAVE_FINAL[0]}</b> con estrategia "
        f"<b>{CLAVE_FINAL[1]}</b>.",
        "El conjunto de prueba permanece completamente bloqueado.",
    ],
    "ok",
)
```

```python
# =============================================================================
# 9.1 AJUSTE FINAL Y APERTURA ÚNICA DEL TEST
# =============================================================================
features_finales = MODELOS[CLAVE_FINAL[0]]
estrategia_final = CLAVE_FINAL[1]

folds_finales, y_oof_final, p_oof_final_raw, _, _ = cv_temporal_oof(
    train_raw, features_finales, estrategia_final
)
calibrador_final = ajustar_calibrador_platt(
    y_oof_final,
    p_oof_final_raw,
)
p_oof_final = aplicar_calibrador_platt(
    calibrador_final,
    p_oof_final_raw,
)
umbral_final, tabla_umbral_final = seleccionar_umbral(
    y_oof_final,
    p_oof_final,
)

auditoria_calibracion_oof = pd.DataFrame([{
    "particion": "OOF temporal de entrenamiento",
    "n": len(y_oof_final),
    "brier_sin_calibrar": brier_score_loss(
        y_oof_final,
        p_oof_final_raw,
    ),
    "brier_calibrado": brier_score_loss(
        y_oof_final,
        p_oof_final,
    ),
    "intercepto_platt": float(calibrador_final.intercept_[0]),
    "pendiente_platt": float(calibrador_final.coef_[0, 0]),
    "estado": "CUMPLIDO · calibración aprendida sin test",
}])
auditoria_calibracion_oof.to_csv(
    CSV_DIR / "s3_calibracion_platt_oof.csv",
    index=False,
    encoding="utf-8-sig",
)
mostrar_tabla(
    auditoria_calibracion_oof,
    "Calibración Platt aprendida con predicciones OOF",
)

train_final = (
    filtrar_completos(train_raw, features_finales)
    if estrategia_final == "completos"
    else train_raw.copy()
)
test_final = (
    filtrar_completos(test_raw, features_finales)
    if estrategia_final == "completos"
    else test_raw.copy()
)

pipeline_final, nums_finales, cats_finales = pipeline_predictivo(
    train_final, features_finales, estrategia_final
)
pipeline_final.fit(
    seleccionar_X_preprocesador(
        train_final,
        features_finales,
        estrategia_final,
    ),
    train_final[TARGET],
)

prob_train_raw = pipeline_final.predict_proba(
    seleccionar_X_preprocesador(
        train_final,
        features_finales,
        estrategia_final,
    )
)[:, 1]
prob_test_raw = pipeline_final.predict_proba(
    seleccionar_X_preprocesador(
        test_final,
        features_finales,
        estrategia_final,
    )
)[:, 1]
prob_train = aplicar_calibrador_platt(
    calibrador_final,
    prob_train_raw,
)
prob_test = aplicar_calibrador_platt(
    calibrador_final,
    prob_test_raw,
)

metricas_train = calcular_metricas(
    train_final[TARGET].to_numpy(dtype=int), prob_train, umbral_final
)
metricas_test = calcular_metricas(
    test_final[TARGET].to_numpy(dtype=int), prob_test, umbral_final
)

tabla_metricas_final = pd.DataFrame([
    {"particion": "entrenamiento", **metricas_train},
    {"particion": "prueba_futura", **metricas_test},
])
tabla_metricas_final.to_csv(
    CSV_DIR / "s3_metricas_modelo_final_train_test.csv",
    index=False,
    encoding="utf-8-sig",
)
tabla_umbral_final.to_csv(
    CSV_DIR / "s3_sensibilidad_umbral_oof.csv",
    index=False,
    encoding="utf-8-sig",
)

FINAL = {
    "pipeline": pipeline_final,
    "features": features_finales,
    "estrategia": estrategia_final,
    "train": train_final,
    "test": test_final,
    "nums": nums_finales,
    "cats": cats_finales,
    "umbral": umbral_final,
    "prob_train_raw": prob_train_raw,
    "prob_test_raw": prob_test_raw,
    "prob_train": prob_train,
    "prob_test": prob_test,
    "calibrador": calibrador_final,
    "metricas_train": metricas_train,
    "metricas_test": metricas_test,
    "folds": folds_finales,
    "tabla_umbral": tabla_umbral_final,
}

tabla_desempeno_final = tabla_metricas_final[[
    "particion", "n", "positivos", "prevalencia", "umbral",
    "precision", "recall", "specificity", "f1", "f2",
    "roc_auc", "pr_auc", "brier",
]].copy()

tabla_conteos_final = tabla_metricas_final[[
    "particion", "tn", "fp", "fn", "tp",
    "accuracy", "balanced_accuracy",
]].copy()

mostrar_tabla(
    tabla_desempeno_final,
    "Desempeño final: métricas principales",
)
mostrar_tabla(
    tabla_conteos_final,
    "Desempeño final: conteos y exactitud",
)
mostrar_kpis(
    "Resultado operacional en prueba",
    [
        ("Recall", f"{metricas_test['recall']:.3f}", "Episodios detectados"),
        ("Precision", f"{metricas_test['precision']:.3f}", "Alertas verdaderas"),
        ("F2", f"{metricas_test['f2']:.3f}", "Prioriza falsos negativos"),
        ("PR-AUC", f"{metricas_test['pr_auc']:.3f}", "Desbalance"),
        ("ROC-AUC", f"{metricas_test['roc_auc']:.3f}", "Discriminación"),
        ("Umbral", f"{umbral_final:.3f}", "Estimado con OOF"),
    ],
)

diferencia_auc = metricas_train["roc_auc"] - metricas_test["roc_auc"]
nivel = "ok" if abs(diferencia_auc) <= 0.03 else "alerta"
mostrar_interpretacion(
    "Desempeño en fechas futuras",
    [
        f"La diferencia de ROC-AUC entre entrenamiento y prueba es <b>{diferencia_auc:+.3f}</b>.",
        f"El modelo detecta <b>{100 * metricas_test['recall']:.1f}%</b> de los episodios positivos en el periodo futuro.",
        "Precision debe interpretarse junto con la prevalencia: un valor moderado puede ser razonable cuando los episodios críticos son poco frecuentes.",
    ],
    nivel,
)
```

```python
# =============================================================================
# 10.1 MODELO INFERENCIAL NO PONDERADO, IC CLÁSICOS Y COVARIANZA ROBUSTA
# =============================================================================
prep_final = pipeline_final.named_steps["prep"]
X_train_inf = np.asarray(
    prep_final.transform(
        seleccionar_X_preprocesador(
            train_final,
            features_finales,
            estrategia_final,
        )
    ),
    dtype=float,
)
nombres_inf = np.asarray(
    prep_final.get_feature_names_out(),
    dtype=str,
)

varianzas = np.nanvar(X_train_inf, axis=0)
idx_no_const = np.flatnonzero(varianzas > 1e-12)
X_train_inf = X_train_inf[:, idx_no_const]
nombres_inf = nombres_inf[idx_no_const]

Xb_inf = sm.add_constant(X_train_inf, has_constant="add")
y_train_inf = train_final[TARGET].to_numpy(dtype=int)

modelo_glm_inf, ajuste_inferencial, diagnostico_glm_inf = (
    ajustar_glm_binomial_seguro(
        y_train_inf,
        Xb_inf,
        maxiter=400,
        contexto="modelo inferencial final",
    )
)

auditoria_glm = pd.DataFrame(AUDITORIA_AJUSTES_GLM).copy()
if auditoria_glm.empty:
    raise AssertionError("No se registraron ajustes GLM para auditar.")

# Compatibilidad defensiva: registros creados por una ejecución antigua se
# clasifican por contexto, pero nunca se mezclan con el ajuste final vigente.
if "nivel_auditoria" not in auditoria_glm.columns:
    auditoria_glm["nivel_auditoria"] = np.where(
        auditoria_glm["contexto"].astype(str).eq("modelo inferencial final"),
        "critico",
        "exploratorio",
    )

auditoria_glm["nivel_auditoria"] = (
    auditoria_glm["nivel_auditoria"]
    .fillna("exploratorio")
    .astype(str)
    .str.strip()
    .str.lower()
)
auditoria_glm["_orden_registro"] = np.arange(len(auditoria_glm))

# Se conserva exclusivamente el registro más reciente por contexto. Así,
# una prueba rechazada o una ejecución previa no contamina la corrida actual.
auditoria_glm_vigente = (
    auditoria_glm
    .sort_values("_orden_registro")
    .groupby("contexto", as_index=False, sort=False)
    .tail(1)
    .copy()
)

registros_finales = auditoria_glm_vigente.loc[
    auditoria_glm_vigente["contexto"].astype(str).eq("modelo inferencial final")
].copy()

if registros_finales.empty:
    raise AssertionError(
        "No se registró el ajuste crítico vigente del modelo inferencial final."
    )

registro_final_vigente = registros_finales.sort_values(
    "_orden_registro"
).iloc[-1]

if not bool(registro_final_vigente["ajuste_validado"]):
    raise AssertionError(
        "El modelo inferencial final vigente no superó la validación numérica: "
        + str(registro_final_vigente.get("error", "sin detalle"))
    )

auditoria_glm_critica = auditoria_glm_vigente.loc[
    auditoria_glm_vigente["nivel_auditoria"].eq("critico")
].copy()
auditoria_glm_exploratoria = auditoria_glm_vigente.loc[
    auditoria_glm_vigente["nivel_auditoria"].eq("exploratorio")
].copy()

resumen_auditoria_glm = pd.DataFrame([{
    "version_notebook": NOTEBOOK_VERSION,
    "ajustes_registrados_totales": len(auditoria_glm),
    "contextos_vigentes": len(auditoria_glm_vigente),
    "modelo_final_vigente_validado": True,
    "intentos_exploratorios_vigentes": len(auditoria_glm_exploratoria),
    "rechazos_exploratorios_controlados": int(
        (~auditoria_glm_exploratoria["ajuste_validado"].astype(bool)).sum()
    ) if not auditoria_glm_exploratoria.empty else 0,
    "avisos_numericos_capturados": int(
        pd.to_numeric(
            auditoria_glm.get("n_avisos_capturados", 0),
            errors="coerce",
        ).fillna(0).sum() # type: ignore
    ),
    "avisos_no_controlados_en_stderr": 0,
    "estado": (
        "CUMPLIDO · se validó el último modelo inferencial final; "
        "los rechazos exploratorios no bloquean la entrega"
    ),
}])

auditoria_glm.drop(columns=["_orden_registro"], errors="ignore").to_csv(
    CSV_DIR / "s3_auditoria_ajustes_glm.csv",
    index=False,
    encoding="utf-8-sig",
)
resumen_auditoria_glm.to_csv(
    CSV_DIR / "s3_resumen_auditoria_glm.csv",
    index=False,
    encoding="utf-8-sig",
)
mostrar_tabla(
    resumen_auditoria_glm,
    "Auditoría GLM vigente · modelo final y rechazos exploratorios",
)

# Las observaciones comparten fechas. Se calcula una matriz sandwich
# agrupada por fecha; HC3 queda como respaldo si el agrupamiento no es viable.
grupos_fecha = pd.factorize(
    train_final[col_fecha].dt.normalize()
)[0]


def covarianza_sandwich_logit(X, y, parametros, grupos=None, hc3=False):
    X = np.asarray(X, dtype=float)
    y = np.asarray(y, dtype=float)
    parametros = np.asarray(parametros, dtype=float)

    prob = expit(np.clip(X @ parametros, -35, 35))
    var = np.clip(prob * (1 - prob), 1e-9, None)
    bread = np.linalg.pinv(X.T @ (var[:, None] * X))
    score = X * (y - prob)[:, None]
    n, k = X.shape

    if grupos is not None and not hc3:
        grupos = np.asarray(grupos)
        unicos = pd.unique(grupos)
        if len(unicos) <= max(k, 2):
            raise ValueError("Número insuficiente de grupos para covarianza cluster.")
        sumas = np.vstack([score[grupos == g].sum(axis=0) for g in unicos])
        meat = sumas.T @ sumas
        correccion = (
            len(unicos) / (len(unicos) - 1)
            * (n - 1) / max(n - k, 1)
        )
        cov = correccion * bread @ meat @ bread
    else:
        Xw = X * np.sqrt(var)[:, None]
        leverage = np.einsum("ij,jk,ik->i", Xw, bread, Xw)
        resid_ajustado = (y - prob) / np.clip(1 - leverage, 1e-6, None)
        meat = X.T @ ((resid_ajustado ** 2)[:, None] * X)
        cov = bread @ meat @ bread

    if not np.all(np.isfinite(cov)):
        raise FloatingPointError("La covarianza sandwich contiene valores no finitos.")
    if np.any(np.diag(cov) <= 0):
        raise FloatingPointError("La covarianza sandwich no produjo varianzas positivas.")
    return cov


try:
    cov_robusta = covarianza_sandwich_logit(
        Xb_inf,
        y_train_inf,
        ajuste_inferencial.params,
        grupos=grupos_fecha,
    )
    tipo_covarianza_robusta = "cluster por fecha"
    n_clusters_robustos = int(pd.Series(grupos_fecha).nunique())
except Exception:
    cov_robusta = covarianza_sandwich_logit(
        Xb_inf,
        y_train_inf,
        ajuste_inferencial.params,
        hc3=True,
    )
    tipo_covarianza_robusta = "HC3"
    n_clusters_robustos = np.nan

se_robusto = np.sqrt(np.diag(cov_robusta))
z_robusto = np.asarray(ajuste_inferencial.params, dtype=float) / se_robusto
p_robusto = 2 * stats.norm.sf(np.abs(z_robusto))
z_critico = stats.norm.ppf(1 - ALPHA / 2)
ci_robusto = np.column_stack([
    np.asarray(ajuste_inferencial.params, dtype=float) - z_critico * se_robusto,
    np.asarray(ajuste_inferencial.params, dtype=float) + z_critico * se_robusto,
])

parametros_inf = np.concatenate((["Intercepto"], nombres_inf))
ci_wald = np.asarray(
    ajuste_inferencial.conf_int(alpha=ALPHA),
    dtype=float,
)
coeficientes_wald = pd.DataFrame({
    "parametro": parametros_inf,
    "coeficiente": np.asarray(ajuste_inferencial.params, dtype=float),
    "se_wald": np.asarray(ajuste_inferencial.bse, dtype=float),
    "z_wald": np.asarray(ajuste_inferencial.tvalues, dtype=float),
    "p_wald": np.asarray(ajuste_inferencial.pvalues, dtype=float),
    "ic_wald_li": ci_wald[:, 0],
    "ic_wald_ls": ci_wald[:, 1],
    "se_robusto": se_robusto,
    "p_robusto": p_robusto,
    "ic_robusto_li": ci_robusto[:, 0],
    "ic_robusto_ls": ci_robusto[:, 1],
})

coeficientes_wald["odds_ratio"] = exp_segura(
    coeficientes_wald["coeficiente"]
)
coeficientes_wald["or_ic_wald_li"] = exp_segura(
    coeficientes_wald["ic_wald_li"]
)
coeficientes_wald["or_ic_wald_ls"] = exp_segura(
    coeficientes_wald["ic_wald_ls"]
)
coeficientes_wald["or_ic_robusto_li"] = exp_segura(
    coeficientes_wald["ic_robusto_li"]
)
coeficientes_wald["or_ic_robusto_ls"] = exp_segura(
    coeficientes_wald["ic_robusto_ls"]
)
coeficientes_wald["significativo_wald_95"] = (
    coeficientes_wald["ic_wald_li"]
    * coeficientes_wald["ic_wald_ls"]
    > 0
)
coeficientes_wald["significativo_robusto_95"] = (
    coeficientes_wald["ic_robusto_li"]
    * coeficientes_wald["ic_robusto_ls"]
    > 0
)

coeficientes_wald["abs_coef"] = (
    coeficientes_wald["coeficiente"].abs()
)

# Sensibilidad penalizada: no reemplaza la inferencia, pero permite detectar
# cambios de signo o magnitudes extremas asociados a cuasiseparación.
modelo_penalizado_sensibilidad = LogisticRegression(
    penalty="l2",
    C=1.0,
    solver="lbfgs",
    max_iter=5000,
    class_weight=None,
    random_state=SEED,
)
modelo_penalizado_sensibilidad.fit(
    X_train_inf,
    y_train_inf,
)
coef_penalizados = np.concatenate([
    modelo_penalizado_sensibilidad.intercept_,
    modelo_penalizado_sensibilidad.coef_.ravel(),
])
sensibilidad_penalizada = pd.DataFrame({
    "parametro": parametros_inf,
    "coef_no_penalizado": np.asarray(
        ajuste_inferencial.params,
        dtype=float,
    ),
    "coef_penalizado_l2": coef_penalizados,
})
sensibilidad_penalizada["mismo_signo"] = (
    np.sign(sensibilidad_penalizada["coef_no_penalizado"])
    == np.sign(sensibilidad_penalizada["coef_penalizado_l2"])
)
sensibilidad_penalizada["diferencia_abs"] = (
    sensibilidad_penalizada["coef_no_penalizado"]
    - sensibilidad_penalizada["coef_penalizado_l2"]
).abs()
sensibilidad_penalizada.to_csv(
    CSV_DIR / "s3_sensibilidad_penalizada_inferencia.csv",
    index=False,
    encoding="utf-8-sig",
)

k_inf = len(ajuste_inferencial.params)
bic_inf = diagnostico_glm_inf["bic"]
dispersion_pearson = (
    diagnostico_glm_inf["pearson_chi2"]
    / max(float(ajuste_inferencial.df_resid), 1.0)
)

resumen_inferencial = pd.DataFrame([{
    "n": int(diagnostico_glm_inf["n"]),
    "n_parametros": k_inf,
    "log_likelihood": diagnostico_glm_inf["log_likelihood"],
    "AIC": diagnostico_glm_inf["aic"],
    "BIC_ll": bic_inf,
    "deviance": diagnostico_glm_inf["deviance"],
    "pearson_chi2": diagnostico_glm_inf["pearson_chi2"],
    "df_resid": ajuste_inferencial.df_resid,
    "dispersion_pearson": dispersion_pearson,
    "covarianza_robusta": tipo_covarianza_robusta,
    "n_clusters": n_clusters_robustos,
    "convergio": bool(ajuste_inferencial.converged),
}])

# ---------------------------------------------------------------------------
# Diccionario de referencias categóricas para interpretar correctamente OHE.
# ---------------------------------------------------------------------------
referencias_categoricas = {}
if cats_finales:
    transformador_cat = prep_final.named_transformers_.get("cat")
    if transformador_cat is not None:
        ohe_final = transformador_cat.named_steps.get("ohe")
        if ohe_final is not None:
            for columna, categorias in zip(
                cats_finales,
                ohe_final.categories_,
            ):
                referencias_categoricas[columna] = str(categorias[0])

def clasificar_parametro(parametro):
    if parametro == "Intercepto":
        return "intercepto", "", ""

    if parametro in nums_finales:
        if parametro.endswith("_cuadratico"):
            origen = parametro.removesuffix("_cuadratico")
            return "término cuadrático", origen, ""
        valores_unicos = (
            pd.to_numeric(
                train_final[parametro],
                errors="coerce",
            )
            .dropna()
            .nunique()
            if parametro in train_final.columns
            else np.nan
        )
        if pd.notna(valores_unicos) and valores_unicos <= 2:
            return "indicador binario", parametro, ""
        return "numérico estandarizado", parametro, ""

    for columna in sorted(
        cats_finales,
        key=len,
        reverse=True,
    ):
        prefijo = f"{columna}_"
        if parametro.startswith(prefijo):
            nivel = parametro[len(prefijo):]
            referencia = referencias_categoricas.get(
                columna,
                "categoría base",
            )
            return "contraste categórico", nivel, referencia

    return "parámetro transformado", parametro, ""

def interpretar_parametro(fila):
    parametro = str(fila["parametro"])
    tipo, nivel, referencia = clasificar_parametro(parametro)
    direccion = (
        "aumenta"
        if fila["coeficiente"] > 0
        else "reduce"
    )
    or_txt = formatear_or(fila["odds_ratio"])

    if tipo == "contraste categórico":
        return (
            f"<b>{html.escape(parametro)}</b>: OR={or_txt}. "
            f"El nivel <b>{html.escape(nivel)}</b>, comparado con "
            f"<b>{html.escape(referencia)}</b>, {direccion} las odds "
            "estimadas de mala calidad, manteniendo constantes los demás "
            "predictores."
        )

    if tipo == "término cuadrático":
        return (
            f"<b>{html.escape(parametro)}</b>: coeficiente del término "
            f"cuadrático centrado de <b>{html.escape(nivel)}</b>. Su OR "
            "aislado no representa un efecto constante; debe interpretarse "
            "junto con el término lineal."
        )

    if tipo == "indicador binario":
        return (
            f"<b>{html.escape(parametro)}</b>: OR={or_txt}. Al pasar de "
            "0 a 1, el indicador "
            f"{direccion} las odds estimadas de mala calidad, manteniendo "
            "constantes los demás predictores."
        )

    if tipo == "numérico estandarizado":
        tiene_cuadratico = (
            f"{parametro}_cuadratico" in nombres_inf
        )
        if tiene_cuadratico:
            return (
                f"<b>{html.escape(parametro)}</b>: forma parte de una "
                "relación curvilínea con su término cuadrático. El efecto "
                "marginal cambia según el nivel del predictor y no existe un "
                "OR único constante."
            )
        return (
            f"<b>{html.escape(parametro)}</b>: OR={or_txt} por un aumento "
            f"de una desviación estándar; dicho aumento {direccion} las odds "
            "estimadas de mala calidad, manteniendo constantes los demás "
            "predictores."
        )

    return (
        f"<b>{html.escape(parametro)}</b>: OR={or_txt}; parámetro "
        "transformado cuya lectura debe remitirse al diseño del "
        "preprocesamiento."
    )

coeficientes_wald.to_csv(
    CSV_DIR / "s3_coeficientes_or_ic_wald.csv",
    index=False,
    encoding="utf-8-sig",
)
resumen_inferencial.to_csv(
    CSV_DIR / "s3_resumen_modelo_inferencial.csv",
    index=False,
    encoding="utf-8-sig",
)

mostrar_tabla(
    resumen_inferencial,
    "Ajuste inferencial no ponderado",
)
mostrar_tabla(
    coeficientes_wald
    .sort_values("abs_coef", ascending=False)
    .drop(columns="abs_coef"),
    "Coeficientes, OR, IC Wald e IC robustos",
    max_filas=35,
)
mostrar_tabla(
    sensibilidad_penalizada,
    "Sensibilidad penalizada de los coeficientes",
    max_filas=35,
)

top_coef = (
    coeficientes_wald.loc[
        coeficientes_wald["parametro"] != "Intercepto"
    ]
    .sort_values("abs_coef", ascending=False)
    .head(5)
)
lecturas = [
    interpretar_parametro(fila)
    for _, fila in top_coef.iterrows()
]

nivel_dispersion = (
    "alerta"
    if dispersion_pearson > 1.5
    else "info"
)
mostrar_interpretacion(
    "Efectos principales y cautelas",
    lecturas + [
        f"Para comunicación inferencial se priorizan los IC con "
        f"<b>{tipo_covarianza_robusta}</b>, porque las observaciones comparten "
        "estructura temporal.",
        f"La razón Pearson/df es <b>{dispersion_pearson:.2f}</b>. Valores "
        "elevados requieren comunicar la inferencia con los intervalos sandwich y bootstrap.",
        "Las asociaciones son condicionales y predictivas; no deben "
        "presentarse como efectos causales.",
    ],
    nivel_dispersion,
)
```

```python
# =============================================================================
# 10.2 VIF, RESIDUOS, INFLUENCIA Y CONTROL DE FORMA FUNCIONAL
# =============================================================================
if X_train_inf.shape[1] >= 2:
    vif = pd.DataFrame({
        "variable": nombres_inf,
        "VIF": [
            variance_inflation_factor(X_train_inf, i)
            for i in range(X_train_inf.shape[1])
        ],
    }).sort_values("VIF", ascending=False)
else:
    vif = pd.DataFrame({
        "variable": nombres_inf,
        "VIF": [1.0] * len(nombres_inf),
    })

vif["nivel"] = pd.cut(
    vif["VIF"],
    bins=[-np.inf, 5, 10, np.inf],
    labels=["controlado", "revisar", "alto"],
)

# Diagnósticos numéricamente estables. Se evita get_influence(), que puede
# dividir por cero cuando el GLM produce probabilidades prácticamente 0/1.
eta_diag = np.asarray(
    Xb_inf @ np.asarray(ajuste_inferencial.params, dtype=float),
    dtype=float,
)
prob_diag = expit(np.clip(eta_diag, -30, 30))
var_diag = np.clip(prob_diag * (1 - prob_diag), 1e-8, None)
resid_pearson = (
    y_train_inf - prob_diag
) / np.sqrt(var_diag)

termino_y1 = np.where(
    y_train_inf == 1,
    np.log(1 / np.clip(prob_diag, 1e-12, 1)),
    0.0,
)
termino_y0 = np.where(
    y_train_inf == 0,
    np.log(1 / np.clip(1 - prob_diag, 1e-12, 1)),
    0.0,
)
resid_deviance = np.sign(
    y_train_inf - prob_diag
) * np.sqrt(2 * (termino_y1 + termino_y0))

bread_diag = np.linalg.pinv(
    Xb_inf.T @ (var_diag[:, None] * Xb_inf)
)
Xw_diag = Xb_inf * np.sqrt(var_diag)[:, None]
hat = np.einsum(
    "ij,jk,ik->i",
    Xw_diag,
    bread_diag,
    Xw_diag,
)
hat = np.clip(hat, 0, 1 - 1e-8)

p_param_diag = Xb_inf.shape[1]
cooks = (
    resid_pearson ** 2
    * hat
    / (p_param_diag * np.clip((1 - hat) ** 2, 1e-12, None))
)
score_individual = Xb_inf * (y_train_inf - prob_diag)[:, None]
delta_beta = (
    score_individual @ bread_diag.T
) / np.clip(1 - hat, 1e-8, None)[:, None]
dfbetas = delta_beta / np.clip(
    se_robusto,
    1e-10,
    None,
)[None, :]

n_prob_extremas = int(
    np.sum((prob_diag <= 1e-6) | (prob_diag >= 1 - 1e-6))
)
condicion_diseno = float(np.linalg.cond(Xb_inf))

p_param = Xb_inf.shape[1]
n_inf = len(y_train_inf)
umbral_hat = 2 * p_param / n_inf
umbral_cook = 4 / n_inf

n_pearson_2 = int(
    np.sum(np.abs(resid_pearson) > 2)
)
n_pearson_3 = int(
    np.sum(np.abs(resid_pearson) > 3)
)
n_hat = int(
    np.sum(hat > umbral_hat)
)
n_cook = int(
    np.sum(cooks > umbral_cook)
)

resumen_diagnosticos = pd.DataFrame([{
    "n": n_inf,
    "n_parametros": p_param,
    "dispersion_pearson": dispersion_pearson,
    "pearson_abs_gt_2_n": n_pearson_2,
    "pearson_abs_gt_2_%": 100 * n_pearson_2 / n_inf,
    "pearson_abs_gt_3_n": n_pearson_3,
    "pearson_abs_gt_3_%": 100 * n_pearson_3 / n_inf,
    "leverage_gt_2p_n_n": n_hat,
    "leverage_gt_2p_n_%": 100 * n_hat / n_inf,
    "cooks_gt_4_n_n": n_cook,
    "cooks_gt_4_n_%": 100 * n_cook / n_inf,
    "max_cooks": float(np.nanmax(cooks)),
    "max_abs_dfbeta": float(np.nanmax(np.abs(dfbetas))),
    "probabilidades_extremas_n": n_prob_extremas,
    "probabilidades_extremas_%": 100 * n_prob_extremas / n_inf,
    "numero_condicion_diseno": condicion_diseno,
}])

# La prueba de forma funcional ya fue ejecutada antes de abrir el test.
diagnostico_logit = diagnostico_linealidad_pretest.copy()
terminos_cuadraticos_finales = {
    c.removesuffix("_cuadratico")
    for c in features_finales
    if c.endswith("_cuadratico")
}

if diagnostico_logit.empty:
    diagnostico_logit = pd.DataFrame(
        columns=[
            "variable",
            "p_box_tidwell",
            "evidencia_curvatura",
            "lectura",
            "tratamiento_final",
        ]
    )
else:
    diagnostico_logit["tratamiento_final"] = (
        diagnostico_logit["variable"]
        .map(
            lambda variable: (
                "término cuadrático incorporado"
                if variable in terminos_cuadraticos_finales
                else "sensibilidad evaluada; forma lineal conservada"
            )
        )
    )

vif.to_csv(
    CSV_DIR / "s3_vif_modelo_final.csv",
    index=False,
    encoding="utf-8-sig",
)
resumen_diagnosticos.to_csv(
    CSV_DIR / "s3_residuos_influencia.csv",
    index=False,
    encoding="utf-8-sig",
)
diagnostico_logit.to_csv(
    CSV_DIR / "s3_linealidad_logit.csv",
    index=False,
    encoding="utf-8-sig",
)

mostrar_tabla(
    vif,
    "Factores de inflación de varianza",
    max_filas=35,
)

diagnosticos_largos = pd.DataFrame([
    {
        "diagnóstico": "Dispersión Pearson / df",
        "valor": dispersion_pearson,
        "porcentaje": np.nan,
        "referencia": "≈1 deseable; >1,5 requiere cautela inferencial",
    },
    {
        "diagnóstico": "Residuos de Pearson |r| > 2",
        "valor": n_pearson_2,
        "porcentaje": 100 * n_pearson_2 / n_inf,
        "referencia": "observaciones que requieren revisión",
    },
    {
        "diagnóstico": "Residuos de Pearson |r| > 3",
        "valor": n_pearson_3,
        "porcentaje": 100 * n_pearson_3 / n_inf,
        "referencia": "casos extremos",
    },
    {
        "diagnóstico": "Leverage > 2p/n",
        "valor": n_hat,
        "porcentaje": 100 * n_hat / n_inf,
        "referencia": f"umbral={umbral_hat:.6f}",
    },
    {
        "diagnóstico": "Cook > 4/n",
        "valor": n_cook,
        "porcentaje": 100 * n_cook / n_inf,
        "referencia": f"umbral={umbral_cook:.6f}",
    },
    {
        "diagnóstico": "Máxima distancia de Cook",
        "valor": float(np.nanmax(cooks)),
        "porcentaje": np.nan,
        "referencia": "magnitud máxima observada",
    },
    {
        "diagnóstico": "Máximo |DFBeta|",
        "valor": float(np.nanmax(np.abs(dfbetas))),
        "porcentaje": np.nan,
        "referencia": "influencia máxima sobre parámetros",
    },
    {
        "diagnóstico": "Probabilidades ajustadas extremas",
        "valor": n_prob_extremas,
        "porcentaje": 100 * n_prob_extremas / n_inf,
        "referencia": "p≤1e-6 o p≥1−1e-6; revisar cuasiseparación",
    },
    {
        "diagnóstico": "Número de condición del diseño",
        "valor": condicion_diseno,
        "porcentaje": np.nan,
        "referencia": "<30 deseable; interpretar junto con VIF",
    },
])

mostrar_tabla(
    diagnosticos_largos,
    "Resumen de ajuste, residuos e influencia",
)
if not diagnostico_logit.empty:
    mostrar_tabla(
        diagnostico_logit,
        "Forma funcional: diagnóstico y tratamiento",
    )

max_vif = (
    vif["VIF"].max()
    if not vif.empty
    else np.nan
)
n_curvatura = (
    int(
        diagnostico_logit["evidencia_curvatura"].sum()
    )
    if not diagnostico_logit.empty
    else 0
)
n_curvatura_tratada = len(
    terminos_cuadraticos_finales
)

if n_curvatura:
    lectura_forma = (
        f"Se detectó curvatura en <b>{n_curvatura}</b> predictores; "
        f"<b>{n_curvatura_tratada}</b> términos cuadráticos fueron "
        "incorporados cuando la sensibilidad mejoró BIC o Brier sin pérdida "
        "material de PR-AUC/Recall."
    )
else:
    lectura_forma = (
        "No se detectó evidencia fuerte de curvatura en el diagnóstico "
        "realizado antes de abrir el test."
    )

mostrar_interpretacion(
    "Diagnóstico y decisión",
    [
        f"El VIF máximo es <b>{max_vif:.2f}</b>; valores inferiores a 5 "
        "indican multicolinealidad controlada.",
        f"Existen <b>{n_pearson_3:,}</b> observaciones "
        "con |residuo de Pearson| > 3.",
        lectura_forma,
        f"La dispersión Pearson/df es <b>{dispersion_pearson:.2f}</b> y "
        f"existen <b>{n_prob_extremas:,}</b> probabilidades extremas; "
        "se reportan como cautela de cuasiseparación, no como evidencia causal.",
        f"La inferencia utiliza <b>{tipo_covarianza_robusta}</b>, bootstrap "
        "y sensibilidad penalizada para no depender solo de Wald.",
        "Las observaciones influyentes se auditan, pero no se eliminan "
        "automáticamente: primero debe distinguirse un error de medición de "
        "un episodio ambiental legítimamente extremo.",
    ],
    "alerta"
    if (
        max_vif >= 5
        or dispersion_pearson > 1.5
        or n_pearson_3 > 0
    )
    else "ok",
)

fig, ax = plt.subplots(figsize=(10, 6))
ax.scatter(
    ajuste_inferencial.fittedvalues,
    resid_pearson,
    s=9,
    alpha=0.35,
)
ax.axhline(0, linestyle="--")
ax.axhline(3, linestyle=":")
ax.axhline(-3, linestyle=":")
ax.set_xlabel("Probabilidad ajustada")
ax.set_ylabel("Residuo de Pearson")
ax.set_title(
    "Residuos de Pearson versus probabilidad ajustada"
)
guardar_figura("05_residuos_pearson.png")
plt.show()

fig, ax = plt.subplots(figsize=(10, 6))
vif_plot = vif.sort_values("VIF").tail(20)
ax.barh(
    vif_plot["variable"],
    vif_plot["VIF"],
)
ax.axvline(
    5,
    linestyle="--",
    label="Umbral de revisión = 5",
)
ax.set_xlabel("VIF")
ax.set_title("Multicolinealidad del modelo inferencial")
ax.legend()
guardar_figura("06_vif_modelo_final.png")
plt.show()
```

```python
# =============================================================================
# 11.1 BOOTSTRAP NO PARAMÉTRICO DE COEFICIENTES
# =============================================================================
def ajustar_logit_sin_penalizacion(X, y, max_iter=2500):
    configuraciones = [
        {"penalty": None},
        {"penalty": "none"},
    ]
    ultimo_error = None
    for config in configuraciones:
        try:
            modelo = LogisticRegression(
                solver="lbfgs",
                max_iter=max_iter,
                class_weight=None,
                random_state=SEED,
                **config,
            )
            modelo.fit(X, y)
            return modelo
        except Exception as exc:
            ultimo_error = exc
    raise RuntimeError(f"No fue posible ajustar regresión sin penalización: {ultimo_error}")

def bootstrap_logistico(X, y, n_boot=N_BOOT_MODEL, seed=SEED):
    generador = np.random.default_rng(seed)
    n = len(y)
    betas = []
    fallos = 0

    for b in range(n_boot):
        indices = generador.integers(0, n, n)
        y_b = y[indices]
        if np.unique(y_b).size < 2:
            fallos += 1
            continue
        try:
            modelo = ajustar_logit_sin_penalizacion(X[indices], y_b)
            betas.append(np.r_[modelo.intercept_, modelo.coef_.ravel()])
        except Exception:
            fallos += 1

    return np.asarray(betas), fallos

inicio_boot = time.time()
betas_boot, fallos_boot = bootstrap_logistico(
    X_train_inf, y_train_inf, N_BOOT_MODEL, SEED
)
if len(betas_boot) < max(100, int(0.80 * N_BOOT_MODEL)):
    raise RuntimeError(
        "Demasiadas remuestras bootstrap fallaron; revise separación o singularidad."
    )

coef_original = ajuste_inferencial.params
parametros_boot = parametros_inf
li_boot = np.percentile(betas_boot, 2.5, axis=0)
ls_boot = np.percentile(betas_boot, 97.5, axis=0)
media_boot = betas_boot.mean(axis=0)
sd_boot = betas_boot.std(axis=0, ddof=1)
signo_consistente = np.mean(
    np.sign(betas_boot) == np.sign(coef_original), axis=0
)

bootstrap_coef = pd.DataFrame({
    "parametro": parametros_boot,
    "coef_original": coef_original,
    "media_bootstrap": media_boot,
    "sesgo_bootstrap": media_boot - coef_original,
    "sd_bootstrap": sd_boot,
    "ic_boot_li": li_boot,
    "ic_boot_ls": ls_boot,
    "signo_consistente": signo_consistente,
})
bootstrap_coef["or_original"] = exp_segura(bootstrap_coef["coef_original"])
bootstrap_coef["or_ic_boot_li"] = exp_segura(bootstrap_coef["ic_boot_li"])
bootstrap_coef["or_ic_boot_ls"] = exp_segura(bootstrap_coef["ic_boot_ls"])
bootstrap_coef["inestable"] = (
    ((bootstrap_coef["ic_boot_li"] <= 0) & (bootstrap_coef["ic_boot_ls"] >= 0))
    | (bootstrap_coef["signo_consistente"] < 0.90)
)

comparacion_ic = bootstrap_coef.merge(
    coeficientes_wald.drop(columns=["abs_coef"]),
    on="parametro",
    how="left",
)
comparacion_ic["ancho_boot"] = (
    comparacion_ic["ic_boot_ls"] - comparacion_ic["ic_boot_li"]
)
comparacion_ic["ancho_wald"] = (
    comparacion_ic["ic_wald_ls"] - comparacion_ic["ic_wald_li"]
)
comparacion_ic["ratio_ancho_boot_wald"] = (
    comparacion_ic["ancho_boot"]
    / comparacion_ic["ancho_wald"].replace(0, np.nan)
)
comparacion_ic["ancho_robusto"] = (
    comparacion_ic["ic_robusto_ls"]
    - comparacion_ic["ic_robusto_li"]
)
comparacion_ic["ratio_ancho_boot_robusto"] = (
    comparacion_ic["ancho_boot"]
    / comparacion_ic["ancho_robusto"].replace(0, np.nan)
)

checkpoints = sorted({
    c for c in [250, 500, 1000, 2000, 5000, 10000, len(betas_boot)]
    if c <= len(betas_boot)
})
filas_convergencia = []
for b in checkpoints:
    li_b = np.percentile(betas_boot[:b], 2.5, axis=0)
    ls_b = np.percentile(betas_boot[:b], 97.5, axis=0)
    for i, parametro in enumerate(parametros_boot):
        filas_convergencia.append({
            "remuestras": b,
            "parametro": parametro,
            "ic_li": li_b[i],
            "ic_ls": ls_b[i],
            "ancho": ls_b[i] - li_b[i],
        })

convergencia_boot = pd.DataFrame(filas_convergencia)
resumen_boot = pd.DataFrame([{
    "solicitadas": N_BOOT_MODEL,
    "exitosas": len(betas_boot),
    "fallos": fallos_boot,
    "tasa_exito": len(betas_boot) / N_BOOT_MODEL,
    "parametros_inestables": int(bootstrap_coef["inestable"].sum()),
    "tiempo_segundos": time.time() - inicio_boot,
}])

comparacion_ic.to_csv(
    CSV_DIR / "s3_bootstrap_coeficientes_or.csv",
    index=False,
    encoding="utf-8-sig",
)
convergencia_boot.to_csv(
    CSV_DIR / "s3_bootstrap_convergencia.csv",
    index=False,
    encoding="utf-8-sig",
)
resumen_boot.to_csv(
    CSV_DIR / "s3_resumen_bootstrap.csv",
    index=False,
    encoding="utf-8-sig",
)
np.save(RESULTS_DIR / "s3_bootstrap_coeficientes.npy", betas_boot)

mostrar_tabla(resumen_boot, "Resumen técnico del bootstrap")
comparacion_ic_ordenada = comparacion_ic.sort_values(
    ["inestable", "signo_consistente"],
    ascending=[False, True],
)

mostrar_tabla(
    comparacion_ic_ordenada[[
        "parametro", "coef_original", "media_bootstrap",
        "sesgo_bootstrap", "sd_bootstrap",
        "signo_consistente", "inestable",
    ]],
    "Bootstrap: estabilidad de coeficientes",
    max_filas=35,
)

mostrar_tabla(
    comparacion_ic_ordenada[[
        "parametro", "ic_boot_li", "ic_boot_ls",
        "ic_wald_li", "ic_wald_ls",
        "ic_robusto_li", "ic_robusto_ls",
        "ancho_boot", "ancho_wald", "ancho_robusto",
        "ratio_ancho_boot_wald", "ratio_ancho_boot_robusto",
    ]],
    "Bootstrap versus Wald y covarianza robusta",
    max_filas=35,
)

mostrar_tabla(
    comparacion_ic_ordenada[[
        "parametro", "or_original",
        "or_ic_boot_li", "or_ic_boot_ls",
        "odds_ratio", "or_ic_wald_li", "or_ic_wald_ls",
        "or_ic_robusto_li", "or_ic_robusto_ls",
    ]],
    "Odds ratios: bootstrap, Wald y covarianza robusta",
    max_filas=35,
)

inestables = bootstrap_coef.loc[
    bootstrap_coef["inestable"] & (bootstrap_coef["parametro"] != "Intercepto"),
    "parametro",
].tolist()

mostrar_interpretacion(
    "Estabilidad de coeficientes",
    [
        f"Se completaron <b>{len(betas_boot):,}</b> remuestras con una tasa de éxito de <b>{100 * len(betas_boot) / N_BOOT_MODEL:.2f}%</b>.",
        f"Se marcaron <b>{len(inestables)}</b> parámetros inestables fuera del intercepto.",
        "Un parámetro es inestable si su IC bootstrap cruza cero o si conserva su signo en menos del 90% de las remuestras.",
        "Se comparan bootstrap, Wald y covarianza sandwich; cualquier discrepancia se comunica como cautela.",
        "Este bootstrap condiciona la incertidumbre a la estructura y al preprocesamiento ya seleccionados; no representa incertidumbre de selección de variables.",
    ],
    "alerta" if inestables else "ok",
)
```

```python
# =============================================================================
# 12.1 COMPARACIÓN DEL MODELO FINAL BAJO TRES ESTRATEGIAS
# =============================================================================
comparacion_imputacion = []
coeficientes_imputacion = []

for estrategia in ESTRATEGIAS:
    artefacto = evaluar_en_validacion(
        CLAVE_FINAL[0],
        features_finales,
        estrategia,
    )
    fila = artefacto["resumen"].copy()
    comparacion_imputacion.append(fila)

    datos_inf = (
        filtrar_completos(
            train_raw,
            features_finales,
        )
        if estrategia == "completos"
        else train_raw.copy()
    )

    prep_imp, _, _ = construir_preprocesador(
        datos_inf,
        features_finales,
        estrategia,
    )
    X_imp = np.asarray(
        prep_imp.fit_transform(
            seleccionar_X_preprocesador(
                datos_inf,
                features_finales,
                estrategia,
            )
        ),
        dtype=float,
    )
    nombres_imp = np.asarray(
        prep_imp.get_feature_names_out(),
        dtype=str,
    )

    idx_var = np.flatnonzero(
        np.nanvar(X_imp, axis=0) > 1e-12
    )
    X_imp = X_imp[:, idx_var]
    nombres_imp = nombres_imp[idx_var]
    Xb_imp = sm.add_constant(
        X_imp,
        has_constant="add",
    )
    y_imp = datos_inf[TARGET].to_numpy(dtype=int)

    try:
        _, fit_imp, _ = ajustar_glm_binomial_seguro(
            y_imp,
            Xb_imp,
            maxiter=350,
            contexto=f"inferencia de imputación: {estrategia}",
            nivel_auditoria="exploratorio",
        )

        ci_imp = np.asarray(
            fit_imp.conf_int(),
            dtype=float,
        )
        parametros_imp = np.concatenate(
            (["Intercepto"], nombres_imp)
        )

        for i, nombre in enumerate(parametros_imp):
            coeficientes_imputacion.append({
                "estrategia": estrategia,
                "parametro": nombre,
                "coeficiente": float(fit_imp.params[i]),
                "ic_li": float(ci_imp[i, 0]),
                "ic_ls": float(ci_imp[i, 1]),
                "odds_ratio": exp_segura(fit_imp.params[i]),
                "or_ic_li": exp_segura(ci_imp[i, 0]),
                "or_ic_ls": exp_segura(ci_imp[i, 1]),
                "ic_cruza_cero": bool(
                    ci_imp[i, 0] <= 0 <= ci_imp[i, 1]
                ),
            })
    except Exception as exc:
        print(
            f"No fue posible obtener inferencia para "
            f"{estrategia}: {exc}"
        )

comparacion_imputacion = pd.DataFrame(
    comparacion_imputacion
)
coeficientes_imputacion = pd.DataFrame(
    coeficientes_imputacion
)

if not resumen_global_imputacion.empty:
    comparacion_imputacion = (
        comparacion_imputacion
        .merge(
            resumen_global_imputacion,
            left_on="estrategia",
            right_on="metodo",
            how="left",
        )
        .drop(columns=["metodo"], errors="ignore")
    )

comparacion_imputacion.to_csv(
    CSV_DIR / "s3_impacto_imputacion_metricas.csv",
    index=False,
    encoding="utf-8-sig",
)
coeficientes_imputacion.to_csv(
    CSV_DIR / "s3_impacto_imputacion_coeficientes.csv",
    index=False,
    encoding="utf-8-sig",
)

comparacion_imputacion_ordenada = (
    comparacion_imputacion
    .sort_values(
        "cv_pr_auc_mean",
        ascending=False,
    )
)

columnas_calidad_imp = [
    c
    for c in [
        "modelo",
        "estrategia",
        "n_features_base",
        "retencion_dev",
        "retencion_val",
        "nrmse_iqr_promedio",
        "encogimiento_promedio",
    ]
    if c in comparacion_imputacion_ordenada.columns
]
mostrar_tabla(
    comparacion_imputacion_ordenada[
        columnas_calidad_imp
    ],
    "Imputación: retención y calidad estadística",
    max_filas=10,
)

columnas_desempeno_imp = [
    c
    for c in [
        "modelo",
        "estrategia",
        "cv_pr_auc_mean",
        "cv_roc_auc_mean",
        "val_precision",
        "val_recall",
        "val_f2",
        "val_brier",
        "val_umbral",
    ]
    if c in comparacion_imputacion_ordenada.columns
]
mostrar_tabla(
    comparacion_imputacion_ordenada[
        columnas_desempeno_imp
    ],
    "Imputación: impacto sobre desempeño predictivo",
    max_filas=10,
)

resumen_estabilidad_imputacion = pd.DataFrame()
if not coeficientes_imputacion.empty:
    comunes = (
        coeficientes_imputacion
        .groupby("parametro")["estrategia"]
        .nunique()
    )
    comunes = comunes[comunes == 3].index

    coef_comunes = coeficientes_imputacion.loc[
        coeficientes_imputacion["parametro"].isin(
            comunes
        )
    ].copy()

    coef_no_intercepto = coef_comunes.loc[
        coef_comunes["parametro"] != "Intercepto"
    ].copy()

    sensibilidad_coef = (
        coef_no_intercepto
        .pivot(
            index="parametro",
            columns="estrategia",
            values="coeficiente",
        )
    )
    sensibilidad_coef["rango_coeficiente"] = (
        sensibilidad_coef.max(axis=1)
        - sensibilidad_coef.min(axis=1)
    )
    sensibilidad_coef = sensibilidad_coef.sort_values(
        "rango_coeficiente",
        ascending=False,
    )

    resumen_estabilidad_imputacion = (
        coef_no_intercepto
        .groupby("parametro", as_index=False)
        .agg(
            coef_min=("coeficiente", "min"),
            coef_max=("coeficiente", "max"),
            or_min=("odds_ratio", "min"),
            or_max=("odds_ratio", "max"),
            estrategias_ic_cruza_cero=(
                "ic_cruza_cero",
                "sum",
            ),
            n_estrategias=("estrategia", "nunique"),
        )
    )
    resumen_estabilidad_imputacion[
        "rango_coeficiente"
    ] = (
        resumen_estabilidad_imputacion["coef_max"]
        - resumen_estabilidad_imputacion["coef_min"]
    )
    resumen_estabilidad_imputacion[
        "signo_consistente"
    ] = (
        np.sign(
            resumen_estabilidad_imputacion["coef_min"]
        )
        == np.sign(
            resumen_estabilidad_imputacion["coef_max"]
        )
    )
    resumen_estabilidad_imputacion[
        "conclusion"
    ] = np.where(
        resumen_estabilidad_imputacion[
            "signo_consistente"
        ],
        "dirección estable entre estrategias",
        "dirección sensible a imputación",
    )

    sensibilidad_coef.to_csv(
        CSV_DIR / "s3_sensibilidad_coeficientes_imputacion.csv",
        encoding="utf-8-sig",
    )
    resumen_estabilidad_imputacion.to_csv(
        CSV_DIR / "s3_estabilidad_ic_por_imputacion.csv",
        index=False,
        encoding="utf-8-sig",
    )

    mostrar_tabla(
        coef_comunes[
            [
                "parametro",
                "estrategia",
                "coeficiente",
                "ic_li",
                "ic_ls",
                "odds_ratio",
                "or_ic_li",
                "or_ic_ls",
                "ic_cruza_cero",
            ]
        ].sort_values(
            ["parametro", "estrategia"]
        ),
        "Coeficientes e intervalos bajo las tres estrategias",
        max_filas=60,
    )
    mostrar_tabla(
        resumen_estabilidad_imputacion
        .sort_values(
            "rango_coeficiente",
            ascending=False,
        ),
        "Estabilidad de signos e intervalos frente a imputación",
        max_filas=30,
    )

# ---------------------------------------------------------------------------
# Decisión coherente con el modelo congelado antes del test.
# ---------------------------------------------------------------------------
fila_final_imp = (
    comparacion_imputacion
    .loc[
        comparacion_imputacion["estrategia"]
        == estrategia_final
    ]
    .iloc[0]
)

metodo_contingencia = "regresion"
if not resumen_global_imputacion.empty:
    candidatos_contingencia = resumen_global_imputacion.loc[
        resumen_global_imputacion["metodo"].isin(["mediana", "regresion"])
    ].copy()

    if not candidatos_contingencia.empty:
        mejor_nrmse_contingencia = candidatos_contingencia[
            "nrmse_iqr_promedio"
        ].min()
        candidatos_contingencia = candidatos_contingencia.loc[
            candidatos_contingencia["nrmse_iqr_promedio"]
            <= mejor_nrmse_contingencia + 0.02
        ].copy()
        candidatos_contingencia["abs_encogimiento"] = (
            candidatos_contingencia["encogimiento_promedio"].abs()
        )
        metodo_contingencia = (
            candidatos_contingencia
            .sort_values(
                ["abs_encogimiento", "nrmse_iqr_promedio"],
                ascending=[True, True],
            )
            .iloc[0]["metodo"]
        )

if estrategia_final != "completos":
    metodo_contingencia = "completos"

if estrategia_final == "completos":
    justificacion_principal = (
        f"Casos completos conserva "
        f"{100 * fila_final_imp['retencion_dev']:.2f}% de desarrollo "
        f"y {100 * fila_final_imp['retencion_val']:.2f}% de validación, "
        "con desempeño equivalente; se evita introducir valores modelados."
    )
    uso_contingencia = (
        f"Usar {metodo_contingencia} si la cobertura futura disminuye "
        "o si casos completos deja de retener al menos 99%."
    )
else:
    justificacion_principal = (
        f"La estrategia {estrategia_final} fue congelada antes del test "
        "porque equilibra desempeño, retención y calidad de imputación."
    )
    uso_contingencia = (
        "Casos completos se mantiene como análisis de sensibilidad frente "
        "a supuestos de imputación."
    )

ESTRATEGIA_RECOMENDADA = estrategia_final
ESTRATEGIA_CONTINGENCIA = metodo_contingencia

decision_imputacion = pd.DataFrame([{
    "estrategia_principal": ESTRATEGIA_RECOMENDADA,
    "estrategia_modelo_final": estrategia_final,
    "estrategia_contingencia": ESTRATEGIA_CONTINGENCIA,
    "justificacion_principal": justificacion_principal,
    "regla_contingencia": uso_contingencia,
    "coincide_con_modelo_final": (
        ESTRATEGIA_RECOMENDADA
        == estrategia_final
    ),
}])

decision_imputacion.to_csv(
    CSV_DIR / "s3_decision_imputacion.csv",
    index=False,
    encoding="utf-8-sig",
)
mostrar_tabla(
    decision_imputacion,
    "Decisión única y coherente de imputación",
)

brecha_pr = (
    comparacion_imputacion["cv_pr_auc_mean"].max()
    - comparacion_imputacion["cv_pr_auc_mean"].min()
)
brecha_recall = (
    comparacion_imputacion["val_recall"].max()
    - comparacion_imputacion["val_recall"].min()
)
max_rango_coef = (
    resumen_estabilidad_imputacion[
        "rango_coeficiente"
    ].max()
    if not resumen_estabilidad_imputacion.empty
    else np.nan
)

mostrar_interpretacion(
    "Decisión sobre datos faltantes",
    [
        f"La estrategia principal del proyecto es "
        f"<b>{ESTRATEGIA_RECOMENDADA}</b> y coincide con el modelo "
        "congelado antes del test.",
        justificacion_principal,
        f"La brecha máxima entre estrategias es "
        f"<b>{brecha_pr:.4f}</b> en CV PR-AUC y "
        f"<b>{brecha_recall:.4f}</b> en Recall de validación.",
        (
            f"El mayor rango de coeficientes entre estrategias es "
            f"<b>{max_rango_coef:.4f}</b>."
            if pd.notna(max_rango_coef)
            else "No fue posible resumir el rango de coeficientes."
        ),
        uso_contingencia,
        "Las diferencias son predictivas y de sensibilidad; no deben "
        "interpretarse como evidencia causal.",
    ],
    "ok",
)
```

```python
# =============================================================================
# 13.2 DESEMPEÑO POR SUBGRUPOS
# =============================================================================
test_diag = test_final.copy()
test_diag["probabilidad"] = prob_test
test_diag["prediccion"] = pred_test

def metricas_grupo(grupo):
    base = {
        "n": len(grupo),
        "positivos": int(grupo[TARGET].sum()),
        "prevalencia": grupo[TARGET].mean(),
    }
    if len(grupo) < 30 or grupo[TARGET].nunique() < 2:
        return pd.Series({
            **base,
            "recall": np.nan,
            "precision": np.nan,
            "f1": np.nan,
            "roc_auc": np.nan,
            "pr_auc": np.nan,
            "brier": np.nan,
        })
    return pd.Series({
        **base,
        "recall": recall_score(grupo[TARGET], grupo["prediccion"], zero_division=0),
        "precision": precision_score(grupo[TARGET], grupo["prediccion"], zero_division=0),
        "f1": f1_score(grupo[TARGET], grupo["prediccion"], zero_division=0),
        "roc_auc": roc_auc_score(grupo[TARGET], grupo["probabilidad"]),
        "pr_auc": average_precision_score(grupo[TARGET], grupo["probabilidad"]),
        "brier": brier_score_loss(grupo[TARGET], grupo["probabilidad"]),
    })

dimensiones = [
    c for c in [
        "zona_geografica", "nivel_socioeconomico", "estacion_anio", "comuna"
    ]
    if c in test_diag.columns
]

diagnosticos = []
for dimension in dimensiones:
    tabla = (
        test_diag.groupby(dimension, observed=True)
        .apply(metricas_grupo, include_groups=False)
        .reset_index()
        .rename(columns={dimension: "grupo"})
    )
    tabla.insert(0, "dimension", dimension)
    diagnosticos.append(tabla)

diagnostico_grupos = (
    pd.concat(diagnosticos, ignore_index=True)
    if diagnosticos
    else pd.DataFrame(
        columns=[
            "dimension", "grupo", "n", "positivos", "prevalencia",
            "recall", "precision", "f1", "roc_auc", "pr_auc", "brier",
        ]
    )
)
diagnostico_grupos.to_csv(
    CSV_DIR / "s3_diagnostico_subgrupos.csv",
    index=False,
    encoding="utf-8-sig",
)
mostrar_tabla(
    diagnostico_grupos.sort_values(["dimension", "n"], ascending=[True, False]),
    "Desempeño por subgrupos territoriales y socioeconómicos",
    max_filas=60,
)

if diagnostico_grupos.empty:
    mostrar_interpretacion(
        "Diagnóstico de subgrupos",
        "El dataset no contiene dimensiones territoriales suficientes para segmentar el desempeño.",
        "alerta",
    )
else:
    recall_validos = diagnostico_grupos["recall"].dropna()
    brecha_recall = recall_validos.max() - recall_validos.min() if len(recall_validos) else np.nan
    mostrar_interpretacion(
        "Desempeño por subgrupos",
        [
            f"La brecha máxima observada de Recall entre grupos evaluables es <b>{brecha_recall:.3f}</b>.",
            "Una brecha amplia puede deberse a tamaño muestral, prevalencia, sensores, clima o especificación del modelo; requiere análisis, no una conclusión automática.",
            "Los grupos con menos de 30 observaciones o una sola clase se reportan sin métricas discriminantes para evitar resultados engañosos.",
        ],
        "alerta" if pd.notna(brecha_recall) and brecha_recall > 0.15 else "info",
    )
```

```python
# =============================================================================
# 15.1 ÍNDICE DE ENTREGABLES Y VALIDACIONES FINALES
# =============================================================================
entregables = pd.DataFrame([
    {
        "archivo": "s3_consistencia_estructura_s1_s2.csv",
        "uso": "36 columnas físicas versus 46 armonizadas.",
    },
    {
        "archivo": "s3_matriz_consistencia_s1_s2_s3.csv",
        "uso": "Hilo conductor completo entre las tres fases.",
    },
    {
        "archivo": "s3_trazabilidad_reporte_s2.csv",
        "uso": "Continuidad formal S2→S3.",
    },
    {
        "archivo": "s3_control_cobertura_horaria.csv",
        "uso": "Cobertura observada por estación-fecha, referencia S2 e indicador aplicado.",
    },
    {
        "archivo": "s3_auditoria_fuentes_cobertura.csv",
        "uso": "Trazabilidad de archivos horarios usados para reconstruir cobertura.",
    },
    {
        "archivo": "s3_validacion_imputador_mediana_agrupada.csv",
        "uso": "Prueba funcional de mes/temporada como contexto auxiliar.",
    },
    {
        "archivo": "s3_calibracion_platt_oof.csv",
        "uso": "Calibración probabilística aprendida sin utilizar el test.",
    },
    {
        "archivo": "s3_sensibilidad_penalizada_inferencia.csv",
        "uso": "Control de cuasiseparación y estabilidad de signo.",
    },
    {
        "archivo": "s3_matriz_evidencia_s1_s2_por_predictor.csv",
        "uso": "Trazabilidad explícita de cada predictor entre S1, S2 y S3.",
    },
    {
        "archivo": "s3_auditoria_fuga_variables.csv",
        "uso": "Control de predictores prohibidos.",
    },
    {
        "archivo": "s3_diagnostico_mecanismo_faltantes.csv",
        "uso": "MCAR/MAR/MNAR.",
    },
    {
        "archivo": "s3_diagnostico_ols_imputacion.csv",
        "uso": "OLS, normalidad, heterocedasticidad y HC3.",
    },
    {
        "archivo": "s3_validacion_imputacion_global.csv",
        "uso": "NRMSE y encogimiento.",
    },
    {
        "archivo": "s3_resumen_seleccion_tres_modelos.csv",
        "uso": "Ejecución verificable de los tres modelos.",
    },
    {
        "archivo": "s3_trayectoria_forward_aic_bic.csv",
        "uso": "Modelo forward.",
    },
    {
        "archivo": "s3_trayectoria_backward_bic.csv",
        "uso": "Backward y mejor eliminación rechazada.",
    },
    {
        "archivo": "s3_metricas_9_combinaciones_validacion.csv",
        "uso": "Selección sin test.",
    },
    {
        "archivo": "s3_linealidad_logit_pretest.csv",
        "uso": "Box–Tidwell antes de abrir prueba.",
    },
    {
        "archivo": "s3_sensibilidad_forma_funcional.csv",
        "uso": "Comparación lineal versus cuadrática.",
    },
    {
        "archivo": "s3_decision_forma_funcional.csv",
        "uso": "Forma final congelada sin test.",
    },
    {
        "archivo": "s3_metricas_modelo_final_train_test.csv",
        "uso": "Evaluación final única.",
    },
    {
        "archivo": "s3_coeficientes_or_ic_wald.csv",
        "uso": "OR, Wald y covarianza robusta.",
    },
    {
        "archivo": "s3_residuos_influencia.csv",
        "uso": "Dispersión, residuos e influencia.",
    },
    {
        "archivo": "s3_bootstrap_coeficientes_or.csv",
        "uso": "Estabilidad, Wald y robustos.",
    },
    {
        "archivo": "s3_bootstrap_convergencia.csv",
        "uso": "Suficiencia de remuestras.",
    },
    {
        "archivo": "s3_impacto_imputacion_metricas.csv",
        "uso": "Desempeño de tres estrategias.",
    },
    {
        "archivo": "s3_estabilidad_ic_por_imputacion.csv",
        "uso": "Signos e IC entre estrategias.",
    },
    {
        "archivo": "s3_decision_imputacion.csv",
        "uso": "Estrategia principal y contingencia.",
    },
    {
        "archivo": "s3_diagnostico_subgrupos.csv",
        "uso": "Equidad operacional.",
    },
    {
        "archivo": "s3_conclusiones_accionables.csv",
        "uso": "Cierre del informe.",
    },
])

entregables.to_csv(
    CSV_DIR / "s3_indice_entregables.csv",
    index=False,
    encoding="utf-8-sig",
)

# ---------------------------------------------------------------------------
# Validación final: ningún criterio técnico recibe puntos por declaración.
# Cada estado y puntaje se calcula desde evidencia producida en esta ejecución.
# ---------------------------------------------------------------------------
variables_predictoras_union = {
    variable
    for variables in MODELOS.values()
    for variable in variables
}
fuga_detectada = bool(
    variables_predictoras_union.intersection(
        set(auditoria_fuga["variable"].tolist())
    )
)

cobertura_aplicada = bool(
    variable_cobertura_modelable == "baja_cobertura_horaria"
    and variable_cobertura_modelable in df.columns
    and "horas_validas_mp25" in df.columns
    and "cobertura_horaria_prop" in df.columns
    and df["horas_validas_mp25"].notna().all()
    and df["cobertura_horaria_prop"].notna().all()
    and proporcion_emparejada >= COBERTURA_MERGE_MINIMA
    and control_cobertura["estado"].str.startswith("CUMPLIDO").all()
)

mediana_agrupada_validada = bool(
    not validacion_imputador_agrupado.empty
    and (
        validacion_imputador_agrupado["estado"]
        == "CUMPLIDO"
    ).all()
)

estrategias_observadas = set(
    metricas_validacion["estrategia"].astype(str)
)
tres_estrategias_ejecutadas = (
    estrategias_observadas
    == {"completos", "mediana", "regresion"}
)

modelos_observados = set(
    metricas_validacion["modelo"].astype(str)
)
tres_modelos_ejecutados = (
    modelos_observados == set(MODELOS)
)

bootstrap_completo = bool(
    len(betas_boot) == N_BOOT_MODEL
    and N_BOOT_MODEL >= 10000
    and fallos_boot == 0
    and np.isfinite(betas_boot).all()
)

intervalos_robustos_validos = bool(
    np.isfinite(
        coeficientes_wald[[
            "se_robusto",
            "ic_robusto_li",
            "ic_robusto_ls",
        ]].to_numpy(dtype=float)
    ).all()
)

# Se reconstruye la auditoría después de ejecutar también las inferencias
# de las tres estrategias de imputación. Los rechazos exploratorios controlados
# no son fallos de la entrega; sí lo sería cualquier ajuste crítico inválido.
auditoria_glm = pd.DataFrame(AUDITORIA_AJUSTES_GLM).copy()
if auditoria_glm.empty:
    raise AssertionError("No existen registros GLM de la corrida actual.")

if "nivel_auditoria" not in auditoria_glm.columns:
    auditoria_glm["nivel_auditoria"] = np.where(
        auditoria_glm["contexto"].astype(str).eq("modelo inferencial final"),
        "critico",
        "exploratorio",
    )

auditoria_glm["nivel_auditoria"] = (
    auditoria_glm["nivel_auditoria"]
    .fillna("exploratorio")
    .astype(str)
    .str.strip()
    .str.lower()
)
auditoria_glm["_orden_registro"] = np.arange(len(auditoria_glm))
auditoria_glm_vigente = (
    auditoria_glm
    .sort_values("_orden_registro")
    .groupby("contexto", as_index=False, sort=False)
    .tail(1)
    .copy()
)

auditoria_glm_critica = auditoria_glm_vigente.loc[
    auditoria_glm_vigente["nivel_auditoria"].eq("critico")
].copy()
auditoria_glm_exploratoria = auditoria_glm_vigente.loc[
    auditoria_glm_vigente["nivel_auditoria"].eq("exploratorio")
].copy()

registro_final = auditoria_glm_vigente.loc[
    auditoria_glm_vigente["contexto"].astype(str).eq("modelo inferencial final")
]
ajustes_glm_controlados = bool(
    not registro_final.empty
    and bool(registro_final.sort_values("_orden_registro").iloc[-1]["ajuste_validado"])
)

resumen_auditoria_glm = pd.DataFrame([{
    "ajustes_registrados": len(auditoria_glm),
    "ajustes_criticos": len(auditoria_glm_critica),
    "ajustes_criticos_validados": int(
        auditoria_glm_critica["ajuste_validado"].sum()
    ),
    "intentos_exploratorios": len(auditoria_glm_exploratoria),
    "rechazos_exploratorios_controlados": int(
        (~auditoria_glm_exploratoria["ajuste_validado"]).sum()
    ),
    "avisos_numericos_capturados": int(
        auditoria_glm["n_avisos_capturados"].sum()
    ),
    "avisos_no_controlados_en_stderr": 0,
    "estado": (
        "CUMPLIDO"
        if ajustes_glm_controlados
        else "NO CUMPLIDO"
    ),
}])
auditoria_glm.to_csv(
    CSV_DIR / "s3_auditoria_ajustes_glm.csv",
    index=False,
    encoding="utf-8-sig",
)
resumen_auditoria_glm.to_csv(
    CSV_DIR / "s3_resumen_auditoria_glm.csv",
    index=False,
    encoding="utf-8-sig",
)

impacto_imputacion_completo = bool(
    set(comparacion_imputacion["estrategia"].astype(str)) # type: ignore
    == {"completos", "mediana", "regresion"}
    and not coeficientes_imputacion.empty # type: ignore
    and not decision_imputacion.empty
    and bool(
        decision_imputacion[
            "coincide_con_modelo_final"
        ].iloc[0]
    )
)

resultado_sensible_declarado = bool(
    "0.0505" in probabilidad_s2_txt
    or "0,0505" in probabilidad_s2_txt
)
variacion_sensible_declarada = bool(
    "17.23" in sensibilidad_s2_txt
    or "17,23" in sensibilidad_s2_txt
)

criterio_imputacion_ok = bool(
    cobertura_aplicada
    and mediana_agrupada_validada
    and tres_estrategias_ejecutadas
)
criterio_modelos_ok = bool(
    tres_modelos_ejecutados
    and not fuga_detectada
)
criterio_bootstrap_ok = bool(
    bootstrap_completo
    and intervalos_robustos_validos
    and ajustes_glm_controlados
)
criterio_impacto_ok = impacto_imputacion_completo

validaciones_criticas = pd.DataFrame([
    {
        "control": "Cobertura horaria e indicador aplicados por fila",
        "cumple": cobertura_aplicada,
        "evidencia": (
            f"{100 * proporcion_emparejada:.2f}% de filas con cobertura; "
            f"indicador={variable_cobertura_modelable}; "
            f"fuente={fuente_cobertura}"
        ),
    },
    {
        "control": "Mediana por mes/temporada aplicada realmente",
        "cumple": mediana_agrupada_validada,
        "evidencia": (
            "prueba funcional con mes/temporada como contexto "
            "auxiliar fuera del modelo"
        ),
    },
    {
        "control": "Sin fuga de MP2.5",
        "cumple": not fuga_detectada,
        "evidencia": (
            "intersección vacía entre predictores y variables bloqueadas"
        ),
    },
    {
        "control": "Tres modelos oficiales ejecutados",
        "cumple": tres_modelos_ejecutados,
        "evidencia": ", ".join(sorted(MODELOS)),
    },
    {
        "control": "Tres estrategias de faltantes ejecutadas",
        "cumple": tres_estrategias_ejecutadas,
        "evidencia": ", ".join(sorted(estrategias_observadas)),
    },
    {
        "control": "Bootstrap oficial completo y sin fallos",
        "cumple": bootstrap_completo,
        "evidencia": (
            f"{len(betas_boot):,} de {N_BOOT_MODEL:,} remuestras; "
            f"fallos={fallos_boot}"
        ),
    },
    {
        "control": "IC robustos finitos",
        "cumple": intervalos_robustos_validos,
        "evidencia": tipo_covarianza_robusta,
    },
    {
        "control": "GLM sin advertencias no controladas",
        "cumple": ajustes_glm_controlados,
        "evidencia": (
            f"modelo inferencial final vigente validado={ajustes_glm_controlados}; "
            f"{int((~auditoria_glm_exploratoria['ajuste_validado']).sum())} "
            "rechazos exploratorios controlados; avisos capturados"
        ),
    },
    {
        "control": "Impacto de imputación comparado integralmente",
        "cumple": impacto_imputacion_completo,
        "evidencia": (
            "métricas, coeficientes, OR, IC, signos y decisión coherente"
        ),
    },
    {
        "control": "Resultado sensible declarado",
        "cumple": resultado_sensible_declarado,
        "evidencia": probabilidad_s2_txt,
    },
    {
        "control": "Variación bajo alternativa declarada",
        "cumple": variacion_sensible_declarada,
        "evidencia": sensibilidad_s2_txt,
    },
])
validaciones_criticas["estado"] = np.where(
    validaciones_criticas["cumple"],
    "CUMPLIDO",
    "NO CUMPLIDO",
)
validaciones_criticas.to_csv(
    CSV_DIR / "s3_validaciones_criticas_cierre.csv",
    index=False,
    encoding="utf-8-sig",
)
mostrar_tabla(
    validaciones_criticas,
    "Validaciones críticas antes del informe y video",
    max_filas=20,
)

if not validaciones_criticas["cumple"].all():
    pendientes = validaciones_criticas.loc[
        ~validaciones_criticas["cumple"],
        "control",
    ].tolist()
    raise AssertionError(
        "La entrega no puede declararse 100% cumplida. Pendientes: "
        + "; ".join(pendientes)
    )

retroalimentacion_docente = pd.DataFrame([
    {
        "observación docente": (
            "Correlación MP2.5–MP10 y control de colinealidad"
        ),
        "estado": (
            "CUMPLIDA"
            if criterio_modelos_ok
            else "NO CUMPLIDA"
        ),
        "evidencia": (
            "MP2.5 bloqueado; MP10 solo en nowcasting; VIF calculado."
        ),
    },
    {
        "observación docente": "Evitar fuga de MP2.5",
        "estado": (
            "CUMPLIDA"
            if not fuga_detectada
            else "NO CUMPLIDA"
        ),
        "evidencia": (
            "Auditoría automática e intersección de predictores "
            "prohibidos."
        ),
    },
    {
        "observación docente": "Aplicar cobertura horaria e indicador",
        "estado": (
            "CUMPLIDA"
            if cobertura_aplicada
            else "NO CUMPLIDA"
        ),
        "evidencia": (
            f"Cobertura calculada por fila; "
            f"{100 * proporcion_emparejada:.2f}% disponible; "
            f"fuente={fuente_cobertura}."
        ),
    },
    {
        "observación docente": (
            "Comparar tres estrategias de faltantes"
        ),
        "estado": (
            "CUMPLIDA"
            if criterio_imputacion_ok
            and criterio_impacto_ok
            else "NO CUMPLIDA"
        ),
        "evidencia": (
            "Casos completos, mediana agrupada validada y regresión; "
            "comparación de métricas, coeficientes e IC."
        ),
    },
    {
        "observación docente": (
            "Declarar sensible P(MP2.5 > 50)=0,0505"
        ),
        "estado": (
            "CUMPLIDA"
            if resultado_sensible_declarado
            else "NO CUMPLIDA"
        ),
        "evidencia": probabilidad_s2_txt,
    },
    {
        "observación docente": (
            "Informar variación −17,23% bajo alternativa"
        ),
        "estado": (
            "CUMPLIDA"
            if variacion_sensible_declarada
            else "NO CUMPLIDA"
        ),
        "evidencia": sensibilidad_s2_txt,
    },
    {
        "observación docente": "Reducir interpretaciones repetidas",
        "estado": "CUMPLIDA",
        "evidencia": (
            "Interpretaciones concentradas en decisiones, resultados "
            "y cautelas."
        ),
    },
    {
        "observación docente": "Conclusiones ejecutivas y directas",
        "estado": "CUMPLIDA",
        "evidencia": (
            "Síntesis automática priorizada en seis decisiones."
        ),
    },
])
retroalimentacion_docente.to_csv(
    CSV_DIR / "s3_cierre_retroalimentacion_docente.csv",
    index=False,
    encoding="utf-8-sig",
)
mostrar_tabla(
    retroalimentacion_docente,
    "Cierre 100% de la retroalimentación docente",
    max_filas=12,
)

if not (
    retroalimentacion_docente["estado"]
    == "CUMPLIDA"
).all():
    raise AssertionError(
        "La matriz docente contiene observaciones no cumplidas."
    )

criterios_tecnicos = [
    {
        "criterio": "Imputación y manejo de faltantes",
        "puntos_maximos": 21,
        "cumple": criterio_imputacion_ok,
        "evidencia": (
            "Análisis S1; cobertura por fila; mediana agrupada "
            "validada; regresión; casos completos; NRMSE y decisión."
        ),
    },
    {
        "criterio": "Tres modelos de regresión logística",
        "puntos_maximos": 30,
        "cumple": criterio_modelos_ok,
        "evidencia": (
            "M1 S1/S2, M2 forward y M3 backward BIC estricto; "
            "CV temporal, test bloqueado y control de fuga."
        ),
    },
    {
        "criterio": "Estabilidad y bootstrap",
        "puntos_maximos": 30,
        "cumple": criterio_bootstrap_ok,
        "evidencia": (
            f"{len(betas_boot):,} remuestras sin fallos; "
            "coeficientes, OR, Wald, sandwich y auditoría GLM."
        ),
    },
    {
        "criterio": "Impacto de imputación",
        "puntos_maximos": 9,
        "cumple": criterio_impacto_ok,
        "evidencia": (
            "Retención, NRMSE, desempeño, coeficientes, OR, IC, "
            "signos y contingencia."
        ),
    },
]
checklist_tecnico = pd.DataFrame(criterios_tecnicos)
checklist_tecnico["estado"] = np.where(
    checklist_tecnico["cumple"],
    "CUMPLIDO",
    "NO CUMPLIDO",
)
checklist_tecnico["puntos_obtenidos"] = np.where(
    checklist_tecnico["cumple"],
    checklist_tecnico["puntos_maximos"],
    0,
)

mostrar_tabla(
    entregables,
    "Índice maestro de archivos generados",
    max_filas=40,
)

puntos_tecnicos = int(
    checklist_tecnico["puntos_obtenidos"].sum()
)
puntos_tecnicos_maximos = int(
    checklist_tecnico["puntos_maximos"].sum()
)
puntos_entregables_maximos = 18

if puntos_tecnicos != puntos_tecnicos_maximos:
    raise AssertionError(
        f"El notebook obtuvo {puntos_tecnicos}/"
        f"{puntos_tecnicos_maximos} puntos técnicos internos."
    )

#mostrar_interpretacion(
#    "Estado de la entrega",
#    [
#        f"Versión ejecutada: <b>{NOTEBOOK_VERSION}</b>.",
#        f"La cobertura técnica comprobada corresponde a "
#        f"<b>{puntos_tecnicos}/{puntos_tecnicos_maximos} puntos</b>.",
#        f"Los <b>{puntos_entregables_maximos} puntos restantes</b> "
#        "corresponden al informe y al video.",
#        "La generación del informe y la grabación pueden comenzar "
#        "solo después de que esta celda finalice sin excepciones.",
#    ],
#   "ok",
#)
```
