```python
# =============================================================================
# 1. CONFIGURACIÓN, DEPENDENCIAS Y FUNCIONES HTML PREMIUM
# =============================================================================
import os
import sys
import time
import math
import html
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

import statsmodels.api as sm
from statsmodels.stats.diagnostic import het_breuschpagan
from statsmodels.stats.outliers_influence import variance_inflation_factor

from IPython.display import display, HTML

from sklearn.experimental import enable_iterative_imputer  # noqa: F401
from sklearn.impute import SimpleImputer, IterativeImputer
from sklearn.linear_model import LogisticRegression, BayesianRidge
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    confusion_matrix, precision_score, recall_score, f1_score, fbeta_score,
    roc_auc_score, average_precision_score, balanced_accuracy_score,
    accuracy_score, brier_score_loss, roc_curve, precision_recall_curve
)
from sklearn.calibration import calibration_curve

warnings.filterwarnings("ignore")

SEED = 2026
ALPHA = 0.05
MIN_RECALL = 0.70
COSTO_FN = 5.0
COSTO_FP = 1.0

FAST_MODE = os.getenv("S3_FAST", "0") == "1"
N_BOOT_MODEL = int(os.getenv("S3_BOOTSTRAP", "500" if FAST_MODE else "10000"))
N_MASK_REPEATS = 1 if FAST_MODE else 5
N_CV_SPLITS = 3 if FAST_MODE else 5
rng = np.random.default_rng(SEED)

CWD = Path.cwd().resolve()
if CWD.name.lower() == "notebooks":
    PROJECT_ROOT = CWD.parent
elif (CWD / "data").exists():
    PROJECT_ROOT = CWD
else:
    PROJECT_ROOT = next((p for p in [CWD, *CWD.parents] if (p / "data").exists()), CWD)

DATA_DIR = PROJECT_ROOT / "data"
FIGURES_DIR = PROJECT_ROOT / "figures" / "S3"
RESULTS_DIR = PROJECT_ROOT / "resultados" / "S3"
FIGURES_DIR.mkdir(parents=True, exist_ok=True)
RESULTS_DIR.mkdir(parents=True, exist_ok=True)

plt.rcParams.update({
    "figure.figsize": (11.5, 6.2), "figure.dpi": 130, "savefig.dpi": 300,
    "axes.grid": True, "grid.alpha": 0.25, "axes.titleweight": "bold",
    "axes.titlesize": 14, "axes.labelsize": 11, "font.size": 10,
    "legend.frameon": True, "legend.framealpha": 0.95,
    "axes.spines.top": False, "axes.spines.right": False,
})

def guardar_figura(nombre):
    ruta = FIGURES_DIR / nombre
    plt.savefig(ruta, bbox_inches="tight", pad_inches=0.25, facecolor="white")
    return ruta

def _fmt_html(valor, decimales=3):
    """Formatea valores sin permitir saltos dentro de números."""
    if pd.isna(valor):
        return "—"
    if isinstance(valor, (pd.Timestamp, np.datetime64)):
        return pd.Timestamp(valor).strftime("%Y-%m-%d")
    if isinstance(valor, (bool, np.bool_)):
        return "Sí" if bool(valor) else "No"
    if isinstance(valor, (float, np.floating)):
        return f"{float(valor):,.{decimales}f}"
    if isinstance(valor, (int, np.integer)):
        return f"{int(valor):,}"
    return html.escape(str(valor))

def _etiqueta_columna(columna):
    """Convierte nombres técnicos en encabezados legibles."""
    return html.escape(str(columna).replace("_", " ").strip())

def _configurar_columna(serie, nombre):
    """
    Calcula un ancho mínimo por columna.

    La tabla puede desplazarse horizontalmente, pero no comprime números,
    métricas ni nombres hasta volverlos ilegibles.
    """
    nombre_limpio = str(nombre).replace("_", " ")
    nombre_lower = str(nombre).lower()

    columnas_texto_largo = {
        "lectura", "advertencia", "predictores", "regla", "criterio",
        "evidencia", "implicancia", "hallazgo", "uso", "uso_s3",
        "motivo_exclusion", "contenido", "resultado_validado",
    }

    if pd.api.types.is_bool_dtype(serie):
        return 92, "nowrap", "center"

    if pd.api.types.is_numeric_dtype(serie):
        ancho = max(96, min(138, 18 + len(nombre_limpio) * 7))
        return ancho, "nowrap", "right"

    muestra = serie.dropna().astype(str).head(100)
    max_chars = max([len(nombre_limpio)] + [len(x) for x in muestra.tolist()])

    if nombre_lower in columnas_texto_largo or max_chars > 44:
        ancho = max(230, min(390, 60 + min(max_chars, 48) * 6))
        return ancho, "normal", "left"

    ancho = max(125, min(245, 28 + max_chars * 7))
    return ancho, "nowrap", "left"

def tabla_html(df, titulo=None, max_filas=30, decimales=3):
    """
    Genera una tabla legible en GitHub, JupyterLab, VS Code y HTML exportado.

    - Cada columna recibe un ancho mínimo calculado.
    - Los valores numéricos nunca se dividen entre líneas.
    - Las columnas narrativas sí pueden envolver texto.
    - Las tablas anchas usan scroll horizontal sin comprimir su contenido.
    - La primera columna queda visualmente destacada durante el desplazamiento.
    """
    tabla = df.copy()
    recortada = len(tabla) > max_filas
    if recortada:
        tabla = tabla.head(max_filas)

    configuracion = {
        columna: _configurar_columna(tabla[columna], columna)
        for columna in tabla.columns
    }
    anchos = {columna: valores[0] for columna, valores in configuracion.items()}
    ancho_total = sum(anchos.values())
    ancho_tabla = max(760, ancho_total)

    colgroup = "".join(
        f'<col style="width:{anchos[col]}px;min-width:{anchos[col]}px;">'
        for col in tabla.columns
    )

    encabezados = []
    for indice, columna in enumerate(tabla.columns):
        ancho, _, alineacion = configuracion[columna]
        sticky = (
            "position:sticky;left:0;z-index:4;"
            "box-shadow:5px 0 8px rgba(15,23,42,.08);"
            if indice == 0 else ""
        )
        encabezados.append(
            f"""<th style="box-sizing:border-box;width:{ancho}px;min-width:{ancho}px;
                         padding:11px 13px;text-align:{alineacion};
                         background:#17375E !important;color:#FFFFFF !important;
                         border-bottom:3px solid #3B82F6;font-weight:900;
                         white-space:normal;word-break:normal;overflow-wrap:normal;
                         vertical-align:middle;line-height:1.25;{sticky}">
                  {_etiqueta_columna(columna)}
                </th>"""
        )

    filas_html = []
    for indice_fila, (_, fila) in enumerate(tabla.iterrows()):
        fondo = "#FFFFFF" if indice_fila % 2 == 0 else "#F3F6F9"
        celdas = []

        for indice_columna, columna in enumerate(tabla.columns):
            ancho, white_space, alineacion = configuracion[columna]
            sticky = (
                f"position:sticky;left:0;z-index:2;background:{fondo} !important;"
                "box-shadow:5px 0 8px rgba(15,23,42,.06);font-weight:700;"
                if indice_columna == 0 else f"background:{fondo} !important;"
            )

            celdas.append(
                f"""<td style="box-sizing:border-box;width:{ancho}px;min-width:{ancho}px;
                             padding:10px 13px;{sticky}color:#172033 !important;
                             border-bottom:1px solid #DDE5EE;vertical-align:middle;
                             line-height:1.45;white-space:{white_space};
                             word-break:normal;overflow-wrap:normal;text-align:{alineacion};
                             font-variant-numeric:tabular-nums;">
                      {_fmt_html(fila[columna], decimales)}
                    </td>"""
            )

        filas_html.append(f"<tr>{''.join(celdas)}</tr>")

    titulo_html = ""
    if titulo:
        titulo_html = f"""
        <div style="padding:14px 17px;background:#EAF2FA !important;
                    color:#17375E !important;border-bottom:1px solid #CBD5E1;
                    font-size:17px;font-weight:900;line-height:1.3;">
          {html.escape(titulo)}
        </div>"""

    nota = ""
    if recortada:
        nota = f"""
        <div style="padding:10px 14px;background:#FFF7ED !important;
                    color:#7C2D12 !important;border-top:1px solid #FED7AA;
                    font-size:12.5px;">
          Se muestran {max_filas} de {len(df):,} filas. La tabla completa se exporta a CSV.
        </div>"""

    ayuda_scroll = ""
    if ancho_total > 1250:
        ayuda_scroll = """
        <div style="padding:8px 14px;background:#F8FAFC !important;
                    color:#475569 !important;border-bottom:1px solid #E2E8F0;
                    font-size:12px;">
          ↔ Tabla extendida: desplácese horizontalmente. Los valores se mantienen completos.
        </div>"""

    return f"""
    <div style="box-sizing:border-box;width:100%;max-width:100%;margin:16px 0 20px 0;
                background:#FFFFFF !important;color:#172033 !important;color-scheme:light;
                border:1px solid #CBD5E1;border-radius:14px;overflow:hidden;
                box-shadow:0 7px 20px rgba(15,23,42,.08);
                font-family:Arial,Helvetica,sans-serif;">
      {titulo_html}
      {ayuda_scroll}
      <div style="box-sizing:border-box;width:100%;max-width:100%;overflow-x:auto;
                  overflow-y:hidden;scrollbar-gutter:stable;padding-bottom:3px;">
        <table style="width:{ancho_tabla}px;min-width:{ancho_tabla}px;
                      table-layout:fixed;border-collapse:separate;border-spacing:0;
                      background:#FFFFFF !important;color:#172033 !important;
                      font-size:13.5px;">
          <colgroup>{colgroup}</colgroup>
          <thead><tr>{''.join(encabezados)}</tr></thead>
          <tbody>{''.join(filas_html)}</tbody>
        </table>
      </div>
      {nota}
    </div>
    """

def mostrar_tabla(df, titulo=None, max_filas=30, decimales=3):
    display(HTML(tabla_html(df, titulo, max_filas, decimales)))

def mostrar_interpretacion(titulo, parrafos, nivel="info"):
    colores = {
        "info": ("#EAF4FF", "#1D6FB8", "#17375E"),
        "ok": ("#ECFDF3", "#15803D", "#14532D"),
        "alerta": ("#FFF7ED", "#B45309", "#7C2D12"),
        "riesgo": ("#FFF1F2", "#B42335", "#7F1D1D"),
    }
    fondo, borde, texto = colores.get(nivel, colores["info"])
    if isinstance(parrafos, str):
        parrafos = [parrafos]
    cuerpo = "".join(
        f"<p style='margin:{0 if i == 0 else 9}px 0 0 0;color:#26364D !important;'>{p}</p>"
        for i, p in enumerate(parrafos)
    )
    display(HTML(f"""
    <div style="box-sizing:border-box;width:100%;max-width:100%;overflow-wrap:anywhere;
                background:{fondo} !important;color:{texto} !important;color-scheme:light;
                border:1px solid #CBD5E1;border-left:7px solid {borde};border-radius:14px;
                padding:18px 21px;margin:16px 0;font-family:Arial,Helvetica,sans-serif;
                line-height:1.62;">
      <div style="color:{texto} !important;font-size:17px;font-weight:900;margin-bottom:7px;">
        {titulo}
      </div>
      {cuerpo}
    </div>
    """))

def mostrar_kpis(titulo, items):
    tarjetas = []
    for etiqueta, valor, detalle in items:
        tarjetas.append(f"""
        <div style="box-sizing:border-box;background:#FFFFFF !important;color:#172033 !important;
                    border:1px solid #CBD5E1;border-top:5px solid #2563EB;border-radius:12px;
                    padding:14px 15px;min-width:180px;">
          <div style="color:#40536B !important;font-size:12px;font-weight:900;text-transform:uppercase;
                      letter-spacing:.5px;">{html.escape(str(etiqueta))}</div>
          <div style="color:#17375E !important;font-size:22px;font-weight:900;margin-top:5px;">
            {html.escape(str(valor))}
          </div>
          <div style="color:#40536B !important;font-size:12.5px;margin-top:5px;line-height:1.4;">
            {html.escape(str(detalle))}
          </div>
        </div>""")
    display(HTML(f"""
    <div style="box-sizing:border-box;width:100%;max-width:100%;background:#F7FAFC !important;
                color:#172033 !important;color-scheme:light;border:1px solid #CBD5E1;
                border-radius:15px;padding:18px;margin:16px 0;font-family:Arial,Helvetica,sans-serif;">
      <div style="color:#17375E !important;font-size:18px;font-weight:900;margin-bottom:13px;">
        {html.escape(titulo)}
      </div>
      <div style="display:grid;grid-template-columns:repeat(auto-fit,minmax(180px,1fr));gap:12px;">
        {''.join(tarjetas)}
      </div>
    </div>
    """))

mostrar_kpis(
    "Entorno técnico",
    [
        ("Python", sys.version.split()[0], "Versión activa"),
        ("Semilla", SEED, "Reproducibilidad"),
        ("Bootstrap", f"{N_BOOT_MODEL:,}", "Remuestras del modelo"),
        ("CV temporal", N_CV_SPLITS, "Bloques expansivos"),
        ("Modo", "RÁPIDO" if FAST_MODE else "COMPLETO", "Control mediante S3_FAST"),
    ],
)
mostrar_interpretacion(
    "✅ Configuración validada",
    [
        "El notebook separa explícitamente el modelo <b>predictivo ponderado</b> del modelo <b>inferencial no ponderado</b>.",
        "Los cuadros y tablas fijan fondo, texto y bordes mediante estilos inline para evitar pérdida de contraste en GitHub.",
    ],
    "ok",
)
```

```python
# =============================================================================
# 2.1 CARGA ROBUSTA Y RECONSTRUCCIÓN DE LA MATRIZ ARMONIZADA S1/S2
# =============================================================================
def leer_csv_robusto(ruta: str | Path) -> pd.DataFrame:
    """
    Lee un archivo CSV probando codificaciones y separadores habituales.

    Las llamadas a ``pd.read_csv`` se realizan de forma explícita para que
    Pylance pueda validar correctamente los tipos de cada argumento. Esto
    evita las advertencias falsas generadas por un diccionario ``**kwargs``
    construido dinámicamente.
    """
    errores: list[str] = []
    codificaciones: tuple[str, ...] = ("utf-8-sig", "utf-8", "latin-1")
    separadores: tuple[str | None, ...] = (None, ";", ",")

    for codificacion in codificaciones:
        for separador in separadores:
            try:
                if separador is None:
                    # Detección automática del separador mediante el motor Python.
                    temporal = pd.read_csv(
                        ruta,
                        encoding=codificacion,
                        sep=None,
                        engine="python",
                    )
                else:
                    temporal = pd.read_csv(
                        ruta,
                        encoding=codificacion,
                        sep=separador,
                    )

                if temporal.shape[1] > 1:
                    return temporal

                errores.append(
                    f"{codificacion}/{separador}: solo se detectó una columna"
                )

            except (UnicodeDecodeError, pd.errors.ParserError, OSError, ValueError) as exc:
                errores.append(f"{codificacion}/{separador}: {exc}")

    detalle = " | ".join(errores[-8:])
    raise RuntimeError(
        "No fue posible leer el archivo CSV con las combinaciones "
        f"de codificación y separador disponibles. Detalle: {detalle}"
    )

CANDIDATOS_DATA = [
    DATA_DIR / "calidad_aire_diario.csv",
    PROJECT_ROOT / "calidad_aire_diario.csv",
    CWD / "data" / "calidad_aire_diario.csv",
    CWD.parent / "data" / "calidad_aire_diario.csv",
]
ruta_dataset = next((p for p in CANDIDATOS_DATA if p.exists()), None)
if ruta_dataset is None:
    raise FileNotFoundError(
        "No se encontró calidad_aire_diario.csv.\nRutas revisadas:\n- "
        + "\n- ".join(str(p) for p in CANDIDATOS_DATA)
    )

# ---------------------------------------------------------------------------
# A. Estructura física original: 16.071 filas × 36 columnas.
# ---------------------------------------------------------------------------
df_base = leer_csv_robusto(ruta_dataset)
df_base.columns = (
    df_base.columns.astype(str).str.strip().str.replace("\ufeff", "", regex=False)
)

col_fecha = next((c for c in ["fecha", "date", "Fecha"] if c in df_base.columns), None)
if col_fecha is None:
    raise ValueError("Se requiere una columna de fecha para realizar validación temporal.")

df_base[col_fecha] = pd.to_datetime(df_base[col_fecha], errors="coerce")
n_fechas_invalidas = int(df_base[col_fecha].isna().sum())
df_base = (
    df_base.loc[df_base[col_fecha].notna()]
    .sort_values(col_fecha)
    .reset_index(drop=True)
)

shape_base_original = df_base.shape
columnas_base_originales = df_base.columns.tolist()

# ---------------------------------------------------------------------------
# B. Enriquecimiento idéntico al utilizado en S1 y reconstruido en S2.
#    Cinco variables medias generan diez variables físicas min/max:
#    temperatura, presión, humedad, viento y radiación.
# ---------------------------------------------------------------------------
DELTAS_FISICOS_S1_S2 = {
    "temperatura_mean": {"min_sub": 6.5, "max_add": 8.0},
    "presion_mean": {"min_sub": 2.1, "max_add": 2.3},
    "humedad_mean": {"min_sub": 20.0, "max_add": 22.0},
    "viento_mean": {"min_sub": 1.2, "max_add": 1.8},
    "radiacion_mean": {"min_sub": 150.0, "max_add": 300.0},
}

df_raw = df_base.copy()
orden_enriquecido = []

def agregar_columna_unica(lista, columna):
    if columna in df_raw.columns and columna not in lista:
        lista.append(columna)

for columna in columnas_base_originales:
    agregar_columna_unica(orden_enriquecido, columna)

    if columna not in DELTAS_FISICOS_S1_S2:
        continue

    configuracion = DELTAS_FISICOS_S1_S2[columna]
    columna_min = columna.replace("_mean", "_min")
    columna_max = columna.replace("_mean", "_max")

    if columna_min not in df_raw.columns:
        df_raw[columna_min] = (
            pd.to_numeric(df_raw[columna], errors="coerce")
            - configuracion["min_sub"]
        )
    if columna_max not in df_raw.columns:
        df_raw[columna_max] = (
            pd.to_numeric(df_raw[columna], errors="coerce")
            + configuracion["max_add"]
        )

    if "humedad" in columna:
        df_raw[columna_min] = df_raw[columna_min].clip(lower=0, upper=100)
        df_raw[columna_max] = df_raw[columna_max].clip(lower=0, upper=100)

    if "viento" in columna or "radiacion" in columna:
        df_raw[columna_min] = df_raw[columna_min].clip(lower=0)
        df_raw[columna_max] = df_raw[columna_max].clip(lower=0)

    agregar_columna_unica(orden_enriquecido, columna_min)
    agregar_columna_unica(orden_enriquecido, columna_max)

# Si el archivo ya contiene alguna derivada, se preserva una sola vez.
for columna in df_raw.columns:
    agregar_columna_unica(orden_enriquecido, columna)

df_raw = df_raw.loc[:, orden_enriquecido]
shape_enriquecida = df_raw.shape

# Categorías ordenadas, consistentes con S1/S2.
orden_calidad = ["Bueno", "Regular", "Alerta", "Preemergencia", "Emergencia"]
orden_nse = [
    "Q1 (bajo)", "Q2 (medio-bajo)", "Q3 (medio)",
    "Q4 (medio-alto)", "Q5 (alto)",
]

if "calidad_aire_mp25" in df_raw.columns:
    df_raw["calidad_aire_mp25"] = pd.Categorical(
        df_raw["calidad_aire_mp25"],
        categories=orden_calidad,
        ordered=True,
    )
if "calidad_aire_mp10" in df_raw.columns:
    df_raw["calidad_aire_mp10"] = pd.Categorical(
        df_raw["calidad_aire_mp10"],
        categories=orden_calidad,
        ordered=True,
    )
if "nivel_socioeconomico" in df_raw.columns:
    df_raw["nivel_socioeconomico"] = pd.Categorical(
        df_raw["nivel_socioeconomico"],
        categories=orden_nse,
        ordered=True,
    )

# Línea base histórica del proyecto.
BASELINE_PROYECTO = {
    "filas_base": 16071,
    "columnas_base": 36,
    "filas_inferenciales": 11603,
    "columnas_enriquecidas": 46,
}

resumen_carga = pd.DataFrame([
    {
        "estructura": "CSV físico original",
        "filas": shape_base_original[0],
        "columnas": shape_base_original[1],
        "esperado": "16.071 × 36",
        "estado": (
            "Consistente"
            if shape_base_original == (
                BASELINE_PROYECTO["filas_base"],
                BASELINE_PROYECTO["columnas_base"],
            )
            else "Revisar versión del archivo"
        ),
    },
    {
        "estructura": "Matriz armonizada S1/S2",
        "filas": shape_enriquecida[0],
        "columnas": shape_enriquecida[1],
        "esperado": "16.071 × 46",
        "estado": (
            "Consistente"
            if shape_enriquecida == (
                BASELINE_PROYECTO["filas_base"],
                BASELINE_PROYECTO["columnas_enriquecidas"],
            )
            else "Revisar enriquecimiento"
        ),
    },
])

resumen_archivo = pd.DataFrame([{
    "archivo": ruta_dataset.name,
    "fecha mínima": df_raw[col_fecha].min(),
    "fecha máxima": df_raw[col_fecha].max(),
    "fechas inválidas excluidas": n_fechas_invalidas,
    "duplicados exactos": int(df_raw.duplicated().sum()),
    "variables físicas incorporadas": shape_enriquecida[1] - shape_base_original[1],
}])

resumen_carga.to_csv(
    RESULTS_DIR / "s3_consistencia_estructura_s1_s2.csv",
    index=False,
    encoding="utf-8-sig",
)
resumen_archivo.to_csv(
    RESULTS_DIR / "s3_resumen_carga.csv",
    index=False,
    encoding="utf-8-sig",
)

mostrar_tabla(resumen_carga, "Continuidad estructural: archivo físico y matriz enriquecida")
mostrar_tabla(resumen_archivo, "Control técnico del archivo")
mostrar_tabla(
    df_raw.head(5),
    "Vista de la matriz armonizada de 46 variables",
    max_filas=5,
)

mostrar_interpretacion(
    "📌 Interpretación de la carga y enriquecimiento",
    [
        f"El archivo físico contiene <b>{shape_base_original[0]:,}</b> observaciones y "
        f"<b>{shape_base_original[1]}</b> columnas originales.",
        f"Al replicar la preparación de S1/S2 se recupera una matriz de "
        f"<b>{shape_enriquecida[0]:,} × {shape_enriquecida[1]}</b>.",
        "Las diez columnas adicionales corresponden a mínimos y máximos físicos derivados "
        "de temperatura, presión, humedad, viento y radiación; no son diez fuentes nuevas.",
        "Esta distinción evita confundir el inventario total de variables con el subconjunto "
        "de predictores que finalmente utiliza cada modelo.",
    ],
    "ok" if shape_enriquecida[1] == 46 else "alerta",
)
```

```python
# =============================================================================
# 2.2 CARGA Y AUDITORÍA DEL REPORTE PUENTE DE SUMATIVA 2
# =============================================================================
CANDIDATOS_S2 = [
    PROJECT_ROOT / "resultados" / "s2_reporte_resultados_validados_para_s3.csv",
    PROJECT_ROOT / "s2_reporte_resultados_validados_para_s3.csv",
    CWD / "resultados" / "s2_reporte_resultados_validados_para_s3.csv",
    CWD / "s2_reporte_resultados_validados_para_s3.csv",
]
ruta_reporte_s2 = next((p for p in CANDIDATOS_S2 if p.exists()), None)
if ruta_reporte_s2 is None:
    raise FileNotFoundError(
        "No se encontró s2_reporte_resultados_validados_para_s3.csv.\n"
        "Ubicación recomendada: resultados/s2_reporte_resultados_validados_para_s3.csv"
    )

reporte_s2 = leer_csv_robusto(ruta_reporte_s2)
reporte_s2.columns = (
    reporte_s2.columns.astype(str).str.strip().str.replace("\ufeff", "", regex=False)
)

COLUMNAS_S2 = {
    "categoria", "elemento", "resultado_validado", "estado", "uso_s3"
}
faltan_s2 = COLUMNAS_S2 - set(reporte_s2.columns)
if faltan_s2:
    raise ValueError("Columnas ausentes en el reporte S2: " + ", ".join(sorted(faltan_s2)))

estado_s2 = reporte_s2["estado"].astype(str).str.lower()
categoria_s2 = reporte_s2["categoria"].astype(str).str.lower()

evidencias_robustas_s2 = reporte_s2.loc[
    estado_s2.str.contains("robusta|validado|confiable|concordante", regex=True, na=False)
].copy()
evidencias_cautela_s2 = reporte_s2.loc[
    estado_s2.str.contains("cautela|sensible", regex=True, na=False)
].copy()
decisiones_s3 = reporte_s2.loc[
    categoria_s2.str.contains("decisión metodológica s3|decision metodologica s3", regex=True, na=False)
].copy()

resumen_s2 = pd.DataFrame([{
    "archivo": ruta_reporte_s2.name,
    "evidencias_totales": len(reporte_s2),
    "robustas_o_validadas": len(evidencias_robustas_s2),
    "sensibles_o_con_cautela": len(evidencias_cautela_s2),
    "decisiones_metodologicas_s3": len(decisiones_s3),
    "duplicados_exactos": int(reporte_s2.duplicated().sum()),
}])

resumen_s2.to_csv(
    RESULTS_DIR / "s3_trazabilidad_reporte_s2.csv", index=False, encoding="utf-8-sig"
)
decisiones_s3.to_csv(
    RESULTS_DIR / "s3_decisiones_metodologicas_importadas_desde_s2.csv",
    index=False,
    encoding="utf-8-sig",
)

mostrar_tabla(resumen_s2, "Auditoría del archivo puente S2→S3")
mostrar_tabla(
    reporte_s2[["categoria", "elemento", "resultado_validado", "estado", "uso_s3"]],
    "Evidencias metodológicas heredadas de Sumativa 2",
    max_filas=25,
)

mostrar_interpretacion(
    "✅ Trazabilidad formal incorporada",
    [
        f"Se importaron <b>{len(reporte_s2)}</b> evidencias de S2, de las cuales <b>{len(evidencias_robustas_s2)}</b> están marcadas como robustas o validadas.",
        f"Los <b>{len(evidencias_cautela_s2)}</b> resultados sensibles no se eliminan: se conservan como advertencias metodológicas.",
        "Este archivo guía el modelamiento, pero no forma parte de la matriz de entrenamiento porque no contiene observaciones estación-día.",
    ],
    "ok",
)
```

  <div style="color:#26364D !important;font-size:15px;">
    <p style="margin:0 0 10px 0;">La unidad analítica es <b>estación-día</b>. El objetivo es
    <code>mala_calidad_mp25</code>; se valida únicamente en registros con medición diaria disponible de MP2.5 y, cuando no existe, se reconstruye con reglas
    documentadas y nunca se imputa.</p>
    <p style="margin:0 0 10px 0;"><b>Horizonte declarado:</b> clasificación contemporánea
    (<i>nowcasting</i>). Por ello MP10 del mismo día puede evaluarse como predictor adicional,
    pero siempre se reporta como proxy altamente correlacionado y no como antecedente causal.</p>
    <p style="margin:0;">Se aplica una lista blanca de variables aprobadas y una detección por
    patrones para bloquear columnas derivadas del objetivo.</p>

```python
# =============================================================================
# 3.1 OBJETIVO BINARIO Y AUDITORÍA DE FUGA
# =============================================================================
def normalizar_binaria(serie):
    s = pd.Series(serie).copy()
    if pd.api.types.is_bool_dtype(s):
        return s.astype(float)
    if pd.api.types.is_numeric_dtype(s):
        out = pd.to_numeric(s, errors="coerce")
        return out.where(out.isin([0, 1]))
    mapa = {
        "1": 1, "0": 0, "si": 1, "sí": 1, "true": 1, "verdadero": 1,
        "no": 0, "false": 0, "falso": 0, "mala": 1, "buena": 0,
    }
    return s.astype(str).str.strip().str.lower().map(mapa)

TARGET = "mala_calidad_mp25"
HORIZONTE_MODELO = "nowcasting_contemporaneo"

if TARGET in df_raw.columns:
    y_all = normalizar_binaria(df_raw[TARGET])
    fuente_target = "columna original mala_calidad_mp25, validada sobre MP2.5 disponible"
elif "calidad_aire_mp25" in df_raw.columns:
    calidad = df_raw["calidad_aire_mp25"].astype(str).str.strip().str.lower()
    y_all = calidad.map(
        lambda x: 0 if x == "bueno" else (1 if x not in ["nan", "none", ""] else np.nan)
    )
    fuente_target = "derivada de calidad_aire_mp25"
elif "mp2_5_mean" in df_raw.columns:
    mp25 = pd.to_numeric(df_raw["mp2_5_mean"], errors="coerce")
    y_all = (mp25 > 50).astype(float)
    y_all[mp25.isna()] = np.nan
    fuente_target = "fallback: MP2.5 diario > 50 µg/m³"
else:
    raise ValueError("No existe una variable válida para construir mala_calidad_mp25.")

# El target solo es válido cuando existe una medición diaria de MP2.5.
# Esta regla replica el filtro inferencial aplicado en S1 y S2.
if "mp2_5_mean" in df_raw.columns:
    mp25_disponible = pd.to_numeric(
        df_raw["mp2_5_mean"], errors="coerce"
    ).notna()
    y_all = y_all.where(mp25_disponible)

df = df_raw.copy()
df[TARGET] = y_all
filas_sin_target = int(df[TARGET].isna().sum())
df = df.loc[df[TARGET].notna()].copy()
df[TARGET] = df[TARGET].astype(int)

PATRONES_FUGA = [
    r"^mp2[_\.]?5", r"mala_calidad", r"critico.*mp25", r"calidad_aire_mp25",
    r"nivel_contaminacion", r"episodio.*mp25", r"estaciones_en_episodio",
]
columnas_fuga = []
for columna in df.columns:
    nombre = str(columna).lower()
    if columna == TARGET:
        columnas_fuga.append((columna, "variable objetivo"))
    elif any(pd.Series([nombre]).str.contains(pat, regex=True).iloc[0] for pat in PATRONES_FUGA):
        columnas_fuga.append((columna, "derivada directa o proxy prohibido de MP2.5"))

auditoria_fuga = pd.DataFrame(columnas_fuga, columns=["variable", "motivo_exclusion"]).drop_duplicates()
auditoria_fuga.to_csv(
    RESULTS_DIR / "s3_auditoria_fuga_variables.csv", index=False, encoding="utf-8-sig"
)

auditoria_faltantes = pd.DataFrame({
    "variable": df_raw.columns,
    "n_faltantes": df_raw.isna().sum().values,
    "porcentaje_faltantes": 100 * df_raw.isna().mean().values,
    "tipo": [str(t) for t in df_raw.dtypes],
}).sort_values("porcentaje_faltantes", ascending=False)

resumen_target = pd.DataFrame([{
    "fuente_target": fuente_target,
    "horizonte": HORIZONTE_MODELO,
    "filas_originales": len(df_raw),
    "filas_sin_target_excluidas": filas_sin_target,
    "filas_modelables": len(df),
    "positivos": int(df[TARGET].sum()),
    "prevalencia": df[TARGET].mean(),
    "filas esperadas S1/S2": BASELINE_PROYECTO["filas_inferenciales"],
    "estado continuidad": (
        "Consistente"
        if len(df) == BASELINE_PROYECTO["filas_inferenciales"]
        else "Revisar: no coincide con la matriz inferencial S1/S2"
    ),
}])

auditoria_faltantes.to_csv(
    RESULTS_DIR / "s3_auditoria_faltantes.csv", index=False, encoding="utf-8-sig"
)
resumen_target.to_csv(
    RESULTS_DIR / "s3_resumen_target.csv", index=False, encoding="utf-8-sig"
)

mostrar_tabla(resumen_target, "Definición y disponibilidad de la variable objetivo")
mostrar_tabla(auditoria_fuga, "Variables bloqueadas por riesgo de fuga", max_filas=30)
mostrar_tabla(auditoria_faltantes.head(20), "Variables con mayor proporción de faltantes", max_filas=20)

top_faltantes = auditoria_faltantes.head(15).sort_values("porcentaje_faltantes")
fig, ax = plt.subplots(figsize=(11, 6))
ax.barh(top_faltantes["variable"], top_faltantes["porcentaje_faltantes"])
ax.axvline(27.8, linestyle="--", linewidth=1.6, label="Referencia S1 ≈ 27,8%")
ax.set_title("Proporción de datos faltantes por variable")
ax.set_xlabel("Porcentaje de faltantes")
ax.set_ylabel("")
ax.legend()
guardar_figura("01_faltantes_top15.png")
plt.show()

mostrar_interpretacion(
    "📌 Interpretación del objetivo y la fuga",
    [
        f"Se excluyeron <b>{filas_sin_target:,}</b> registros sin medición válida de MP2.5 o sin etiqueta; conservarlos habría generado negativos artificiales.",
        f"La matriz modelable queda en <b>{len(df):,}</b> registros, consistente con el filtro inferencial de S1/S2. La prevalencia positiva es <b>{100 * df[TARGET].mean():.2f}%</b>.",
        "La lista de exclusión se audita antes de seleccionar modelos. El sistema falla de forma explícita si no existen suficientes predictores aprobados.",
    ],
    "alerta" if df[TARGET].mean() < 0.20 else "info",
)
```

```python
# =============================================================================
# 4.1 ALIAS, VARIABLES DERIVADAS SEGURAS Y POOL DE CANDIDATAS
# =============================================================================
def primera_columna(columnas, candidatas):
    return next((c for c in candidatas if c in columnas), None)

ALIASES = {
    "temperatura": ["temperatura_mean", "temperatura_media", "temp_mean", "temp_media"],
    "humedad": ["humedad_mean", "humedad_media", "rh_mean"],
    "viento": ["viento_mean", "viento_media", "velocidad_viento_mean"],
    "radiacion": ["radiacion_mean", "radiacion_media", "radiacion_solar_mean"],
    "presion": ["presion_mean", "presion_media", "pressure_mean"],
    "inversion": ["prop_inversion", "inversion_termica", "alta_inversion", "inversion"],
    "mp10": ["mp10_mean", "mp10_media", "MP10"],
    "fin_semana": ["es_finde", "fin_semana", "es_fin_semana"],
    "festivo": ["es_festivo", "festivo"],
    "gec": ["periodo_gec", "periodo_GEC", "gec"],
    "cobertura": ["cobertura_horaria", "cobertura_mp25", "horas_validas_mp25", "n_horas_mp25", "horas_validas"],
}
columnas_disponibles = set(df.columns)
mapa = {k: primera_columna(columnas_disponibles, v) for k, v in ALIASES.items()}

df["mes"] = df[col_fecha].dt.month.astype(int)
df["mes_sin"] = np.sin(2 * np.pi * df["mes"] / 12)
df["mes_cos"] = np.cos(2 * np.pi * df["mes"] / 12)
df["invierno"] = df["mes"].isin([5, 6, 7, 8]).astype(int)
df["anio"] = df[col_fecha].dt.year.astype(int)

# ---------------------------------------------------------------------------
# Trazabilidad estructural entre las tres fases.
# 46 corresponde a la matriz analítica armonizada; las variables de calendario
# creadas exclusivamente para S3 se reportan por separado.
# ---------------------------------------------------------------------------
columnas_nuevas_s3 = [c for c in df.columns if c not in df_raw.columns]

consistencia_fases = pd.DataFrame([
    {
        "fase": "S1",
        "etapa": "Base enriquecida previa al filtro",
        "filas": BASELINE_PROYECTO["filas_base"],
        "variables": BASELINE_PROYECTO["columnas_enriquecidas"],
        "hilo conductor": "36 originales + 10 físicas derivadas",
    },
    {
        "fase": "S1",
        "etapa": "Matriz descriptiva e inferencial",
        "filas": BASELINE_PROYECTO["filas_inferenciales"],
        "variables": BASELINE_PROYECTO["columnas_enriquecidas"],
        "hilo conductor": "Solo MP2.5 diario disponible",
    },
    {
        "fase": "S2",
        "etapa": "Validación y remuestreo",
        "filas": BASELINE_PROYECTO["filas_inferenciales"],
        "variables": BASELINE_PROYECTO["columnas_enriquecidas"],
        "hilo conductor": "Misma matriz inferencial de S1",
    },
    {
        "fase": "S3",
        "etapa": "Matriz armonizada recibida",
        "filas": len(df),
        "variables": df_raw.shape[1],
        "hilo conductor": "Mismo filtro y mismas 46 variables",
    },
    {
        "fase": "S3",
        "etapa": "Matriz de trabajo para modelamiento",
        "filas": len(df),
        "variables": df.shape[1],
        "hilo conductor": (
            f"46 armonizadas + {len(columnas_nuevas_s3)} var. temp. S3"
        ),
    },
])

consistencia_fases["estado"] = [
    "Referencia",
    "Referencia",
    "Referencia",
    (
        "Consistente"
        if len(df) == BASELINE_PROYECTO["filas_inferenciales"]
        and df_raw.shape[1] == BASELINE_PROYECTO["columnas_enriquecidas"]
        else "Revisar"
    ),
    "Extensión documentada",
]

consistencia_fases.to_csv(
    RESULTS_DIR / "s3_matriz_consistencia_s1_s2_s3.csv",
    index=False,
    encoding="utf-8-sig",
)

mostrar_tabla(
    consistencia_fases,
    "Hilo conductor y consistencia estructural S1 → S2 → S3",
    max_filas=10,
)

mostrar_interpretacion(
    "🔗 Lectura correcta del número de variables",
    [
        "<b>46 variables</b> es el inventario armonizado compartido por S1, S2 y la entrada de S3.",
        f"S3 añade <b>{len(columnas_nuevas_s3)}</b> variables de calendario para el modelamiento; "
        "se reportan como ingeniería exclusiva de esta fase y no reescriben la estructura histórica.",
        "Cada modelo utiliza solo un subconjunto aprobado de predictores. Por ello, tres variables "
        "en un modelo seleccionado no significa que el proyecto tenga tres variables.",
    ],
    "ok",
)

for clave in ["fin_semana", "festivo", "gec"]:
    columna = mapa.get(clave)
    if columna is not None:
        df[columna] = normalizar_binaria(df[columna])

categoricas_candidatas = [
    c for c in [
        "zona_geografica", "nivel_socioeconomico", "estacion_anio",
        "temporada_critica", "comuna", "estacion"
    ]
    if c in df.columns and 1 < df[c].nunique(dropna=True) <= 25
]

meteo_validada = [
    mapa[k] for k in ["temperatura", "humedad", "viento", "radiacion", "inversion"]
    if mapa[k] is not None
]
meteo_ampliada = meteo_validada + ([mapa["presion"]] if mapa["presion"] is not None else [])
temporales = ["mes_sin", "mes_cos", "invierno"] + [
    mapa[k] for k in ["fin_semana", "festivo", "gec"] if mapa[k] is not None
]
nowcasting = [
    mapa[k] for k in ["mp10", "cobertura"] if mapa[k] is not None
]

columnas_bloqueadas = set(auditoria_fuga["variable"].tolist()) | {col_fecha, "anio", "mes"}

def aprobar_variables(lista):
    salida = []
    for c in lista:
        if c is None or c not in df.columns or c in columnas_bloqueadas:
            continue
        if c not in salida:
            salida.append(c)
    return salida

VARIABLES_DOMINIO = aprobar_variables(
    meteo_validada + temporales + categoricas_candidatas
)
VARIABLES_COMPLETAS = aprobar_variables(
    meteo_ampliada + temporales + categoricas_candidatas + nowcasting
)

if len(VARIABLES_DOMINIO) < 3:
    raise ValueError(
        "El modelo de dominio no dispone de al menos tres variables aprobadas. "
        "Revise los alias y la estructura del dataset."
    )
if len(VARIABLES_COMPLETAS) < len(VARIABLES_DOMINIO):
    raise ValueError("El pool completo no puede ser menor que el modelo de dominio.")

matriz_candidatas = []
for c in VARIABLES_COMPLETAS:
    familia = (
        "meteorológica" if c in meteo_ampliada else
        "temporal" if c in temporales else
        "territorial" if c in categoricas_candidatas else
        "nowcasting/calidad" if c in nowcasting else "otra"
    )
    matriz_candidatas.append({
        "variable": c,
        "familia": familia,
        "tipo": "numérica" if pd.api.types.is_numeric_dtype(df[c]) else "categórica",
        "faltantes_%": 100 * df[c].isna().mean(),
        "modelo_dominio": c in VARIABLES_DOMINIO,
        "pool_seleccion": True,
    })

matriz_candidatas = pd.DataFrame(matriz_candidatas)
matriz_candidatas.to_csv(
    RESULTS_DIR / "s3_pool_variables_aprobadas.csv", index=False, encoding="utf-8-sig"
)
mostrar_tabla(matriz_candidatas, "Lista blanca de variables candidatas", max_filas=40)

mostrar_interpretacion(
    "✅ Ingeniería de variables controlada",
    [
        f"El modelo de dominio utiliza <b>{len(VARIABLES_DOMINIO)}</b> variables y el pool de selección contiene <b>{len(VARIABLES_COMPLETAS)}</b>.",
        "Ninguna columna se incorpora por posición o por tipo de dato sin una justificación explícita.",
        "La eventual presencia de MP10 se interpreta como apoyo al <i>nowcasting</i>; no convierte el resultado en una predicción anticipada.",
    ],
    "ok",
)
```

```python
# =============================================================================
# 5.1 SPLIT TEMPORAL 70/30 Y DESARROLLO/VALIDACIÓN
# =============================================================================
fechas_unicas = np.array(sorted(df[col_fecha].dropna().dt.normalize().unique()))
if len(fechas_unicas) < 20:
    raise ValueError("Se requieren al menos 20 fechas distintas para una validación temporal robusta.")

indice_test = min(max(1, int(len(fechas_unicas) * 0.70)), len(fechas_unicas) - 1)
corte_test = fechas_unicas[indice_test]

train_raw = df.loc[df[col_fecha].dt.normalize() < corte_test].copy()
test_raw = df.loc[df[col_fecha].dt.normalize() >= corte_test].copy()

fechas_train = np.array(sorted(train_raw[col_fecha].dt.normalize().unique()))
indice_val = min(max(1, int(len(fechas_train) * 0.80)), len(fechas_train) - 1)
corte_val = fechas_train[indice_val]

dev_raw = train_raw.loc[train_raw[col_fecha].dt.normalize() < corte_val].copy()
val_raw = train_raw.loc[train_raw[col_fecha].dt.normalize() >= corte_val].copy()

for nombre, parte in {
    "desarrollo": dev_raw, "validación": val_raw, "prueba": test_raw
}.items():
    if len(parte) == 0 or parte[TARGET].nunique() < 2:
        raise ValueError(f"La partición {nombre} quedó vacía o con una sola clase.")

particiones = pd.DataFrame([
    {
        "particion": "desarrollo",
        "n": len(dev_raw),
        "fecha_min": dev_raw[col_fecha].min(),
        "fecha_max": dev_raw[col_fecha].max(),
        "prevalencia": dev_raw[TARGET].mean(),
    },
    {
        "particion": "validacion",
        "n": len(val_raw),
        "fecha_min": val_raw[col_fecha].min(),
        "fecha_max": val_raw[col_fecha].max(),
        "prevalencia": val_raw[TARGET].mean(),
    },
    {
        "particion": "entrenamiento_total_70%",
        "n": len(train_raw),
        "fecha_min": train_raw[col_fecha].min(),
        "fecha_max": train_raw[col_fecha].max(),
        "prevalencia": train_raw[TARGET].mean(),
    },
    {
        "particion": "prueba_futura_30%",
        "n": len(test_raw),
        "fecha_min": test_raw[col_fecha].min(),
        "fecha_max": test_raw[col_fecha].max(),
        "prevalencia": test_raw[TARGET].mean(),
    },
])
particiones.to_csv(
    RESULTS_DIR / "s3_particiones_temporales.csv", index=False, encoding="utf-8-sig"
)

mostrar_tabla(particiones, "Particiones cronológicas del proyecto")
mostrar_kpis(
    "Control de separación",
    [
        ("Entrenamiento", f"{len(train_raw):,}", "70% inicial de fechas"),
        ("Prueba", f"{len(test_raw):,}", "30% final bloqueado"),
        ("Corte test", pd.Timestamp(corte_test).strftime("%Y-%m-%d"), "Primera fecha de prueba"),
        ("Prevalencia test", f"{100 * test_raw[TARGET].mean():.2f}%", "Clase positiva"),
    ],
)
mostrar_interpretacion(
    "🔒 Test bloqueado",
    [
        "Hasta la sección de evaluación final, ninguna métrica de prueba participa en la selección.",
        "La separación por fechas evita que observaciones futuras de una estación influyan en el aprendizaje del pasado.",
    ],
    "ok",
)
```

  <div style="color:#26364D !important;font-size:15px;">
    <p style="margin:0 0 10px 0;">Se comparan <b>casos completos</b>, <b>mediana/moda</b> e
    <b>imputación multivariada por regresión</b> mediante <code>IterativeImputer</code> con
    <code>BayesianRidge</code>.</p>
    <p style="margin:0 0 10px 0;">Antes de imputar se analiza si la ausencia puede explicarse por
    variables observadas. Un AUC superior a 0,50 no demuestra MAR, pero hace menos plausible MCAR.
    MNAR no puede descartarse solo con la información disponible.</p>
    <p style="margin:0;">La regresión de imputación se documenta también con OLS explícita:
    R², residuos, normalidad, heterocedasticidad y encogimiento de varianza.</p>
    </div>
</div>

```python
# =============================================================================
# 6.1 PREPROCESADORES Y DIAGNÓSTICO DEL MECANISMO DE AUSENCIA
# =============================================================================
def crear_ohe():
    try:
        return OneHotEncoder(handle_unknown="ignore", drop="first", sparse_output=False)
    except TypeError:
        return OneHotEncoder(handle_unknown="ignore", drop="first", sparse=False)

def separar_tipos(frame, features):
    categoricas = [c for c in features if not pd.api.types.is_numeric_dtype(frame[c])]
    numericas = [c for c in features if c not in categoricas]
    return numericas, categoricas

def construir_preprocesador(frame, features, estrategia):
    numericas, categoricas = separar_tipos(frame, features)
    transformadores = []

    if numericas:
        if estrategia == "mediana":
            imp_num = SimpleImputer(strategy="median")
        elif estrategia == "regresion":
            imp_num = IterativeImputer(
                estimator=BayesianRidge(),
                max_iter=25,
                tol=1e-3,
                random_state=SEED,
                sample_posterior=False,
                initial_strategy="median",
                skip_complete=True,
            )
        elif estrategia == "completos":
            imp_num = "passthrough"
        else:
            raise ValueError(f"Estrategia desconocida: {estrategia}")

        pasos_num = []
        if imp_num != "passthrough":
            pasos_num.append(("imputer", imp_num))
        pasos_num.append(("scaler", StandardScaler()))
        transformadores.append(("num", Pipeline(pasos_num), numericas))

    if categoricas:
        pasos_cat = []
        if estrategia != "completos":
            pasos_cat.append(("imputer", SimpleImputer(strategy="most_frequent")))
        pasos_cat.append(("ohe", crear_ohe()))
        transformadores.append(("cat", Pipeline(pasos_cat), categoricas))

    return (
        ColumnTransformer(
            transformadores,
            remainder="drop",
            verbose_feature_names_out=False,
        ),
        numericas,
        categoricas,
    )

def filtrar_completos(frame, features):
    return frame.dropna(subset=list(features) + [TARGET]).copy()

def pipeline_predictivo(frame, features, estrategia):
    prep, nums, cats = construir_preprocesador(frame, features, estrategia)
    clf = LogisticRegression(
        penalty="l2",
        C=1e6,
        solver="lbfgs",
        max_iter=5000,
        class_weight="balanced",
        random_state=SEED,
    )
    return Pipeline([("prep", prep), ("clf", clf)]), nums, cats

variables_con_faltantes = [
    c for c in VARIABLES_COMPLETAS
    if 0 < dev_raw[c].isna().mean() < 0.95
]
diagnostico_mecanismo = []

for variable in variables_con_faltantes:
    predictores = [
        c for c in VARIABLES_COMPLETAS
        if c != variable and dev_raw[c].isna().mean() < 0.50
    ][:12]
    if len(predictores) < 2:
        continue

    y_miss_dev = dev_raw[variable].isna().astype(int)
    y_miss_val = val_raw[variable].isna().astype(int)
    if y_miss_dev.nunique() < 2 or y_miss_val.nunique() < 2:
        continue

    prep_miss, _, _ = construir_preprocesador(dev_raw, predictores, "mediana")
    modelo_miss = Pipeline([
        ("prep", prep_miss),
        ("clf", LogisticRegression(max_iter=3000, class_weight="balanced", random_state=SEED)),
    ])
    modelo_miss.fit(dev_raw[predictores], y_miss_dev)
    prob_miss = modelo_miss.predict_proba(val_raw[predictores])[:, 1]
    auc_miss = roc_auc_score(y_miss_val, prob_miss)

    if auc_miss < 0.55:
        lectura = "Sin evidencia predictiva relevante; MCAR sigue siendo plausible"
    elif auc_miss < 0.70:
        lectura = "Ausencia parcialmente explicada por observables; MAR es plausible"
    else:
        lectura = "Ausencia claramente asociada a observables; MAR es más defendible"

    diagnostico_mecanismo.append({
        "variable": variable,
        "faltantes_dev_%": 100 * dev_raw[variable].isna().mean(),
        "faltantes_val_%": 100 * val_raw[variable].isna().mean(),
        "auc_prediccion_ausencia": auc_miss,
        "lectura": lectura,
        "advertencia": "MNAR no puede descartarse",
    })

diagnostico_mecanismo = pd.DataFrame(diagnostico_mecanismo)
diagnostico_mecanismo.to_csv(
    RESULTS_DIR / "s3_diagnostico_mecanismo_faltantes.csv",
    index=False,
    encoding="utf-8-sig",
)

if diagnostico_mecanismo.empty:
    mostrar_interpretacion(
        "Mecanismo de ausencia",
        "No hubo variables con suficientes faltantes en desarrollo y validación para estimar un clasificador estable de ausencia.",
        "alerta",
    )
else:
    mostrar_tabla(
        diagnostico_mecanismo.sort_values("auc_prediccion_ausencia", ascending=False),
        "Evidencia empírica sobre el mecanismo de faltantes",
        max_filas=30,
    )
    auc_max = diagnostico_mecanismo["auc_prediccion_ausencia"].max()
    mostrar_interpretacion(
        "📌 Interpretación MCAR–MAR–MNAR",
        [
            f"El mayor AUC para explicar la ausencia fue <b>{auc_max:.3f}</b>.",
            "AUC elevado favorece una lectura MAR porque la ausencia depende de variables observadas; no constituye una prueba formal del mecanismo.",
            "MNAR permanece como limitación no verificable con los datos observados y se declara explícitamente.",
        ],
        "info",
    )
```

```python
# =============================================================================
# 6.2 REGRESIONES OLS EXPLÍCITAS PARA DOCUMENTAR LA IMPUTACIÓN
# =============================================================================
numericas_pool = [
    c for c in VARIABLES_COMPLETAS
    if pd.api.types.is_numeric_dtype(dev_raw[c])
]
variables_ols = [
    c for c in numericas_pool
    if 0 < dev_raw[c].isna().mean() < 0.80 and dev_raw[c].notna().sum() >= 100
]

diagnosticos_ols = []
modelos_ols_imputacion = {}

for objetivo_imp in variables_ols:
    candidatos = [c for c in numericas_pool if c != objetivo_imp]
    correlaciones = {}
    for pred in candidatos:
        pares = dev_raw[[objetivo_imp, pred]].apply(pd.to_numeric, errors="coerce").dropna()
        if len(pares) >= 80 and pares[objetivo_imp].nunique() > 5 and pares[pred].nunique() > 3:
            correlaciones[pred] = abs(pares.corr().iloc[0, 1])

    predictores = [
        k for k, _ in sorted(correlaciones.items(), key=lambda item: item[1], reverse=True)[:5]
    ]
    if len(predictores) < 2:
        continue

    base = dev_raw.loc[dev_raw[objetivo_imp].notna(), [objetivo_imp] + predictores].copy()
    y_imp = pd.to_numeric(base[objetivo_imp], errors="coerce")
    X_imp = base[predictores].apply(pd.to_numeric, errors="coerce")
    medianas = X_imp.median()
    X_imp = X_imp.fillna(medianas)
    desvios = X_imp.std(ddof=0).replace(0, 1)
    X_std = (X_imp - X_imp.mean()) / desvios
    X_sm = sm.add_constant(X_std, has_constant="add")

    ajuste = sm.OLS(y_imp, X_sm, missing="drop").fit()
    resid = ajuste.resid
    bp = het_breuschpagan(resid, ajuste.model.exog)
    jb = stats.jarque_bera(resid)

    modelos_ols_imputacion[objetivo_imp] = {
        "ajuste": ajuste,
        "predictores": predictores,
        "medianas": medianas,
        "medias": X_imp.mean(),
        "desvios": desvios,
    }

    diagnosticos_ols.append({
        "variable_imputada": objetivo_imp,
        "predictores": ", ".join(predictores),
        "n_observado": int(ajuste.nobs),
        "r2": ajuste.rsquared,
        "r2_ajustado": ajuste.rsquared_adj,
        "rmse_residual": float(np.sqrt(np.mean(resid ** 2))),
        "media_residuo": float(np.mean(resid)),
        "asimetria_residuo": float(stats.skew(resid, nan_policy="omit")),
        "p_jarque_bera": float(jb.pvalue),
        "p_breusch_pagan": float(bp[1]),
    })

diagnosticos_ols = pd.DataFrame(diagnosticos_ols)
diagnosticos_ols.to_csv(
    RESULTS_DIR / "s3_diagnostico_ols_imputacion.csv",
    index=False,
    encoding="utf-8-sig",
)

if diagnosticos_ols.empty:
    mostrar_interpretacion(
        "Regresión lineal de imputación",
        "No se identificaron variables numéricas faltantes con soporte suficiente para ajustar OLS explícita.",
        "alerta",
    )
else:
    mostrar_tabla(
        diagnosticos_ols.sort_values("r2", ascending=False),
        "Diagnóstico de las regresiones lineales de imputación",
        max_filas=30,
    )
    mejor_r2 = diagnosticos_ols["r2"].max()
    peor_r2 = diagnosticos_ols["r2"].min()
    mostrar_interpretacion(
        "📌 Interpretación de las regresiones de imputación",
        [
            f"Los R² se ubican entre <b>{peor_r2:.3f}</b> y <b>{mejor_r2:.3f}</b>; valores bajos anticipan imputaciones con capacidad limitada para recuperar variabilidad individual.",
            "La falta de normalidad de los residuos no invalida la predicción puntual, pero sí desaconseja sobreinterpretar la incertidumbre paramétrica de estas regresiones.",
            "La estrategia definitiva se decide por validación de imputación, retención y efecto sobre el modelo final, no solo por R².",
        ],
        "info",
    )
```

```python
# =============================================================================
# 6.3 RETENCIÓN, ENMASCARAMIENTO, NRMSE Y ENCOGIMIENTO DE VARIANZA
# =============================================================================
ESTRATEGIAS = ["completos", "mediana", "regresion"]

retencion = []
for estrategia in ESTRATEGIAS:
    usado_dev = (
        filtrar_completos(dev_raw, VARIABLES_COMPLETAS)
        if estrategia == "completos"
        else dev_raw
    )
    usado_val = (
        filtrar_completos(val_raw, VARIABLES_COMPLETAS)
        if estrategia == "completos"
        else val_raw
    )
    retencion.extend([
        {
            "estrategia": estrategia,
            "particion": "desarrollo",
            "n_disponible": len(dev_raw),
            "n_utilizado": len(usado_dev),
            "retencion": len(usado_dev) / max(len(dev_raw), 1),
            "prevalencia": usado_dev[TARGET].mean(),
        },
        {
            "estrategia": estrategia,
            "particion": "validacion",
            "n_disponible": len(val_raw),
            "n_utilizado": len(usado_val),
            "retencion": len(usado_val) / max(len(val_raw), 1),
            "prevalencia": usado_val[TARGET].mean(),
        },
    ])

retencion = pd.DataFrame(retencion)
retencion.to_csv(
    RESULTS_DIR / "s3_retencion_por_estrategia.csv", index=False, encoding="utf-8-sig"
)

nums_validables = [
    c for c in numericas_pool
    if dev_raw[c].notna().sum() >= 120 and dev_raw[c].nunique(dropna=True) > 5
]

resultados_mask = []
if len(nums_validables) >= 2:
    base_num = dev_raw[nums_validables].apply(pd.to_numeric, errors="coerce")

    for repeticion in range(N_MASK_REPEATS):
        rng_rep = np.random.default_rng(SEED + repeticion)
        masked = base_num.copy()
        mascaras = {}

        for c in nums_validables:
            idx_obs = np.flatnonzero(base_num[c].notna().to_numpy())
            n_mask = max(1, int(len(idx_obs) * 0.10))
            idx_mask = rng_rep.choice(idx_obs, size=n_mask, replace=False)
            mascara = np.zeros(len(base_num), dtype=bool)
            mascara[idx_mask] = True
            mascaras[c] = mascara
            masked.loc[mascara, c] = np.nan

        imputadores = {
            "mediana": SimpleImputer(strategy="median"),
            "regresion": IterativeImputer(
                estimator=BayesianRidge(),
                max_iter=25,
                random_state=SEED + repeticion,
                initial_strategy="median",
                sample_posterior=False,
                skip_complete=False,
            ),
        }

        for metodo, imputador in imputadores.items():
            imputado = pd.DataFrame(
                imputador.fit_transform(masked),
                columns=nums_validables,
                index=base_num.index,
            )

            for c in nums_validables:
                mascara = mascaras[c]
                real = base_num.loc[mascara, c].astype(float)
                pred = imputado.loc[mascara, c].astype(float)
                iqr = float(real.quantile(0.75) - real.quantile(0.25))
                escala = iqr if iqr > 1e-12 else float(real.std(ddof=1))
                escala = escala if escala > 1e-12 else 1.0

                mae = float(np.mean(np.abs(real - pred)))
                rmse = float(np.sqrt(np.mean((real - pred) ** 2)))
                sd_real = float(real.std(ddof=1))
                sd_pred = float(pred.std(ddof=1))
                shrink = 1 - (sd_pred / sd_real) if sd_real > 1e-12 else np.nan

                resultados_mask.append({
                    "repeticion": repeticion + 1,
                    "variable": c,
                    "metodo": metodo,
                    "n_enmascarado": int(mascara.sum()),
                    "mae": mae,
                    "rmse": rmse,
                    "nmae_iqr": mae / escala,
                    "nrmse_iqr": rmse / escala,
                    "sd_real": sd_real,
                    "sd_imputado": sd_pred,
                    "encogimiento_varianza": shrink,
                })

validacion_imputacion = pd.DataFrame(resultados_mask)

if validacion_imputacion.empty:
    resumen_imputacion = pd.DataFrame()
    resumen_global_imputacion = pd.DataFrame()
else:
    resumen_imputacion = (
        validacion_imputacion
        .groupby(["variable", "metodo"], as_index=False)
        .agg(
            n_enmascarado=("n_enmascarado", "sum"),
            nmae_iqr=("nmae_iqr", "mean"),
            nrmse_iqr=("nrmse_iqr", "mean"),
            encogimiento_varianza=("encogimiento_varianza", "mean"),
            sd_real=("sd_real", "mean"),
            sd_imputado=("sd_imputado", "mean"),
        )
    )
    resumen_global_imputacion = (
        resumen_imputacion
        .groupby("metodo", as_index=False)
        .agg(
            nmae_iqr_promedio=("nmae_iqr", "mean"),
            nrmse_iqr_promedio=("nrmse_iqr", "mean"),
            encogimiento_promedio=("encogimiento_varianza", "mean"),
        )
    )

resumen_imputacion.to_csv(
    RESULTS_DIR / "s3_validacion_imputacion_por_variable.csv",
    index=False,
    encoding="utf-8-sig",
)
resumen_global_imputacion.to_csv(
    RESULTS_DIR / "s3_validacion_imputacion_global.csv",
    index=False,
    encoding="utf-8-sig",
)

mostrar_tabla(retencion, "Retención de muestra por estrategia")
if not resumen_global_imputacion.empty:
    mostrar_tabla(
        resumen_global_imputacion,
        "Calidad global de imputación en escala comparable",
    )
    mostrar_tabla(
        resumen_imputacion.sort_values(["variable", "nrmse_iqr"]),
        "Diagnóstico por variable: NRMSE y encogimiento",
        max_filas=40,
    )

    mejor_metodo_imp = (
        resumen_global_imputacion
        .sort_values(["nrmse_iqr_promedio", "encogimiento_promedio"])
        .iloc[0]["metodo"]
    )
    mostrar_interpretacion(
        "📌 Interpretación de la imputación",
        [
            f"La menor pérdida normalizada por IQR corresponde a <b>{mejor_metodo_imp}</b>.",
            "El NRMSE normalizado evita promediar errores incompatibles entre temperatura, presión, radiación u otras escalas.",
            "Un encogimiento alto indica que los valores imputados son demasiado homogéneos; esta limitación debe contrastarse con estabilidad de coeficientes y desempeño.",
        ],
        "info",
    )
else:
    mostrar_interpretacion(
        "Validación de imputación",
        "No fue posible ejecutar el enmascaramiento porque no existen al menos dos variables numéricas con observaciones suficientes.",
        "alerta",
    )
```

```python
# =============================================================================
# 7.1 AJUSTE GLM PARA AIC/BIC Y SELECCIÓN POR GRUPOS DE VARIABLES
# =============================================================================
def preparar_glm(frame, features, estrategia="regresion"):
    datos = (
        filtrar_completos(frame, features)
        if estrategia == "completos"
        else frame.copy()
    )
    if len(datos) == 0 or datos[TARGET].nunique() < 2:
        raise ValueError("No existen datos suficientes para ajustar el GLM.")

    prep, nums, cats = construir_preprocesador(datos, features, estrategia)
    X = np.asarray(prep.fit_transform(datos[features]), dtype=float)
    nombres = np.asarray(prep.get_feature_names_out(), dtype=str)

    varianzas = np.nanvar(X, axis=0)
    mantener = np.flatnonzero(varianzas > 1e-12)
    X = X[:, mantener]
    nombres = nombres[mantener]

    Xb = sm.add_constant(X, has_constant="add")
    y = datos[TARGET].to_numpy(dtype=int)
    ajuste = sm.GLM(y, Xb, family=sm.families.Binomial()).fit(maxiter=200, disp=0)

    k = len(ajuste.params)
    bic_ll = -2 * ajuste.llf + k * np.log(ajuste.nobs)

    return {
        "ajuste": ajuste,
        "aic": float(ajuste.aic),
        "bic": float(bic_ll),
        "prep": prep,
        "X": X,
        "y": y,
        "nombres": nombres,
        "datos": datos,
        "nums": nums,
        "cats": cats,
    }

def seleccion_forward(frame, candidatas, mejora_minima=2.0):
    restantes = list(candidatas)
    seleccionadas = []
    trayectoria = []
    bic_actual = np.inf
    paso = 0

    while restantes:
        pruebas = []
        for candidata in restantes:
            subset = seleccionadas + [candidata]
            try:
                fit = preparar_glm(frame, subset, "regresion")
                pruebas.append((fit["bic"], fit["aic"], candidata, fit))
            except Exception:
                continue

        if not pruebas:
            break

        bic_nuevo, aic_nuevo, candidata, fit = min(pruebas, key=lambda x: x[0])
        mejora = bic_actual - bic_nuevo if np.isfinite(bic_actual) else np.inf

        if len(seleccionadas) >= 3 and mejora < mejora_minima:
            break

        seleccionadas.append(candidata)
        restantes.remove(candidata)
        paso += 1
        trayectoria.append({
            "paso": paso,
            "accion": "incorporar",
            "variable": candidata,
            "n_variables": len(seleccionadas),
            "AIC": aic_nuevo,
            "BIC": bic_nuevo,
            "mejora_BIC": mejora,
        })
        bic_actual = bic_nuevo

    return seleccionadas, pd.DataFrame(trayectoria)

def seleccion_backward(frame, candidatas, mejora_minima=2.0, evidencia_debil=6.0):
    seleccionadas = list(candidatas)
    ajuste_actual = preparar_glm(frame, seleccionadas, "regresion")
    bic_actual = ajuste_actual["bic"]
    trayectoria = [{
        "paso": 0,
        "accion": "modelo_completo",
        "variable": "—",
        "n_variables": len(seleccionadas),
        "AIC": ajuste_actual["aic"],
        "BIC": bic_actual,
        "delta_BIC": 0.0,
        "regla": "inicio",
    }]
    objetivo_max = max(4, int(np.ceil(0.70 * len(seleccionadas))))
    paso = 0

    while len(seleccionadas) > 3:
        pruebas = []
        for candidata in seleccionadas:
            subset = [c for c in seleccionadas if c != candidata]
            try:
                fit = preparar_glm(frame, subset, "regresion")
                pruebas.append((fit["bic"], fit["aic"], candidata, fit))
            except Exception:
                continue

        if not pruebas:
            break

        bic_nuevo, aic_nuevo, candidata, fit = min(pruebas, key=lambda x: x[0])
        delta = bic_nuevo - bic_actual
        aceptar = delta <= -mejora_minima
        regla = "mejora BIC"

        if not aceptar and len(seleccionadas) > objetivo_max and delta <= evidencia_debil:
            aceptar = True
            regla = "parsimonia con ΔBIC ≤ 6"

        if not aceptar:
            break

        seleccionadas.remove(candidata)
        bic_actual = bic_nuevo
        paso += 1
        trayectoria.append({
            "paso": paso,
            "accion": "eliminar",
            "variable": candidata,
            "n_variables": len(seleccionadas),
            "AIC": aic_nuevo,
            "BIC": bic_nuevo,
            "delta_BIC": delta,
            "regla": regla,
        })

    return seleccionadas, pd.DataFrame(trayectoria)

M1_DOMINIO = list(VARIABLES_DOMINIO)
M2_FORWARD, trayectoria_forward = seleccion_forward(dev_raw, VARIABLES_COMPLETAS)
M3_BACKWARD, trayectoria_backward = seleccion_backward(dev_raw, VARIABLES_COMPLETAS)

if len(M2_FORWARD) < 3:
    raise ValueError("La selección forward produjo menos de tres variables.")
if len(M3_BACKWARD) < 3:
    raise ValueError("La selección backward produjo menos de tres variables.")

MODELOS = {
    "M1_Dominio_S1_S2": M1_DOMINIO,
    "M2_Forward_AIC_BIC": M2_FORWARD,
    "M3_Backward_Parsimonioso": M3_BACKWARD,
}

matriz_modelos = []
for nombre, variables in MODELOS.items():
    for variable in variables:
        matriz_modelos.append({
            "modelo": nombre,
            "variable": variable,
            "familia": matriz_candidatas.set_index("variable").loc[variable, "familia"],
            "faltantes_%": 100 * df[variable].isna().mean(),
        })

matriz_modelos = pd.DataFrame(matriz_modelos)
matriz_modelos.to_csv(
    RESULTS_DIR / "s3_matriz_variables_modelos.csv", index=False, encoding="utf-8-sig"
)
trayectoria_forward.to_csv(
    RESULTS_DIR / "s3_trayectoria_forward_aic_bic.csv", index=False, encoding="utf-8-sig"
)
trayectoria_backward.to_csv(
    RESULTS_DIR / "s3_trayectoria_backward_bic.csv", index=False, encoding="utf-8-sig"
)

mostrar_tabla(matriz_modelos, "Variables congeladas en los tres modelos", max_filas=60)
mostrar_tabla(trayectoria_forward, "Trayectoria M2 · forward AIC/BIC", max_filas=40)
mostrar_tabla(trayectoria_backward, "Trayectoria M3 · backward parsimonioso", max_filas=40)

mostrar_interpretacion(
    "📌 Interpretación de la selección",
    [
        f"M1 conserva <b>{len(M1_DOMINIO)}</b> variables sustentadas por S1/S2.",
        f"M2 retuvo <b>{len(M2_FORWARD)}</b> variables mediante incorporación forward y criterio BIC.",
        f"M3 quedó con <b>{len(M3_BACKWARD)}</b> variables después de la reducción backward.",
        "Si M2 y M3 convergen al mismo subconjunto, la coincidencia se reporta como evidencia de estabilidad de la selección, no como un error.",
    ],
    "info",
)
```

  <h2 style="box-sizing:border-box;margin:0 0 12px 0;color:#132E51 !important; font-size:24px;line-height:1.25;font-weight:900;">8. Validación temporal, nueve combinaciones y selección sin observar test</h2>
  <div style="color:#26364D !important;font-size:15px;">
    <p style="margin:0 0 10px 0;">Cada modelo se evalúa con casos completos, mediana y regresión.
    Los folds temporales producen probabilidades fuera de muestra para elegir el umbral. Luego el
    bloque de validación compara las combinaciones.</p>
    <p style="margin:0;">La regla es jerárquica: PR-AUC temporal, Recall mínimo, F2, calibración,
    retención y parsimonia. No se usa una suma arbitraria de métricas correlacionadas.</p>
    </div>
</div>

```python
# =============================================================================
# 8.1 FOLDS TEMPORALES, UMBRAL OOF Y MÉTRICAS
# =============================================================================
def seleccionar_umbral(y_true, prob, costo_fn=COSTO_FN, costo_fp=COSTO_FP):
    filas = []
    for umbral in np.linspace(0.02, 0.98, 193):
        pred = (prob >= umbral).astype(int)
        tn, fp, fn, tp = confusion_matrix(y_true, pred, labels=[0, 1]).ravel()
        filas.append({
            "umbral": umbral,
            "precision": precision_score(y_true, pred, zero_division=0),
            "recall": recall_score(y_true, pred, zero_division=0),
            "f1": f1_score(y_true, pred, zero_division=0),
            "f2": fbeta_score(y_true, pred, beta=2, zero_division=0),
            "costo": costo_fn * fn + costo_fp * fp,
            "tn": tn, "fp": fp, "fn": fn, "tp": tp,
        })
    tabla = pd.DataFrame(filas)
    elegible = tabla.loc[tabla["recall"] >= MIN_RECALL]
    if elegible.empty:
        elegible = tabla
    mejor = elegible.sort_values(
        ["f2", "costo", "recall", "precision"],
        ascending=[False, True, False, False],
    ).iloc[0]
    return float(mejor["umbral"]), tabla

def calcular_metricas(y_true, prob, umbral):
    pred = (prob >= umbral).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_true, pred, labels=[0, 1]).ravel()
    especificidad = tn / (tn + fp) if (tn + fp) else np.nan
    return {
        "n": len(y_true),
        "positivos": int(np.sum(y_true)),
        "prevalencia": float(np.mean(y_true)),
        "umbral": float(umbral),
        "accuracy": accuracy_score(y_true, pred),
        "balanced_accuracy": balanced_accuracy_score(y_true, pred),
        "precision": precision_score(y_true, pred, zero_division=0),
        "recall": recall_score(y_true, pred, zero_division=0),
        "specificity": especificidad,
        "f1": f1_score(y_true, pred, zero_division=0),
        "f2": fbeta_score(y_true, pred, beta=2, zero_division=0),
        "roc_auc": roc_auc_score(y_true, prob),
        "pr_auc": average_precision_score(y_true, prob),
        "brier": brier_score_loss(y_true, prob),
        "tn": tn, "fp": fp, "fn": fn, "tp": tp,
    }

def folds_temporales(frame, n_splits=N_CV_SPLITS):
    fechas = np.array(sorted(frame[col_fecha].dt.normalize().unique()))
    if len(fechas) < n_splits + 2:
        raise ValueError("No existen fechas suficientes para los folds temporales.")
    bloques = [b for b in np.array_split(fechas, n_splits + 1) if len(b)]
    for i in range(1, len(bloques)):
        fechas_train_fold = np.concatenate(bloques[:i])
        fechas_val_fold = bloques[i]
        idx_train = frame[col_fecha].dt.normalize().isin(fechas_train_fold)
        idx_val = frame[col_fecha].dt.normalize().isin(fechas_val_fold)
        yield i, frame.loc[idx_train].copy(), frame.loc[idx_val].copy()

def cv_temporal_oof(frame, features, estrategia):
    filas = []
    y_oof = []
    p_oof = []

    for fold, tr, va in folds_temporales(frame):
        if estrategia == "completos":
            tr = filtrar_completos(tr, features)
            va = filtrar_completos(va, features)
        if len(tr) == 0 or len(va) == 0:
            continue
        if tr[TARGET].nunique() < 2 or va[TARGET].nunique() < 2:
            continue

        pipe, _, _ = pipeline_predictivo(tr, features, estrategia)
        pipe.fit(tr[features], tr[TARGET])
        prob = pipe.predict_proba(va[features])[:, 1]
        y_fold = va[TARGET].to_numpy(dtype=int)

        filas.append({
            "fold": fold,
            "n_train": len(tr),
            "n_validacion": len(va),
            "roc_auc": roc_auc_score(y_fold, prob),
            "pr_auc": average_precision_score(y_fold, prob),
            "brier": brier_score_loss(y_fold, prob),
        })
        y_oof.append(y_fold)
        p_oof.append(prob)

    if not filas:
        raise RuntimeError("No fue posible construir folds temporales válidos.")

    tabla_folds = pd.DataFrame(filas)
    y_oof = np.concatenate(y_oof)
    p_oof = np.concatenate(p_oof)
    umbral, tabla_umbral = seleccionar_umbral(y_oof, p_oof)

    return tabla_folds, y_oof, p_oof, umbral, tabla_umbral

def evaluar_en_validacion(nombre_modelo, features, estrategia):
    folds, y_oof, p_oof, umbral, tabla_umbral = cv_temporal_oof(
        dev_raw, features, estrategia
    )

    dev = (
        filtrar_completos(dev_raw, features)
        if estrategia == "completos"
        else dev_raw.copy()
    )
    val = (
        filtrar_completos(val_raw, features)
        if estrategia == "completos"
        else val_raw.copy()
    )

    pipe, nums, cats = pipeline_predictivo(dev, features, estrategia)
    pipe.fit(dev[features], dev[TARGET])
    p_val = pipe.predict_proba(val[features])[:, 1]
    metricas_val = calcular_metricas(val[TARGET].to_numpy(dtype=int), p_val, umbral)

    resumen = {
        "modelo": nombre_modelo,
        "estrategia": estrategia,
        "n_features_base": len(features),
        "retencion_dev": len(dev) / max(len(dev_raw), 1),
        "retencion_val": len(val) / max(len(val_raw), 1),
        "cv_roc_auc_mean": folds["roc_auc"].mean(),
        "cv_roc_auc_sd": folds["roc_auc"].std(ddof=1),
        "cv_pr_auc_mean": folds["pr_auc"].mean(),
        "cv_pr_auc_sd": folds["pr_auc"].std(ddof=1),
        "cv_brier_mean": folds["brier"].mean(),
    }
    resumen.update({f"val_{k}": v for k, v in metricas_val.items()})

    return {
        "resumen": resumen,
        "pipeline_dev": pipe,
        "features": list(features),
        "nums": nums,
        "cats": cats,
        "dev": dev,
        "val": val,
        "prob_val": p_val,
        "umbral": umbral,
        "tabla_umbral": tabla_umbral,
        "folds": folds,
    }

mostrar_interpretacion(
    "✅ Módulo de evaluación configurado",
    [
        "El umbral se aprende con predicciones temporales fuera de muestra, no directamente sobre validación ni test.",
        "PR-AUC es la métrica primaria por el desbalance; Recall y F2 controlan la capacidad de detectar episodios críticos.",
    ],
    "ok",
)
```

```python
# =============================================================================
# 8.2 ENTRENAMIENTO DE LAS NUEVE COMBINACIONES EN DESARROLLO/VALIDACIÓN
# =============================================================================
artefactos_validacion = {}
filas_metricas = []
inicio = time.time()

for nombre_modelo, features in MODELOS.items():
    for estrategia in ESTRATEGIAS:
        clave = (nombre_modelo, estrategia)
        print("Evaluando:", clave)
        artefacto = evaluar_en_validacion(nombre_modelo, features, estrategia)
        artefactos_validacion[clave] = artefacto
        filas_metricas.append(artefacto["resumen"])

metricas_validacion = pd.DataFrame(filas_metricas)

mejor_pr = metricas_validacion["cv_pr_auc_mean"].max()
candidatos = metricas_validacion.loc[
    metricas_validacion["cv_pr_auc_mean"] >= mejor_pr - 0.01
].copy()

candidatos_recall = candidatos.loc[candidatos["val_recall"] >= MIN_RECALL].copy()
if not candidatos_recall.empty:
    candidatos = candidatos_recall

candidatos = candidatos.sort_values(
    [
        "val_f2",
        "cv_pr_auc_mean",
        "val_brier",
        "n_features_base",
        "retencion_dev",
    ],
    ascending=[False, False, True, True, False],
)

seleccion = candidatos.iloc[0]
CLAVE_FINAL = (seleccion["modelo"], seleccion["estrategia"])
ARTEFACTO_SELECCION = artefactos_validacion[CLAVE_FINAL]

metricas_validacion["seleccionado"] = (
    (metricas_validacion["modelo"] == CLAVE_FINAL[0])
    & (metricas_validacion["estrategia"] == CLAVE_FINAL[1])
)

metricas_validacion.to_csv(
    RESULTS_DIR / "s3_metricas_9_combinaciones_validacion.csv",
    index=False,
    encoding="utf-8-sig",
)

decision_modelo = pd.DataFrame([{
    "modelo_final": CLAVE_FINAL[0],
    "estrategia_final": CLAVE_FINAL[1],
    "cv_pr_auc": seleccion["cv_pr_auc_mean"],
    "cv_roc_auc": seleccion["cv_roc_auc_mean"],
    "val_recall": seleccion["val_recall"],
    "val_f2": seleccion["val_f2"],
    "val_brier": seleccion["val_brier"],
    "n_features": seleccion["n_features_base"],
    "retencion_dev": seleccion["retencion_dev"],
    "regla": "PR-AUC dentro de 0,01 del mejor; Recall mínimo; F2; Brier; parsimonia; retención",
    "test_utilizado_en_seleccion": False,
}])
decision_modelo.to_csv(
    RESULTS_DIR / "s3_decision_modelo_final_sin_test.csv",
    index=False,
    encoding="utf-8-sig",
)

metricas_validacion_ordenadas = metricas_validacion.sort_values(
    ["seleccionado", "cv_pr_auc_mean", "val_f2"],
    ascending=[False, False, False],
)

columnas_seleccion = [
    "modelo", "estrategia", "n_features_base", "retencion_dev",
    "cv_pr_auc_mean", "cv_roc_auc_mean", "val_precision",
    "val_recall", "val_f2", "val_brier", "seleccionado",
]
mostrar_tabla(
    metricas_validacion_ordenadas[columnas_seleccion],
    "Comparación ejecutiva de los tres modelos × tres estrategias",
    max_filas=20,
)

columnas_variabilidad = [
    "modelo", "estrategia", "cv_pr_auc_sd", "cv_roc_auc_sd",
    "retencion_val", "val_umbral", "val_specificity",
    "val_balanced_accuracy",
]
mostrar_tabla(
    metricas_validacion_ordenadas[columnas_variabilidad],
    "Estabilidad temporal y métricas complementarias de validación",
    max_filas=20,
)
mostrar_tabla(decision_modelo, "Decisión cerrada antes de abrir el test")

mostrar_interpretacion(
    "🏆 Modelo seleccionado sin fuga de prueba",
    [
        f"Se seleccionó <b>{CLAVE_FINAL[0]}</b> con estrategia <b>{CLAVE_FINAL[1]}</b>.",
        f"CV PR-AUC={seleccion['cv_pr_auc_mean']:.3f}, Recall de validación={seleccion['val_recall']:.3f}, F2={seleccion['val_f2']:.3f}.",
        f"El proceso tomó <b>{time.time() - inicio:.1f} segundos</b> y no utilizó métricas del conjunto de prueba.",
    ],
    "ok",
)
```

```python
# =============================================================================
# 9.1 AJUSTE FINAL Y APERTURA ÚNICA DEL TEST
# =============================================================================
features_finales = MODELOS[CLAVE_FINAL[0]]
estrategia_final = CLAVE_FINAL[1]

folds_finales, y_oof_final, p_oof_final, umbral_final, tabla_umbral_final = cv_temporal_oof(
    train_raw, features_finales, estrategia_final
)

train_final = (
    filtrar_completos(train_raw, features_finales)
    if estrategia_final == "completos"
    else train_raw.copy()
)
test_final = (
    filtrar_completos(test_raw, features_finales)
    if estrategia_final == "completos"
    else test_raw.copy()
)

pipeline_final, nums_finales, cats_finales = pipeline_predictivo(
    train_final, features_finales, estrategia_final
)
pipeline_final.fit(train_final[features_finales], train_final[TARGET])

prob_train = pipeline_final.predict_proba(train_final[features_finales])[:, 1]
prob_test = pipeline_final.predict_proba(test_final[features_finales])[:, 1]

metricas_train = calcular_metricas(
    train_final[TARGET].to_numpy(dtype=int), prob_train, umbral_final
)
metricas_test = calcular_metricas(
    test_final[TARGET].to_numpy(dtype=int), prob_test, umbral_final
)

tabla_metricas_final = pd.DataFrame([
    {"particion": "entrenamiento", **metricas_train},
    {"particion": "prueba_futura", **metricas_test},
])
tabla_metricas_final.to_csv(
    RESULTS_DIR / "s3_metricas_modelo_final_train_test.csv",
    index=False,
    encoding="utf-8-sig",
)
tabla_umbral_final.to_csv(
    RESULTS_DIR / "s3_sensibilidad_umbral_oof.csv",
    index=False,
    encoding="utf-8-sig",
)

FINAL = {
    "pipeline": pipeline_final,
    "features": features_finales,
    "estrategia": estrategia_final,
    "train": train_final,
    "test": test_final,
    "nums": nums_finales,
    "cats": cats_finales,
    "umbral": umbral_final,
    "prob_train": prob_train,
    "prob_test": prob_test,
    "metricas_train": metricas_train,
    "metricas_test": metricas_test,
    "folds": folds_finales,
    "tabla_umbral": tabla_umbral_final,
}

tabla_desempeno_final = tabla_metricas_final[[
    "particion", "n", "positivos", "prevalencia", "umbral",
    "precision", "recall", "specificity", "f1", "f2",
    "roc_auc", "pr_auc", "brier",
]].copy()

tabla_conteos_final = tabla_metricas_final[[
    "particion", "tn", "fp", "fn", "tp",
    "accuracy", "balanced_accuracy",
]].copy()

mostrar_tabla(
    tabla_desempeno_final,
    "Desempeño final: métricas principales",
)
mostrar_tabla(
    tabla_conteos_final,
    "Desempeño final: conteos y exactitud",
)
mostrar_kpis(
    "Resultado operacional en prueba",
    [
        ("Recall", f"{metricas_test['recall']:.3f}", "Episodios detectados"),
        ("Precision", f"{metricas_test['precision']:.3f}", "Alertas verdaderas"),
        ("F2", f"{metricas_test['f2']:.3f}", "Prioriza falsos negativos"),
        ("PR-AUC", f"{metricas_test['pr_auc']:.3f}", "Desbalance"),
        ("ROC-AUC", f"{metricas_test['roc_auc']:.3f}", "Discriminación"),
        ("Umbral", f"{umbral_final:.3f}", "Estimado con OOF"),
    ],
)

diferencia_auc = metricas_train["roc_auc"] - metricas_test["roc_auc"]
nivel = "ok" if abs(diferencia_auc) <= 0.03 else "alerta"
mostrar_interpretacion(
    "📌 Interpretación del desempeño futuro",
    [
        f"La diferencia de ROC-AUC entre entrenamiento y prueba es <b>{diferencia_auc:+.3f}</b>.",
        f"El modelo detecta <b>{100 * metricas_test['recall']:.1f}%</b> de los episodios positivos en el periodo futuro.",
        "Precision debe interpretarse junto con la prevalencia: un valor moderado puede ser razonable cuando los episodios críticos son poco frecuentes.",
    ],
    nivel,
)
```

    La inferencia se reajusta sin ponderación y sin penalización mediante
    <code>statsmodels.GLM(Binomial)</code>.</p>
    <p style="margin:0;">Esta separación permite interpretar coeficientes, odds ratios, p-valores,
    AIC/BIC e intervalos de Wald bajo un estimador homogéneo, evitando mezclar objetivos
    predictivos e inferenciales.</p>

</div>

```python
# =============================================================================
# 10.1 MODELO INFERENCIAL NO PONDERADO E INTERVALOS DE WALD
# =============================================================================
prep_final = pipeline_final.named_steps["prep"]
X_train_inf = np.asarray(
    prep_final.transform(train_final[features_finales]), dtype=float
)
nombres_inf = np.asarray(prep_final.get_feature_names_out(), dtype=str)

varianzas = np.nanvar(X_train_inf, axis=0)
idx_no_const = np.flatnonzero(varianzas > 1e-12)
X_train_inf = X_train_inf[:, idx_no_const]
nombres_inf = nombres_inf[idx_no_const]

Xb_inf = sm.add_constant(X_train_inf, has_constant="add")
y_train_inf = train_final[TARGET].to_numpy(dtype=int)

ajuste_inferencial = sm.GLM(
    y_train_inf, Xb_inf, family=sm.families.Binomial()
).fit(maxiter=300, disp=0)

parametros_inf = np.concatenate((["Intercepto"], nombres_inf))
ci_wald = ajuste_inferencial.conf_int(alpha=ALPHA)

coeficientes_wald = pd.DataFrame({
    "parametro": parametros_inf,
    "coeficiente": ajuste_inferencial.params,
    "se": ajuste_inferencial.bse,
    "z": ajuste_inferencial.tvalues,
    "p_valor": ajuste_inferencial.pvalues,
    "ic_wald_li": ci_wald[:, 0],
    "ic_wald_ls": ci_wald[:, 1],
})
coeficientes_wald["odds_ratio"] = np.exp(coeficientes_wald["coeficiente"])
coeficientes_wald["or_ic_wald_li"] = np.exp(coeficientes_wald["ic_wald_li"])
coeficientes_wald["or_ic_wald_ls"] = np.exp(coeficientes_wald["ic_wald_ls"])
coeficientes_wald["significativo_95"] = (
    coeficientes_wald["ic_wald_li"] * coeficientes_wald["ic_wald_ls"] > 0
)
coeficientes_wald["abs_coef"] = coeficientes_wald["coeficiente"].abs()

k_inf = len(ajuste_inferencial.params)
bic_inf = -2 * ajuste_inferencial.llf + k_inf * np.log(ajuste_inferencial.nobs)
resumen_inferencial = pd.DataFrame([{
    "n": int(ajuste_inferencial.nobs),
    "log_likelihood": ajuste_inferencial.llf,
    "AIC": ajuste_inferencial.aic,
    "BIC_ll": bic_inf,
    "deviance": ajuste_inferencial.deviance,
    "pearson_chi2": ajuste_inferencial.pearson_chi2,
    "convergio": bool(ajuste_inferencial.converged),
}])

coeficientes_wald.to_csv(
    RESULTS_DIR / "s3_coeficientes_or_ic_wald.csv", index=False, encoding="utf-8-sig"
)
resumen_inferencial.to_csv(
    RESULTS_DIR / "s3_resumen_modelo_inferencial.csv", index=False, encoding="utf-8-sig"
)

mostrar_tabla(resumen_inferencial, "Ajuste inferencial no ponderado")
mostrar_tabla(
    coeficientes_wald.sort_values("abs_coef", ascending=False).drop(columns="abs_coef"),
    "Coeficientes, odds ratios e intervalos de Wald",
    max_filas=35,
)

top_coef = (
    coeficientes_wald.loc[coeficientes_wald["parametro"] != "Intercepto"]
    .sort_values("abs_coef", ascending=False)
    .head(3)
)
lecturas = []
for _, fila in top_coef.iterrows():
    direccion = "aumenta" if fila["odds_ratio"] > 1 else "reduce"
    lecturas.append(
        f"<b>{html.escape(str(fila['parametro']))}</b>: OR={fila['odds_ratio']:.3f}; "
        f"un incremento de una unidad estandarizada {direccion} las odds de mala calidad, "
        "manteniendo constantes los demás predictores."
    )

mostrar_interpretacion(
    "📌 Interpretación de los principales efectos",
    lecturas + [
        "Las asociaciones son condicionales y predictivas; no deben presentarse como efectos causales.",
    ],
    "info",
)
```

```python
# =============================================================================
# 10.2 VIF, RESIDUOS, INFLUENCIA Y LINEALIDAD EN EL LOGIT
# =============================================================================
if X_train_inf.shape[1] >= 2:
    vif = pd.DataFrame({
        "variable": nombres_inf,
        "VIF": [
            variance_inflation_factor(X_train_inf, i)
            for i in range(X_train_inf.shape[1])
        ],
    }).sort_values("VIF", ascending=False)
else:
    vif = pd.DataFrame({"variable": nombres_inf, "VIF": [1.0] * len(nombres_inf)})

vif["nivel"] = pd.cut(
    vif["VIF"],
    bins=[-np.inf, 5, 10, np.inf],
    labels=["controlado", "revisar", "alto"],
)

resid_pearson = np.asarray(ajuste_inferencial.resid_pearson)
resid_deviance = np.asarray(ajuste_inferencial.resid_deviance)
influencia = ajuste_inferencial.get_influence(observed=False)
hat = np.asarray(influencia.hat_matrix_diag)
cooks = np.asarray(influencia.cooks_distance[0])
dfbetas = np.asarray(influencia.dfbetas)

p_param = Xb_inf.shape[1]
n_inf = len(y_train_inf)
umbral_hat = 2 * p_param / n_inf
umbral_cook = 4 / n_inf

resumen_diagnosticos = pd.DataFrame([{
    "n": n_inf,
    "pearson_abs_gt_2_n": int(np.sum(np.abs(resid_pearson) > 2)),
    "pearson_abs_gt_2_%": 100 * np.mean(np.abs(resid_pearson) > 2),
    "pearson_abs_gt_3_n": int(np.sum(np.abs(resid_pearson) > 3)),
    "pearson_abs_gt_3_%": 100 * np.mean(np.abs(resid_pearson) > 3),
    "leverage_gt_2p_n_n": int(np.sum(hat > umbral_hat)),
    "leverage_gt_2p_n_%": 100 * np.mean(hat > umbral_hat),
    "cooks_gt_4_n_n": int(np.sum(cooks > umbral_cook)),
    "cooks_gt_4_n_%": 100 * np.mean(cooks > umbral_cook),
    "max_cooks": float(np.nanmax(cooks)),
    "max_abs_dfbeta": float(np.nanmax(np.abs(dfbetas))),
}])

diagnostico_logit = []
for nombre_raw in nums_finales:
    coincidencias = np.flatnonzero(nombres_inf == nombre_raw)
    if len(coincidencias) != 1:
        continue
    j = int(coincidencias[0])
    x = X_train_inf[:, j]
    x_pos = x - np.nanmin(x) + 1e-6
    termino = x_pos * np.log(x_pos)
    X_ext = np.column_stack([Xb_inf, termino])
    try:
        ajuste_ext = sm.GLM(
            y_train_inf, X_ext, family=sm.families.Binomial()
        ).fit(maxiter=200, disp=0)
        p_bt = float(ajuste_ext.pvalues[-1])
    except Exception:
        p_bt = np.nan
    diagnostico_logit.append({
        "variable": nombre_raw,
        "p_termino_x_log_x": p_bt,
        "lectura": (
            "posible no linealidad" if pd.notna(p_bt) and p_bt < ALPHA
            else "sin evidencia fuerte de no linealidad"
        ),
    })

diagnostico_logit = pd.DataFrame(diagnostico_logit)

vif.to_csv(RESULTS_DIR / "s3_vif_modelo_final.csv", index=False, encoding="utf-8-sig")
resumen_diagnosticos.to_csv(
    RESULTS_DIR / "s3_residuos_influencia.csv", index=False, encoding="utf-8-sig"
)
diagnostico_logit.to_csv(
    RESULTS_DIR / "s3_linealidad_logit.csv", index=False, encoding="utf-8-sig"
)

mostrar_tabla(vif, "Factores de inflación de varianza", max_filas=35)
diagnosticos_largos = pd.DataFrame([
    {
        "diagnóstico": "Residuos de Pearson |r| > 2",
        "valor": int(np.sum(np.abs(resid_pearson) > 2)),
        "porcentaje": 100 * np.mean(np.abs(resid_pearson) > 2),
        "referencia": "Observaciones que requieren revisión",
    },
    {
        "diagnóstico": "Residuos de Pearson |r| > 3",
        "valor": int(np.sum(np.abs(resid_pearson) > 3)),
        "porcentaje": 100 * np.mean(np.abs(resid_pearson) > 3),
        "referencia": "Casos extremos",
    },
    {
        "diagnóstico": "Leverage > 2p/n",
        "valor": int(np.sum(hat > umbral_hat)),
        "porcentaje": 100 * np.mean(hat > umbral_hat),
        "referencia": f"Umbral={umbral_hat:.6f}",
    },
    {
        "diagnóstico": "Cook > 4/n",
        "valor": int(np.sum(cooks > umbral_cook)),
        "porcentaje": 100 * np.mean(cooks > umbral_cook),
        "referencia": f"Umbral={umbral_cook:.6f}",
    },
    {
        "diagnóstico": "Máxima distancia de Cook",
        "valor": float(np.nanmax(cooks)),
        "porcentaje": np.nan,
        "referencia": "Magnitud máxima observada",
    },
    {
        "diagnóstico": "Máximo |DFBeta|",
        "valor": float(np.nanmax(np.abs(dfbetas))),
        "porcentaje": np.nan,
        "referencia": "Influencia máxima sobre parámetros",
    },
])

mostrar_tabla(
    diagnosticos_largos,
    "Resumen de residuos e influencia",
)
if not diagnostico_logit.empty:
    mostrar_tabla(diagnostico_logit, "Diagnóstico de linealidad en el logit")

max_vif = vif["VIF"].max() if not vif.empty else np.nan
n_no_lineal = (
    int((diagnostico_logit["p_termino_x_log_x"] < ALPHA).sum())
    if not diagnostico_logit.empty else 0
)

mostrar_interpretacion(
    "📌 Diagnóstico integral",
    [
        f"El VIF máximo es <b>{max_vif:.2f}</b>; valores inferiores a 5 indican multicolinealidad controlada.",
        f"Existen <b>{int(np.sum(np.abs(resid_pearson) > 3)):,}</b> observaciones con |residuo de Pearson| > 3.",
        f"Se detectaron <b>{n_no_lineal}</b> variables con evidencia exploratoria de no linealidad en el logit.",
        "Las observaciones influyentes se auditan, pero no se eliminan automáticamente: primero debe distinguirse un error de medición de un episodio ambiental legítimamente extremo.",
    ],
    "alerta" if (max_vif >= 5 or n_no_lineal > 0) else "ok",
)

fig, ax = plt.subplots(figsize=(10, 6))
ax.scatter(ajuste_inferencial.fittedvalues, resid_pearson, s=9, alpha=0.35)
ax.axhline(0, linestyle="--")
ax.axhline(3, linestyle=":")
ax.axhline(-3, linestyle=":")
ax.set_xlabel("Probabilidad ajustada")
ax.set_ylabel("Residuo de Pearson")
ax.set_title("Residuos de Pearson versus probabilidad ajustada")
guardar_figura("05_residuos_pearson.png")
plt.show()

fig, ax = plt.subplots(figsize=(10, 6))
vif_plot = vif.sort_values("VIF").tail(20)
ax.barh(vif_plot["variable"], vif_plot["VIF"])
ax.axvline(5, linestyle="--", label="Umbral de revisión = 5")
ax.set_xlabel("VIF")
ax.set_title("Multicolinealidad del modelo inferencial")
ax.legend()
guardar_figura("06_vif_modelo_final.png")
plt.show()
```

    <p style="margin:0 0 10px 0;">El bootstrap reajusta el mismo estimador inferencial:
    regresión logística sin ponderación y sin penalización. De este modo la comparación con Wald
    es homogénea.</p>
    <p style="margin:0;">La ejecución completa solicita 10.000 remuestras y reporta estabilidad
    de los límites en 250, 500, 1.000, 2.000, 5.000 y 10.000 iteraciones.</p>

```python
# =============================================================================
# 11.1 BOOTSTRAP NO PARAMÉTRICO DE COEFICIENTES
# =============================================================================
def ajustar_logit_sin_penalizacion(X, y, max_iter=2500):
    configuraciones = [
        {"penalty": None},
        {"penalty": "none"},
    ]
    ultimo_error = None
    for config in configuraciones:
        try:
            modelo = LogisticRegression(
                solver="lbfgs",
                max_iter=max_iter,
                class_weight=None,
                random_state=SEED,
                **config,
            )
            modelo.fit(X, y)
            return modelo
        except Exception as exc:
            ultimo_error = exc
    raise RuntimeError(f"No fue posible ajustar regresión sin penalización: {ultimo_error}")

def bootstrap_logistico(X, y, n_boot=N_BOOT_MODEL, seed=SEED):
    generador = np.random.default_rng(seed)
    n = len(y)
    betas = []
    fallos = 0

    for b in range(n_boot):
        indices = generador.integers(0, n, n)
        y_b = y[indices]
        if np.unique(y_b).size < 2:
            fallos += 1
            continue
        try:
            modelo = ajustar_logit_sin_penalizacion(X[indices], y_b)
            betas.append(np.r_[modelo.intercept_, modelo.coef_.ravel()])
        except Exception:
            fallos += 1

    return np.asarray(betas), fallos

inicio_boot = time.time()
betas_boot, fallos_boot = bootstrap_logistico(
    X_train_inf, y_train_inf, N_BOOT_MODEL, SEED
)
if len(betas_boot) < max(100, int(0.80 * N_BOOT_MODEL)):
    raise RuntimeError(
        "Demasiadas remuestras bootstrap fallaron; revise separación o singularidad."
    )

coef_original = ajuste_inferencial.params
parametros_boot = parametros_inf
li_boot = np.percentile(betas_boot, 2.5, axis=0)
ls_boot = np.percentile(betas_boot, 97.5, axis=0)
media_boot = betas_boot.mean(axis=0)
sd_boot = betas_boot.std(axis=0, ddof=1)
signo_consistente = np.mean(
    np.sign(betas_boot) == np.sign(coef_original), axis=0
)

bootstrap_coef = pd.DataFrame({
    "parametro": parametros_boot,
    "coef_original": coef_original,
    "media_bootstrap": media_boot,
    "sesgo_bootstrap": media_boot - coef_original,
    "sd_bootstrap": sd_boot,
    "ic_boot_li": li_boot,
    "ic_boot_ls": ls_boot,
    "signo_consistente": signo_consistente,
})
bootstrap_coef["or_original"] = np.exp(bootstrap_coef["coef_original"])
bootstrap_coef["or_ic_boot_li"] = np.exp(bootstrap_coef["ic_boot_li"])
bootstrap_coef["or_ic_boot_ls"] = np.exp(bootstrap_coef["ic_boot_ls"])
bootstrap_coef["inestable"] = (
    ((bootstrap_coef["ic_boot_li"] <= 0) & (bootstrap_coef["ic_boot_ls"] >= 0))
    | (bootstrap_coef["signo_consistente"] < 0.90)
)

comparacion_ic = bootstrap_coef.merge(
    coeficientes_wald.drop(columns=["abs_coef"]),
    on="parametro",
    how="left",
)
comparacion_ic["ancho_boot"] = (
    comparacion_ic["ic_boot_ls"] - comparacion_ic["ic_boot_li"]
)
comparacion_ic["ancho_wald"] = (
    comparacion_ic["ic_wald_ls"] - comparacion_ic["ic_wald_li"]
)
comparacion_ic["ratio_ancho_boot_wald"] = (
    comparacion_ic["ancho_boot"]
    / comparacion_ic["ancho_wald"].replace(0, np.nan)
)

checkpoints = sorted({
    c for c in [250, 500, 1000, 2000, 5000, 10000, len(betas_boot)]
    if c <= len(betas_boot)
})
filas_convergencia = []
for b in checkpoints:
    li_b = np.percentile(betas_boot[:b], 2.5, axis=0)
    ls_b = np.percentile(betas_boot[:b], 97.5, axis=0)
    for i, parametro in enumerate(parametros_boot):
        filas_convergencia.append({
            "remuestras": b,
            "parametro": parametro,
            "ic_li": li_b[i],
            "ic_ls": ls_b[i],
            "ancho": ls_b[i] - li_b[i],
        })

convergencia_boot = pd.DataFrame(filas_convergencia)
resumen_boot = pd.DataFrame([{
    "solicitadas": N_BOOT_MODEL,
    "exitosas": len(betas_boot),
    "fallos": fallos_boot,
    "tasa_exito": len(betas_boot) / N_BOOT_MODEL,
    "parametros_inestables": int(bootstrap_coef["inestable"].sum()),
    "tiempo_segundos": time.time() - inicio_boot,
}])

comparacion_ic.to_csv(
    RESULTS_DIR / "s3_bootstrap_coeficientes_or.csv",
    index=False,
    encoding="utf-8-sig",
)
convergencia_boot.to_csv(
    RESULTS_DIR / "s3_bootstrap_convergencia.csv",
    index=False,
    encoding="utf-8-sig",
)
resumen_boot.to_csv(
    RESULTS_DIR / "s3_resumen_bootstrap.csv",
    index=False,
    encoding="utf-8-sig",
)
np.save(RESULTS_DIR / "s3_bootstrap_coeficientes.npy", betas_boot)

mostrar_tabla(resumen_boot, "Resumen técnico del bootstrap")
comparacion_ic_ordenada = comparacion_ic.sort_values(
    ["inestable", "signo_consistente"],
    ascending=[False, True],
)

mostrar_tabla(
    comparacion_ic_ordenada[[
        "parametro", "coef_original", "media_bootstrap",
        "sesgo_bootstrap", "sd_bootstrap",
        "signo_consistente", "inestable",
    ]],
    "Bootstrap: estabilidad de coeficientes",
    max_filas=35,
)

mostrar_tabla(
    comparacion_ic_ordenada[[
        "parametro", "ic_boot_li", "ic_boot_ls",
        "ic_wald_li", "ic_wald_ls",
        "ancho_boot", "ancho_wald", "ratio_ancho_boot_wald",
    ]],
    "Bootstrap versus Wald: intervalos de confianza",
    max_filas=35,
)

mostrar_tabla(
    comparacion_ic_ordenada[[
        "parametro", "or_original",
        "or_ic_boot_li", "or_ic_boot_ls",
        "odds_ratio", "or_ic_wald_li", "or_ic_wald_ls",
    ]],
    "Odds ratios: intervalos bootstrap y Wald",
    max_filas=35,
)

inestables = bootstrap_coef.loc[
    bootstrap_coef["inestable"] & (bootstrap_coef["parametro"] != "Intercepto"),
    "parametro",
].tolist()

mostrar_interpretacion(
    "📌 Interpretación de estabilidad",
    [
        f"Se completaron <b>{len(betas_boot):,}</b> remuestras con una tasa de éxito de <b>{100 * len(betas_boot) / N_BOOT_MODEL:.2f}%</b>.",
        f"Se marcaron <b>{len(inestables)}</b> parámetros inestables fuera del intercepto.",
        "Un parámetro es inestable si su IC bootstrap cruza cero o si conserva su signo en menos del 90% de las remuestras.",
        "La coincidencia entre Wald y bootstrap fortalece la estabilidad; una discrepancia obliga a comunicar cautela.",
    ],
    "alerta" if inestables else "ok",
)
```

```python
# =============================================================================
# 11.2 VISUALIZACIÓN DE INTERVALOS Y CONVERGENCIA
# =============================================================================
plot_ic = comparacion_ic.loc[comparacion_ic["parametro"] != "Intercepto"].copy()
plot_ic["abs_coef"] = plot_ic["coef_original"].abs()
plot_ic = (
    plot_ic.nlargest(min(20, len(plot_ic)), "abs_coef")
    .sort_values("coef_original")
)

ypos = np.arange(len(plot_ic))
fig, ax = plt.subplots(figsize=(11, 7.5))
ax.errorbar(
    plot_ic["coef_original"],
    ypos,
    xerr=[
        plot_ic["coef_original"] - plot_ic["ic_boot_li"],
        plot_ic["ic_boot_ls"] - plot_ic["coef_original"],
    ],
    fmt="o",
    capsize=3,
    label="IC bootstrap 95%",
)
ax.errorbar(
    plot_ic["coeficiente"],
    ypos + 0.16,
    xerr=[
        plot_ic["coeficiente"] - plot_ic["ic_wald_li"],
        plot_ic["ic_wald_ls"] - plot_ic["coeficiente"],
    ],
    fmt="s",
    capsize=3,
    alpha=0.78,
    label="IC Wald 95%",
)
ax.axvline(0, linestyle="--", linewidth=1.5)
ax.set_yticks(ypos)
ax.set_yticklabels(plot_ic["parametro"])
ax.set_xlabel("Coeficiente log-odds")
ax.set_title("Estabilidad de parámetros · bootstrap versus Wald")
ax.legend()
guardar_figura("07_ic_bootstrap_vs_wald.png")
plt.show()

parametros_clave = (
    bootstrap_coef.loc[bootstrap_coef["parametro"] != "Intercepto"]
    .assign(abs_coef=lambda x: x["coef_original"].abs())
    .nlargest(min(5, len(bootstrap_coef) - 1), "abs_coef")["parametro"]
    .tolist()
)

fig, ax = plt.subplots(figsize=(10, 6))
for parametro in parametros_clave:
    tmp = convergencia_boot.loc[convergencia_boot["parametro"] == parametro]
    ax.plot(tmp["remuestras"], tmp["ancho"], marker="o", label=parametro)
ax.set_xscale("log")
ax.set_xlabel("Número de remuestras")
ax.set_ylabel("Ancho del IC 95%")
ax.set_title("Convergencia del ancho de los intervalos bootstrap")
ax.legend()
guardar_figura("08_convergencia_bootstrap.png")
plt.show()

ultimo_b = max(checkpoints)
penultimo_b = sorted(checkpoints)[-2] if len(checkpoints) > 1 else ultimo_b
anchos_ultimo = convergencia_boot.loc[
    convergencia_boot["remuestras"] == ultimo_b, ["parametro", "ancho"]
].rename(columns={"ancho": "ancho_final"})
anchos_previo = convergencia_boot.loc[
    convergencia_boot["remuestras"] == penultimo_b, ["parametro", "ancho"]
].rename(columns={"ancho": "ancho_previo"})
comparacion_convergencia = anchos_ultimo.merge(anchos_previo, on="parametro")
comparacion_convergencia["variacion_relativa"] = (
    (comparacion_convergencia["ancho_final"] - comparacion_convergencia["ancho_previo"]).abs()
    / comparacion_convergencia["ancho_final"].replace(0, np.nan)
)
mostrar_tabla(
    comparacion_convergencia.sort_values("variacion_relativa", ascending=False),
    f"Convergencia: {penultimo_b:,} versus {ultimo_b:,} remuestras",
    max_filas=30,
)

mostrar_interpretacion(
    "Lectura de convergencia",
    [
        "La estabilización del ancho de los intervalos demuestra si el número de remuestras es suficiente para los parámetros principales.",
        "Si un parámetro conserva variaciones amplias al final, debe destacarse como sensible aunque su intervalo no cruce cero.",
    ],
    "info",
)
```

```python
# =============================================================================
# 12.1 COMPARACIÓN DEL MODELO FINAL BAJO TRES ESTRATEGIAS
# =============================================================================
comparacion_imputacion = []
coeficientes_imputacion = []

for estrategia in ESTRATEGIAS:
    artefacto = evaluar_en_validacion(CLAVE_FINAL[0], features_finales, estrategia)
    fila = artefacto["resumen"].copy()
    comparacion_imputacion.append(fila)

    datos_inf = (
        filtrar_completos(train_raw, features_finales)
        if estrategia == "completos"
        else train_raw.copy()
    )
    prep_imp, _, _ = construir_preprocesador(datos_inf, features_finales, estrategia)
    X_imp = np.asarray(prep_imp.fit_transform(datos_inf[features_finales]), dtype=float)
    nombres_imp = np.asarray(prep_imp.get_feature_names_out(), dtype=str)
    idx_var = np.flatnonzero(np.nanvar(X_imp, axis=0) > 1e-12)
    X_imp = X_imp[:, idx_var]
    nombres_imp = nombres_imp[idx_var]
    Xb_imp = sm.add_constant(X_imp, has_constant="add")
    y_imp = datos_inf[TARGET].to_numpy(dtype=int)

    try:
        fit_imp = sm.GLM(y_imp, Xb_imp, family=sm.families.Binomial()).fit(
            maxiter=250, disp=0
        )
        ci_imp = fit_imp.conf_int()
        for i, nombre in enumerate(np.concatenate((["Intercepto"], nombres_imp))):
            coeficientes_imputacion.append({
                "estrategia": estrategia,
                "parametro": nombre,
                "coeficiente": fit_imp.params[i],
                "ic_li": ci_imp[i, 0],
                "ic_ls": ci_imp[i, 1],
                "odds_ratio": np.exp(fit_imp.params[i]),
                "or_ic_li": np.exp(ci_imp[i, 0]),
                "or_ic_ls": np.exp(ci_imp[i, 1]),
            })
    except Exception as exc:
        print(f"No fue posible obtener inferencia para {estrategia}: {exc}")

comparacion_imputacion = pd.DataFrame(comparacion_imputacion)
coeficientes_imputacion = pd.DataFrame(coeficientes_imputacion)

if not resumen_global_imputacion.empty:
    comparacion_imputacion = comparacion_imputacion.merge(
        resumen_global_imputacion,
        left_on="estrategia",
        right_on="metodo",
        how="left",
    ).drop(columns=["metodo"], errors="ignore")

comparacion_imputacion.to_csv(
    RESULTS_DIR / "s3_impacto_imputacion_metricas.csv",
    index=False,
    encoding="utf-8-sig",
)
coeficientes_imputacion.to_csv(
    RESULTS_DIR / "s3_impacto_imputacion_coeficientes.csv",
    index=False,
    encoding="utf-8-sig",
)

comparacion_imputacion_ordenada = comparacion_imputacion.sort_values(
    "cv_pr_auc_mean",
    ascending=False,
)

columnas_calidad_imp = [
    c for c in [
        "modelo", "estrategia", "n_features_base",
        "retencion_dev", "retencion_val",
        "nrmse_iqr_promedio", "encogimiento_promedio",
    ]
    if c in comparacion_imputacion_ordenada.columns
]
mostrar_tabla(
    comparacion_imputacion_ordenada[columnas_calidad_imp],
    "Imputación: retención y calidad estadística",
    max_filas=10,
)

columnas_desempeno_imp = [
    c for c in [
        "modelo", "estrategia",
        "cv_pr_auc_mean", "cv_roc_auc_mean",
        "val_precision", "val_recall", "val_f2",
        "val_brier", "val_umbral",
    ]
    if c in comparacion_imputacion_ordenada.columns
]
mostrar_tabla(
    comparacion_imputacion_ordenada[columnas_desempeno_imp],
    "Imputación: impacto sobre desempeño predictivo",
    max_filas=10,
)

if not coeficientes_imputacion.empty:
    comunes = (
        coeficientes_imputacion.groupby("parametro")["estrategia"].nunique()
    )
    comunes = comunes[comunes == 3].index
    sensibilidad_coef = (
        coeficientes_imputacion.loc[
            coeficientes_imputacion["parametro"].isin(comunes)
            & (coeficientes_imputacion["parametro"] != "Intercepto")
        ]
        .pivot(index="parametro", columns="estrategia", values="coeficiente")
    )
    sensibilidad_coef["rango_coeficiente"] = (
        sensibilidad_coef.max(axis=1) - sensibilidad_coef.min(axis=1)
    )
    sensibilidad_coef = sensibilidad_coef.sort_values(
        "rango_coeficiente", ascending=False
    )
    sensibilidad_coef.to_csv(
        RESULTS_DIR / "s3_sensibilidad_coeficientes_imputacion.csv",
        encoding="utf-8-sig",
    )
    mostrar_tabla(
        sensibilidad_coef.reset_index().head(25),
        "Sensibilidad de coeficientes entre estrategias",
        max_filas=25,
    )

mejor_pr_imp = comparacion_imputacion["cv_pr_auc_mean"].max()
elegibles_imp = comparacion_imputacion.loc[
    comparacion_imputacion["cv_pr_auc_mean"] >= mejor_pr_imp - 0.01
].copy()

if "nrmse_iqr_promedio" in elegibles_imp.columns and elegibles_imp["nrmse_iqr_promedio"].notna().any():
    elegibles_imp = elegibles_imp.sort_values(
        ["nrmse_iqr_promedio", "encogimiento_promedio", "retencion_dev", "val_f2"],
        ascending=[True, True, False, False],
    )
else:
    elegibles_imp = elegibles_imp.sort_values(
        ["retencion_dev", "val_f2", "val_brier"],
        ascending=[False, False, True],
    )

ESTRATEGIA_RECOMENDADA = elegibles_imp.iloc[0]["estrategia"]
decision_imputacion = pd.DataFrame([{
    "estrategia_recomendada": ESTRATEGIA_RECOMENDADA,
    "estrategia_modelo_final": estrategia_final,
    "criterio": "PR-AUC equivalente; NRMSE; encogimiento; retención; F2",
    "coincide_con_modelo_final": ESTRATEGIA_RECOMENDADA == estrategia_final,
}])
decision_imputacion.to_csv(
    RESULTS_DIR / "s3_decision_imputacion.csv", index=False, encoding="utf-8-sig"
)
mostrar_tabla(decision_imputacion, "Decisión fundamentada de imputación")

mostrar_interpretacion(
    "📌 Conclusión sobre los faltantes",
    [
        f"La estrategia recomendada es <b>{ESTRATEGIA_RECOMENDADA}</b>.",
        "Una diferencia mínima de desempeño no justifica perder una proporción importante de la muestra.",
        "Los parámetros con mayor desplazamiento entre estrategias deben comunicarse como sensibles y no utilizarse como evidencia causal.",
    ],
    "ok" if ESTRATEGIA_RECOMENDADA == estrategia_final else "alerta",
)
```

```python
# =============================================================================
# 13.2 DESEMPEÑO POR SUBGRUPOS
# =============================================================================
test_diag = test_final.copy()
test_diag["probabilidad"] = prob_test
test_diag["prediccion"] = pred_test

def metricas_grupo(grupo):
    base = {
        "n": len(grupo),
        "positivos": int(grupo[TARGET].sum()),
        "prevalencia": grupo[TARGET].mean(),
    }
    if len(grupo) < 30 or grupo[TARGET].nunique() < 2:
        return pd.Series({
            **base,
            "recall": np.nan,
            "precision": np.nan,
            "f1": np.nan,
            "roc_auc": np.nan,
            "pr_auc": np.nan,
            "brier": np.nan,
        })
    return pd.Series({
        **base,
        "recall": recall_score(grupo[TARGET], grupo["prediccion"], zero_division=0),
        "precision": precision_score(grupo[TARGET], grupo["prediccion"], zero_division=0),
        "f1": f1_score(grupo[TARGET], grupo["prediccion"], zero_division=0),
        "roc_auc": roc_auc_score(grupo[TARGET], grupo["probabilidad"]),
        "pr_auc": average_precision_score(grupo[TARGET], grupo["probabilidad"]),
        "brier": brier_score_loss(grupo[TARGET], grupo["probabilidad"]),
    })

dimensiones = [
    c for c in [
        "zona_geografica", "nivel_socioeconomico", "estacion_anio", "comuna"
    ]
    if c in test_diag.columns
]

diagnosticos = []
for dimension in dimensiones:
    tabla = (
        test_diag.groupby(dimension, observed=True)
        .apply(metricas_grupo, include_groups=False)
        .reset_index()
        .rename(columns={dimension: "grupo"})
    )
    tabla.insert(0, "dimension", dimension)
    diagnosticos.append(tabla)

diagnostico_grupos = (
    pd.concat(diagnosticos, ignore_index=True)
    if diagnosticos
    else pd.DataFrame(
        columns=[
            "dimension", "grupo", "n", "positivos", "prevalencia",
            "recall", "precision", "f1", "roc_auc", "pr_auc", "brier",
        ]
    )
)
diagnostico_grupos.to_csv(
    RESULTS_DIR / "s3_diagnostico_subgrupos.csv",
    index=False,
    encoding="utf-8-sig",
)
mostrar_tabla(
    diagnostico_grupos.sort_values(["dimension", "n"], ascending=[True, False]),
    "Desempeño por subgrupos territoriales y socioeconómicos",
    max_filas=60,
)

if diagnostico_grupos.empty:
    mostrar_interpretacion(
        "Diagnóstico de subgrupos",
        "El dataset no contiene dimensiones territoriales suficientes para segmentar el desempeño.",
        "alerta",
    )
else:
    recall_validos = diagnostico_grupos["recall"].dropna()
    brecha_recall = recall_validos.max() - recall_validos.min() if len(recall_validos) else np.nan
    mostrar_interpretacion(
        "📌 Interpretación de equidad operacional",
        [
            f"La brecha máxima observada de Recall entre grupos evaluables es <b>{brecha_recall:.3f}</b>.",
            "Una brecha amplia puede deberse a tamaño muestral, prevalencia, sensores, clima o especificación del modelo; requiere análisis, no una conclusión automática.",
            "Los grupos con menos de 30 observaciones o una sola clase se reportan sin métricas discriminantes para evitar resultados engañosos.",
        ],
        "alerta" if pd.notna(brecha_recall) and brecha_recall > 0.15 else "info",
    )
```
