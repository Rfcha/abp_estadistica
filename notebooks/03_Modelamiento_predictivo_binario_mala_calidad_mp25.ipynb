```python
# =============================================================================
# 1. CONFIGURACIÓN DEL ENTORNO, RUTAS Y FUNCIONES DE PRESENTACIÓN
# =============================================================================
import os, sys, math, warnings, textwrap, time, json
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
from scipy import stats
from scipy.special import expit

from IPython.display import display, HTML

from sklearn.experimental import enable_iterative_imputer  # noqa: F401
from sklearn.impute import SimpleImputer, IterativeImputer
from sklearn.linear_model import LogisticRegression, BayesianRidge
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    confusion_matrix, precision_score, recall_score, f1_score, fbeta_score,
    roc_auc_score, average_precision_score, balanced_accuracy_score,
    accuracy_score, brier_score_loss, roc_curve, precision_recall_curve,
    auc
)
from sklearn.calibration import calibration_curve
from sklearn.inspection import permutation_importance
from statsmodels.stats.outliers_influence import variance_inflation_factor

warnings.filterwarnings("ignore")

SEED = 2026
ALPHA = 0.05
FAST_MODE = os.getenv("S3_FAST", "0") == "1"
N_BOOT_MODEL = 100 if FAST_MODE else 1_000
N_MASK_REPEATS = 1 if FAST_MODE else 3
rng = np.random.default_rng(SEED)

# Detectar raíz del proyecto desde notebook/ o desde la raíz.
CWD = Path.cwd().resolve()
if CWD.name.lower() == "notebooks":
    PROJECT_ROOT = CWD.parent
elif (CWD / "data").exists():
    PROJECT_ROOT = CWD
else:
    PROJECT_ROOT = next((p for p in [CWD, *CWD.parents] if (p / "data").exists()), CWD)

DATA_DIR = PROJECT_ROOT / "data"
FIGURES_DIR = PROJECT_ROOT / "figures" / "S3"
RESULTS_DIR = PROJECT_ROOT / "resultados" / "S3"
FIGURES_DIR.mkdir(parents=True, exist_ok=True)
RESULTS_DIR.mkdir(parents=True, exist_ok=True)

COLORS = {
    "navy": "#1F3A5F", "blue": "#2563EB", "cyan": "#0F8FA8",
    "green": "#15803D", "amber": "#D97706", "red": "#B22234",
    "slate": "#475569", "light": "#F8FAFC", "grid": "#CBD5E1"
}

plt.rcParams.update({
    "figure.figsize": (11.5, 6.2), "figure.dpi": 130, "savefig.dpi": 300,
    "axes.grid": True, "grid.alpha": 0.28, "axes.titleweight": "bold",
    "axes.titlesize": 14, "axes.labelsize": 11, "font.size": 10,
    "legend.frameon": True, "legend.framealpha": 0.92,
    "axes.spines.top": False, "axes.spines.right": False
})

def tarjeta(titulo, texto, color="#0F8FA8"):
    display(HTML(f"""
    <div style='background:#fff;border:1px solid #d6e4ec;border-left:7px solid {color};border-radius:14px;padding:18px 22px;margin:18px 0;font-family:Arial,Helvetica,sans-serif;color:#1f2937;box-shadow:0 4px 14px rgba(15,143,168,.08);'>
      <div style='font-size:18px;font-weight:800;color:{color};margin-bottom:8px;'>{titulo}</div>
      <div style='font-size:14.5px;color:#374151;line-height:1.65;'>{texto}</div>
    </div>"""))

def guardar_figura(nombre):
    ruta = FIGURES_DIR / nombre
    plt.savefig(ruta, bbox_inches="tight", pad_inches=.25, facecolor="white")
    return ruta

def fmt_pct(x, d=2):
    return "s/d" if pd.isna(x) else f"{100*float(x):.{d}f}%"

def fmt_num(x, d=3):
    return "s/d" if pd.isna(x) else f"{float(x):,.{d}f}"

tarjeta(
    "✅ Entorno configurado",
    f"Python {sys.version.split()[0]} · pandas {pd.__version__} · NumPy {np.__version__} · "
    f"scikit-learn disponible · semilla {SEED} · bootstrap {N_BOOT_MODEL:,} · raíz: <code>{PROJECT_ROOT}</code>",
    COLORS["cyan"]
)
```

<h2 style="margin:0 0 10px 0;font-size:24px;">2. Carga del dataset y trazabilidad formal S1/S2</h2>
<p style="margin:0 0 10px 0;">Se reutiliza <code>data/calidad_aire_diario.csv</code>, la misma base consolidada en S1 y validada computacionalmente en S2. La lectura admite separador coma o punto y coma y distintas codificaciones.</p>
<p style="margin:0 0 10px 0;">Además, se carga el archivo puente <code>resultados/s2_reporte_resultados_validados_para_s3.csv</code>. Este reporte no se combina fila a fila con el dataset estación-día: se utiliza para justificar selección de variables, control de colinealidad, tratamiento de faltantes y comunicación de resultados sensibles.</p>
<p style="margin:0;"><b>Regla metodológica:</b> la variable objetivo nunca se imputa. Los registros sin etiqueta válida no pueden participar en aprendizaje supervisado.</p>

```python
# =============================================================================
# 2. CARGA ROBUSTA DEL DATASET
# =============================================================================
CANDIDATOS = [
    DATA_DIR / "calidad_aire_diario.csv",
    PROJECT_ROOT / "calidad_aire_diario.csv",
    CWD / "data" / "calidad_aire_diario.csv",
    CWD.parent / "data" / "calidad_aire_diario.csv",
]
ruta_dataset = next((p for p in CANDIDATOS if p.exists()), None)
if ruta_dataset is None:
    raise FileNotFoundError(
        "No se encontró calidad_aire_diario.csv. Ubícalo en data/ dentro del repositorio.\n" +
        "Rutas revisadas:\n- " + "\n- ".join(str(p) for p in CANDIDATOS)
    )

def leer_csv_robusto(ruta):
    errores = []
    for enc in ["utf-8-sig", "utf-8", "latin-1"]:
        for sep in [None, ";", ","]:
            try:
                kwargs = {"encoding": enc}
                if sep is None:
                    kwargs.update({"sep": None, "engine": "python"})
                else:
                    kwargs["sep"] = sep
                tmp = pd.read_csv(ruta, **kwargs)
                if tmp.shape[1] > 1:
                    return tmp
            except Exception as e:
                errores.append(f"{enc}/{sep}: {e}")
    raise RuntimeError("No fue posible leer el CSV. " + " | ".join(errores[-4:]))

df_raw = leer_csv_robusto(ruta_dataset)
df_raw.columns = (df_raw.columns.astype(str).str.strip().str.replace("\ufeff", "", regex=False))

# Fecha y orden cronológico.
col_fecha = next((c for c in ["fecha", "date", "Fecha"] if c in df_raw.columns), None)
if col_fecha is None:
    raise ValueError("Se requiere una columna de fecha para separar entrenamiento y prueba sin fuga temporal.")
df_raw[col_fecha] = pd.to_datetime(df_raw[col_fecha], errors="coerce")
df_raw = df_raw.loc[df_raw[col_fecha].notna()].sort_values(col_fecha).reset_index(drop=True)

resumen_carga = pd.DataFrame([{
    "archivo": ruta_dataset.name,
    "filas": len(df_raw),
    "columnas": df_raw.shape[1],
    "fecha_min": df_raw[col_fecha].min(),
    "fecha_max": df_raw[col_fecha].max(),
    "duplicados_exactos": int(df_raw.duplicated().sum())
}])
resumen_carga.to_csv(RESULTS_DIR / "s3_resumen_carga.csv", index=False, encoding="utf-8-sig")
display(resumen_carga)
display(df_raw.head(3))
```

<p style="margin:0 0 10px 0;">El archivo <code>s2_reporte_resultados_validados_para_s3.csv</code> funciona como puente formal entre S2 y S3. Contiene parámetros validados, correlaciones robustas, resultados sensibles y decisiones metodológicas.</p>

```python
# =============================================================================
# 2.1 CARGA DEL REPORTE VALIDADO DE SUMATIVA 2
# -----------------------------------------------------------------------------
# Este archivo funciona como puente metodológico S2 → S3. No se incorpora
# directamente como matriz predictora, ya que contiene resultados agregados,
# decisiones metodológicas y recomendaciones, no observaciones estación-día.
# =============================================================================
CANDIDATOS_S2 = [
    PROJECT_ROOT / "resultados" / "s2_reporte_resultados_validados_para_s3.csv",
    PROJECT_ROOT / "s2_reporte_resultados_validados_para_s3.csv",
    CWD / "resultados" / "s2_reporte_resultados_validados_para_s3.csv",
    CWD / "s2_reporte_resultados_validados_para_s3.csv",
]

ruta_reporte_s2 = next((p for p in CANDIDATOS_S2 if p.exists()), None)

if ruta_reporte_s2 is None:
    raise FileNotFoundError(
        "No se encontró el archivo puente de Sumativa 2:\n"
        "s2_reporte_resultados_validados_para_s3.csv\n\n"
        "Debe ubicarse preferentemente en:\n"
        f"{PROJECT_ROOT / 'resultados' / 's2_reporte_resultados_validados_para_s3.csv'}\n\n"
        "Rutas revisadas:\n- " + "\n- ".join(str(p) for p in CANDIDATOS_S2)
    )

reporte_s2 = leer_csv_robusto(ruta_reporte_s2)
reporte_s2.columns = (
    reporte_s2.columns.astype(str)
    .str.strip()
    .str.replace("\ufeff", "", regex=False)
)

columnas_requeridas_s2 = {
    "categoria",
    "elemento",
    "resultado_validado",
    "estado",
    "uso_s3",
}

faltantes_estructura = columnas_requeridas_s2 - set(reporte_s2.columns)

if faltantes_estructura:
    raise ValueError(
        "El reporte S2 no contiene todas las columnas requeridas: "
        + ", ".join(sorted(faltantes_estructura))
    )

resumen_carga_s2 = pd.DataFrame([{
    "archivo": ruta_reporte_s2.name,
    "ruta": str(ruta_reporte_s2),
    "evidencias": len(reporte_s2),
    "columnas": reporte_s2.shape[1],
    "duplicados_exactos": int(reporte_s2.duplicated().sum()),
    "campos_faltantes_totales": int(reporte_s2[list(columnas_requeridas_s2)].isna().sum().sum()),
}])

display(resumen_carga_s2)
display(reporte_s2)

tarjeta(
    "✅ Reporte puente S2 → S3 cargado",
    f"Archivo: <code>{ruta_reporte_s2.name}</code> · "
    f"{len(reporte_s2)} evidencias metodológicas disponibles. "
    "El reporte será utilizado para trazabilidad y justificación, no como matriz de entrenamiento.",
    COLORS["blue"]
)
```

```python
# =============================================================================
# 2.2 EXTRACCIÓN DE DECISIONES METODOLÓGICAS PARA S3
# =============================================================================
estado_normalizado_s2 = reporte_s2["estado"].astype(str).str.strip().str.lower()
categoria_normalizada_s2 = reporte_s2["categoria"].astype(str).str.strip().str.lower()

evidencias_robustas_s2 = reporte_s2.loc[
    estado_normalizado_s2.str.contains(
        "robusta|validado|confiable|concordante",
        regex=True,
        na=False,
    )
].copy()

evidencias_cautela_s2 = reporte_s2.loc[
    estado_normalizado_s2.str.contains(
        "cautela|sensible",
        regex=True,
        na=False,
    )
].copy()

decisiones_s3 = reporte_s2.loc[
    categoria_normalizada_s2.str.contains(
        "decisión metodológica s3|decision metodologica s3",
        regex=True,
        na=False,
    )
].copy()

# Verificación explícita de la evidencia de colinealidad MP2.5–MP10.
evidencia_colinealidad_s2 = reporte_s2.loc[
    reporte_s2["elemento"].astype(str).str.contains(
        "colinealidad.*mp2.5.*mp10|mp2_5_mean.*mp10_mean",
        case=False,
        regex=True,
        na=False,
    )
].copy()

resumen_trazabilidad_s2 = pd.DataFrame([{
    "archivo_s2": ruta_reporte_s2.name,
    "evidencias_totales": len(reporte_s2),
    "evidencias_robustas_o_validadas": len(evidencias_robustas_s2),
    "resultados_sensibles_o_con_cautela": len(evidencias_cautela_s2),
    "decisiones_metodologicas_s3": len(decisiones_s3),
    "evidencias_colinealidad_mp25_mp10": len(evidencia_colinealidad_s2),
    "uso": "Selección de variables, control de colinealidad, tratamiento de faltantes y comunicación",
}])

resumen_trazabilidad_s2.to_csv(
    RESULTS_DIR / "s3_trazabilidad_reporte_s2.csv",
    index=False,
    encoding="utf-8-sig",
)

decisiones_s3.to_csv(
    RESULTS_DIR / "s3_decisiones_metodologicas_importadas_desde_s2.csv",
    index=False,
    encoding="utf-8-sig",
)

display(resumen_trazabilidad_s2)

if not decisiones_s3.empty:
    display(decisiones_s3[[
        "elemento",
        "resultado_validado",
        "estado",
        "uso_s3",
    ]])
else:
    print("Advertencia: no se identificaron filas con categoría 'Decisión metodológica S3'.")

if not evidencia_colinealidad_s2.empty:
    display(evidencia_colinealidad_s2[[
        "categoria",
        "elemento",
        "resultado_validado",
        "estado",
        "uso_s3",
    ]])
else:
    print(
        "Advertencia: el reporte S2 no contiene una fila explícita de "
        "colinealidad MP2.5–MP10; el notebook mantendrá igualmente el control VIF."
    )

tarjeta(
    "📌 Trazabilidad metodológica consolidada",
    f"Se identificaron <b>{len(evidencias_robustas_s2)}</b> evidencias robustas o validadas, "
    f"<b>{len(evidencias_cautela_s2)}</b> resultados sensibles y "
    f"<b>{len(decisiones_s3)}</b> decisiones metodológicas para S3. "
    "La síntesis quedó exportada en <code>s3_trazabilidad_reporte_s2.csv</code>.",
    COLORS["green"]
)
```

```python
# =============================================================================
# 3. OBJETIVO BINARIO, FALTANTES Y CONTROL DE FUGA DE INFORMACIÓN
# =============================================================================
def normalizar_binaria(s):
    s = pd.Series(s).copy()
    if pd.api.types.is_bool_dtype(s):
        return s.astype(float)
    if pd.api.types.is_numeric_dtype(s):
        out = pd.to_numeric(s, errors="coerce")
        return out.where(out.isin([0, 1]))
    mapa = {
        "1":1, "0":0, "si":1, "sí":1, "true":1, "verdadero":1,
        "no":0, "false":0, "falso":0, "mala":1, "buena":0
    }
    return s.astype(str).str.strip().str.lower().map(mapa)

TARGET = "mala_calidad_mp25"
if TARGET in df_raw.columns:
    y_all = normalizar_binaria(df_raw[TARGET])
    fuente_target = "columna original mala_calidad_mp25"
elif "calidad_aire_mp25" in df_raw.columns:
    calidad = df_raw["calidad_aire_mp25"].astype(str).str.strip().str.lower()
    y_all = calidad.map(lambda x: 0 if x == "bueno" else (1 if x not in ["nan", "none", ""] else np.nan))
    fuente_target = "derivada de calidad_aire_mp25 (distinto de Bueno)"
elif "mp2_5_mean" in df_raw.columns:
    y_all = (pd.to_numeric(df_raw["mp2_5_mean"], errors="coerce") > 50).astype(float)
    y_all[pd.to_numeric(df_raw["mp2_5_mean"], errors="coerce").isna()] = np.nan
    fuente_target = "fallback regulatorio MP2.5 > 50 µg/m³"
else:
    raise ValueError("No existe una variable que permita construir la etiqueta binaria mala_calidad_mp25.")

df = df_raw.copy()
df[TARGET] = y_all
filas_sin_target = int(df[TARGET].isna().sum())
df = df.loc[df[TARGET].notna()].copy()
df[TARGET] = df[TARGET].astype(int)

# Auditoría de faltantes completa.
auditoria_faltantes = pd.DataFrame({
    "variable": df_raw.columns,
    "n_faltantes": df_raw.isna().sum().values,
    "porcentaje_faltantes": 100 * df_raw.isna().mean().values,
    "tipo": [str(t) for t in df_raw.dtypes]
}).sort_values("porcentaje_faltantes", ascending=False)
auditoria_faltantes.to_csv(RESULTS_DIR / "s3_auditoria_faltantes.csv", index=False, encoding="utf-8-sig")

resumen_target = pd.DataFrame([{
    "fuente_target": fuente_target,
    "filas_originales": len(df_raw),
    "filas_sin_target_excluidas": filas_sin_target,
    "filas_modelables": len(df),
    "clase_positiva_n": int(df[TARGET].sum()),
    "prevalencia_positiva": df[TARGET].mean()
}])
resumen_target.to_csv(RESULTS_DIR / "s3_resumen_target.csv", index=False, encoding="utf-8-sig")

display(resumen_target)
display(auditoria_faltantes.head(15))

fig, ax = plt.subplots(figsize=(11, 6))
top = auditoria_faltantes.head(15).sort_values("porcentaje_faltantes")
ax.barh(top["variable"], top["porcentaje_faltantes"])
ax.axvline(27.8, linestyle="--", linewidth=1.8, label="Referencia S1 MP2.5 ≈ 27,8%")
ax.set_title("Variables con mayor proporción de datos faltantes")
ax.set_xlabel("Porcentaje de faltantes")
ax.set_ylabel("")
ax.legend()
guardar_figura("01_faltantes_top15.png")
plt.show()
```

<p style="margin:0 0 10px 0;">Las especificaciones se construyen de forma incremental para distinguir señal física, señal territorial-temporal y señal integrada:</p>

<li><b>M3 · Integrado sin fuga:</b> M2 + MP10 y variables de cobertura disponibles. MP2.5 queda excluido.</li>

```python
# =============================================================================
# 4. INGENIERÍA DE VARIABLES, ALIAS Y ESPECIFICACIONES DE MODELO
# =============================================================================
def primera_columna(columnas, candidatas):
    return next((c for c in candidatas if c in columnas), None)

ALIASES = {
    "temperatura": ["temperatura_mean", "temperatura_media", "temp_mean", "temp_media"],
    "humedad": ["humedad_mean", "humedad_media", "rh_mean"],
    "viento": ["viento_mean", "viento_media", "velocidad_viento_mean"],
    "radiacion": ["radiacion_mean", "radiacion_media", "radiacion_solar_mean"],
    "presion": ["presion_mean", "presion_media", "pressure_mean"],
    "inversion": ["prop_inversion", "inversion_termica", "alta_inversion", "inversion"],
    "mp10": ["mp10_mean", "mp10_media", "MP10"],
    "fin_semana": ["es_finde", "fin_semana", "es_fin_semana"],
    "festivo": ["es_festivo", "festivo"],
    "gec": ["periodo_gec", "periodo_GEC", "gec"],
    "cobertura": ["cobertura_horaria", "cobertura_mp25", "horas_validas_mp25", "n_horas_mp25", "horas_validas"]
}
cols = set(df.columns)
mapa = {k: primera_columna(cols, v) for k, v in ALIASES.items()}

# Variables temporales seguras derivadas exclusivamente de fecha.
df["mes"] = df[col_fecha].dt.month.astype(int)
df["mes_sin"] = np.sin(2 * np.pi * df["mes"] / 12)
df["mes_cos"] = np.cos(2 * np.pi * df["mes"] / 12)
df["invierno"] = df["mes"].isin([5, 6, 7, 8]).astype(int)
df["anio"] = df[col_fecha].dt.year.astype(int)

# Normalizar binarias candidatas sin destruir NaN.
for k in ["fin_semana", "festivo", "gec"]:
    c = mapa.get(k)
    if c is not None:
        df[c] = normalizar_binaria(df[c])

categoricas_posibles = [
    c for c in ["zona_geografica", "nivel_socioeconomico", "estacion_anio", "temporada_critica"]
    if c in df.columns
]
meteo = [mapa[k] for k in ["temperatura", "humedad", "viento", "radiacion", "presion", "inversion"] if mapa[k] is not None]
temporales = ["mes_sin", "mes_cos", "invierno"] + [mapa[k] for k in ["fin_semana", "festivo", "gec"] if mapa[k] is not None]
integradas = [mapa[k] for k in ["mp10", "cobertura"] if mapa[k] is not None]

# Blacklist explícita de variables con fuga o derivadas del objetivo.
BLACKLIST_FUGA = {
    TARGET, "mp2_5_mean", "mp2_5_max", "calidad_aire_mp25", "critico_mp25_dia",
    "nivel_contaminacion", "estaciones_en_episodio"
}

def limpiar_lista(lista):
    out=[]
    for c in lista:
        if c is not None and c in df.columns and c not in BLACKLIST_FUGA and c not in out:
            out.append(c)
    return out

modelos = {
    "M1_Meteorologico": limpiar_lista(meteo),
    "M2_Territorial_Temporal": limpiar_lista(meteo + temporales + categoricas_posibles),
    "M3_Integrado_Sin_Fuga": limpiar_lista(meteo + temporales + categoricas_posibles + integradas),
}

# Respaldo si una familia no existe en el CSV.
seguras_numericas = [c for c in df.select_dtypes(include=np.number).columns if c not in BLACKLIST_FUGA and c not in ["anio", "mes"]]
for nombre, features in modelos.items():
    if len(features) < 2:
        modelos[nombre] = limpiar_lista(features + seguras_numericas[:5])
    if len(modelos[nombre]) < 2:
        raise ValueError(f"{nombre} no dispone de al menos dos predictores seguros.")

matriz_modelos = []
for nombre, features in modelos.items():
    for c in features:
        matriz_modelos.append({
            "modelo": nombre,
            "variable": c,
            "tipo": "categórica" if (c in categoricas_posibles or not pd.api.types.is_numeric_dtype(df[c])) else "numérica",
            "faltantes_%": 100 * df[c].isna().mean(),
            "justificacion": "S1/S2: meteorología, temporalidad, territorio, inequidad o proxy MP10 sin usar MP2.5"
        })
matriz_modelos = pd.DataFrame(matriz_modelos)
matriz_modelos.to_csv(RESULTS_DIR / "s3_matriz_variables_modelos.csv", index=False, encoding="utf-8-sig")
display(matriz_modelos)

print("Columnas excluidas por fuga:", sorted(c for c in BLACKLIST_FUGA if c in df.columns))
```

<h2 style="margin:0 0 10px 0;font-size:24px;">5. Separación temporal y prevención de optimismo</h2>
<p style="margin:0;">El 80% inicial de fechas se usa para entrenamiento y el 20% final para prueba. Dentro del entrenamiento se reserva un bloque temporal de validación para seleccionar el umbral. Así, el futuro nunca participa en el ajuste del modelo ni del umbral.</p>

```python
# =============================================================================
# 5. SPLIT TEMPORAL TRAIN / TEST Y DESARROLLO / VALIDACIÓN
# =============================================================================
fechas_unicas = np.array(sorted(df[col_fecha].dropna().dt.normalize().unique()))
if len(fechas_unicas) < 10:
    raise ValueError("Se requieren al menos 10 fechas distintas para una validación temporal defendible.")

cut_test = fechas_unicas[max(1, int(len(fechas_unicas) * 0.80))]
train_raw = df.loc[df[col_fecha].dt.normalize() < cut_test].copy()
test_raw = df.loc[df[col_fecha].dt.normalize() >= cut_test].copy()

fechas_train = np.array(sorted(train_raw[col_fecha].dt.normalize().unique()))
cut_val = fechas_train[max(1, int(len(fechas_train) * 0.80))]
dev_raw = train_raw.loc[train_raw[col_fecha].dt.normalize() < cut_val].copy()
val_raw = train_raw.loc[train_raw[col_fecha].dt.normalize() >= cut_val].copy()

if min(dev_raw[TARGET].nunique(), val_raw[TARGET].nunique(), test_raw[TARGET].nunique()) < 2:
    raise ValueError("Alguna partición temporal contiene una sola clase. Ajusta el corte o amplía el periodo.")

particiones = pd.DataFrame([
    {"particion":"desarrollo", "n":len(dev_raw), "fecha_min":dev_raw[col_fecha].min(), "fecha_max":dev_raw[col_fecha].max(), "prevalencia":dev_raw[TARGET].mean()},
    {"particion":"validacion", "n":len(val_raw), "fecha_min":val_raw[col_fecha].min(), "fecha_max":val_raw[col_fecha].max(), "prevalencia":val_raw[TARGET].mean()},
    {"particion":"entrenamiento_total", "n":len(train_raw), "fecha_min":train_raw[col_fecha].min(), "fecha_max":train_raw[col_fecha].max(), "prevalencia":train_raw[TARGET].mean()},
    {"particion":"prueba_futura", "n":len(test_raw), "fecha_min":test_raw[col_fecha].min(), "fecha_max":test_raw[col_fecha].max(), "prevalencia":test_raw[TARGET].mean()},
])
particiones.to_csv(RESULTS_DIR / "s3_particiones_temporales.csv", index=False, encoding="utf-8-sig")
display(particiones)
```

<div style="background:#ffffff;border:1px solid #dbe5ef;border-left:7px solid #d97706;border-radius:14px;padding:22px 26px;margin:18px 0;font-family:Arial,Helvetica,sans-serif;color:#0f172a;line-height:1.65;">
<h2 style="margin:0 0 10px 0;font-size:24px;">6. Tres estrategias de manejo de faltantes</h2>
<ul style="margin:0 0 0 20px;padding:0;">
<li><b>Casos completos:</b> elimina filas con algún predictor faltante. Es transparente, pero puede perder muestra y sesgar el periodo.</li>
<li><b>Mediana/moda:</b> imputación simple ajustada solo con entrenamiento.</li>
<li><b>Regresión multivariada:</b> <code>IterativeImputer</code> con <code>BayesianRidge</code>, ajustado solo con entrenamiento; estima cada variable numérica usando las demás.</li>
</ul>
<p style="margin:10px 0 0 0;">La comparación incluye retención de datos y un experimento de enmascaramiento artificial para medir MAE/RMSE de la imputación.</p>
</div>

```python
# =============================================================================
# 6. CONSTRUCCIÓN DE PREPROCESADORES PARA LAS TRES ESTRATEGIAS
# =============================================================================
def crear_ohe():
    try:
        return OneHotEncoder(handle_unknown="ignore", drop="first", sparse_output=False)
    except TypeError:
        return OneHotEncoder(handle_unknown="ignore", drop="first", sparse=False)

def separar_tipos(frame, features):
    cats = [c for c in features if (not pd.api.types.is_numeric_dtype(frame[c]))]
    nums = [c for c in features if c not in cats]
    return nums, cats

def construir_preprocesador(frame, features, estrategia):
    nums, cats = separar_tipos(frame, features)
    transformers=[]
    if nums:
        if estrategia == "mediana":
            imp_num = SimpleImputer(strategy="median")
        elif estrategia == "regresion":
            imp_num = IterativeImputer(
                estimator=BayesianRidge(), max_iter=20, tol=1e-3,
                random_state=SEED, sample_posterior=False,
                initial_strategy="median", skip_complete=True
            )
        elif estrategia == "completos":
            imp_num = "passthrough"
        else:
            raise ValueError(estrategia)
        pasos_num=[]
        if imp_num != "passthrough":
            pasos_num.append(("imputer", imp_num))
        pasos_num.append(("scaler", StandardScaler()))
        transformers.append(("num", Pipeline(pasos_num), nums))
    if cats:
        if estrategia == "completos":
            pasos_cat=[("ohe", crear_ohe())]
        else:
            pasos_cat=[("imputer", SimpleImputer(strategy="most_frequent")), ("ohe", crear_ohe())]
        transformers.append(("cat", Pipeline(pasos_cat), cats))
    return ColumnTransformer(transformers, remainder="drop", verbose_feature_names_out=False), nums, cats

def filtrar_completos(frame, features):
    return frame.dropna(subset=features + [TARGET]).copy()

# Retención por modelo y estrategia.
retencion=[]
for nombre, features in modelos.items():
    for est in ["completos", "mediana", "regresion"]:
        for part_name, part in [("train", train_raw), ("test", test_raw)]:
            usado = filtrar_completos(part, features) if est == "completos" else part
            retencion.append({
                "modelo":nombre, "estrategia":est, "particion":part_name,
                "n_disponible":len(part), "n_utilizado":len(usado),
                "retencion":len(usado)/max(len(part),1), "prevalencia":usado[TARGET].mean()
            })
retencion = pd.DataFrame(retencion)
retencion.to_csv(RESULTS_DIR / "s3_retencion_por_estrategia.csv", index=False, encoding="utf-8-sig")
display(retencion)
```

```python
# =============================================================================
# 6.1 VALIDACIÓN DE IMPUTACIÓN POR ENMASCARAMIENTO ARTIFICIAL
# -----------------------------------------------------------------------------
# Se enmascara una fracción de valores originalmente observados del conjunto de
# entrenamiento. Luego se compara mediana versus regresión en MAE y RMSE.
# =============================================================================
features_union = sorted(set(sum(modelos.values(), [])))
nums_union, _ = separar_tipos(train_raw, features_union)
nums_validables = [c for c in nums_union if train_raw[c].notna().sum() >= 100 and train_raw[c].nunique(dropna=True) > 5]

resultados_mask=[]
if len(nums_validables) >= 2:
    base_num = train_raw[nums_validables].apply(pd.to_numeric, errors="coerce")
    for rep in range(N_MASK_REPEATS):
        rng_rep = np.random.default_rng(SEED + rep)
        masked = base_num.copy()
        mascara_por_col={}
        for c in nums_validables:
            idx_obs = np.flatnonzero(base_num[c].notna().to_numpy())
            n_mask = max(1, int(len(idx_obs) * 0.10))
            idx_mask = rng_rep.choice(idx_obs, size=n_mask, replace=False)
            mask = np.zeros(len(base_num), dtype=bool); mask[idx_mask] = True
            mascara_por_col[c] = mask
            masked.loc[mask, c] = np.nan

        imp_med = SimpleImputer(strategy="median")
        med_arr = imp_med.fit_transform(masked)
        imp_reg = IterativeImputer(
            estimator=BayesianRidge(), max_iter=20, random_state=SEED+rep,
            initial_strategy="median", sample_posterior=False, skip_complete=False
        )
        reg_arr = imp_reg.fit_transform(masked)
        med_df = pd.DataFrame(med_arr, columns=nums_validables, index=base_num.index)
        reg_df = pd.DataFrame(reg_arr, columns=nums_validables, index=base_num.index)

        for c in nums_validables:
            mask = mascara_por_col[c]
            real = base_num.loc[mask, c].astype(float)
            for metodo, pred in [("mediana", med_df.loc[mask, c]), ("regresion", reg_df.loc[mask, c])]:
                mae = np.mean(np.abs(real - pred))
                rmse = np.sqrt(np.mean((real - pred)**2))
                resultados_mask.append({"repeticion":rep+1, "variable":c, "metodo":metodo, "mae":mae, "rmse":rmse, "n_enmascarado":int(mask.sum())})

validacion_imputacion = pd.DataFrame(resultados_mask)
if not validacion_imputacion.empty:
    resumen_imputacion = (validacion_imputacion.groupby(["variable","metodo"], as_index=False)
                           .agg(mae=("mae","mean"), rmse=("rmse","mean"), n_enmascarado=("n_enmascarado","sum")))
    resumen_global = resumen_imputacion.groupby("metodo", as_index=False).agg(mae_promedio=("mae","mean"), rmse_promedio=("rmse","mean"))
else:
    resumen_imputacion = pd.DataFrame(columns=["variable","metodo","mae","rmse","n_enmascarado"])
    resumen_global = pd.DataFrame(columns=["metodo","mae_promedio","rmse_promedio"])

resumen_imputacion.to_csv(RESULTS_DIR / "s3_validacion_imputacion_por_variable.csv", index=False, encoding="utf-8-sig")
resumen_global.to_csv(RESULTS_DIR / "s3_validacion_imputacion_global.csv", index=False, encoding="utf-8-sig")
display(resumen_global)
display(resumen_imputacion.head(20))
```

<div style="background:#ffffff;border:1px solid #dbe5ef;border-left:7px solid #2563eb;border-radius:14px;padding:22px 26px;margin:18px 0;font-family:Arial,Helvetica,sans-serif;color:#0f172a;line-height:1.65;">
<h2 style="margin:0 0 10px 0;font-size:24px;">7. Entrenamiento de 9 combinaciones y selección de umbral</h2>
<p style="margin:0 0 10px 0;">Cada uno de los tres modelos se entrena bajo las tres estrategias de datos. Se usa regresión logística con ponderación balanceada y regularización prácticamente nula (<code>C=1e6</code>) para conservar interpretabilidad.</p>
<p style="margin:0;">El umbral no se fija automáticamente en 0,50: se selecciona en validación maximizando F2, que penaliza con mayor fuerza los falsos negativos y prioriza Recall.</p>
</div>

```python
# =============================================================================
# 7. FUNCIONES DE ENTRENAMIENTO, UMBRAL Y MÉTRICAS
# =============================================================================
def construir_pipeline(frame, features, estrategia):
    prep, nums, cats = construir_preprocesador(frame, features, estrategia)
    clf = LogisticRegression(
        penalty="l2", C=1e6, solver="lbfgs", max_iter=5000,
        class_weight="balanced", random_state=SEED
    )
    return Pipeline([("prep", prep), ("clf", clf)]), nums, cats

def seleccionar_umbral(y_true, prob):
    filas=[]
    for t in np.linspace(0.05, 0.95, 181):
        pred=(prob >= t).astype(int)
        filas.append({
            "umbral":t,
            "precision":precision_score(y_true,pred,zero_division=0),
            "recall":recall_score(y_true,pred,zero_division=0),
            "f1":f1_score(y_true,pred,zero_division=0),
            "f2":fbeta_score(y_true,pred,beta=2,zero_division=0)
        })
    tabla=pd.DataFrame(filas).sort_values(["f2","recall","precision"], ascending=False)
    return float(tabla.iloc[0]["umbral"]), tabla

def calcular_metricas(y_true, prob, umbral):
    pred=(prob >= umbral).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_true,pred,labels=[0,1]).ravel()
    especificidad = tn / (tn + fp) if (tn + fp) else np.nan
    return {
        "n_test":len(y_true), "positivos_test":int(np.sum(y_true)), "umbral":umbral,
        "accuracy":accuracy_score(y_true,pred),
        "balanced_accuracy":balanced_accuracy_score(y_true,pred),
        "precision":precision_score(y_true,pred,zero_division=0),
        "recall":recall_score(y_true,pred,zero_division=0),
        "specificity":especificidad,
        "f1":f1_score(y_true,pred,zero_division=0),
        "f2":fbeta_score(y_true,pred,beta=2,zero_division=0),
        "roc_auc":roc_auc_score(y_true,prob),
        "pr_auc":average_precision_score(y_true,prob),
        "brier":brier_score_loss(y_true,prob),
        "tn":tn, "fp":fp, "fn":fn, "tp":tp
    }

def ajustar_combinacion(nombre_modelo, features, estrategia):
    # Particiones nativas de la estrategia.
    dev = filtrar_completos(dev_raw, features) if estrategia == "completos" else dev_raw.copy()
    val = filtrar_completos(val_raw, features) if estrategia == "completos" else val_raw.copy()
    train = filtrar_completos(train_raw, features) if estrategia == "completos" else train_raw.copy()
    test = filtrar_completos(test_raw, features) if estrategia == "completos" else test_raw.copy()
    if min(len(dev),len(val),len(train),len(test)) == 0:
        raise ValueError(f"Sin filas disponibles para {nombre_modelo}/{estrategia}")
    if min(dev[TARGET].nunique(), val[TARGET].nunique(), train[TARGET].nunique(), test[TARGET].nunique()) < 2:
        raise ValueError(f"Una partición quedó con una sola clase en {nombre_modelo}/{estrategia}")

    pipe_val, nums, cats = construir_pipeline(dev, features, estrategia)
    pipe_val.fit(dev[features], dev[TARGET])
    p_val = pipe_val.predict_proba(val[features])[:,1]
    umbral, tabla_umbral = seleccionar_umbral(val[TARGET].to_numpy(), p_val)

    pipe_final, _, _ = construir_pipeline(train, features, estrategia)
    pipe_final.fit(train[features], train[TARGET])
    p_test = pipe_final.predict_proba(test[features])[:,1]
    metricas = calcular_metricas(test[TARGET].to_numpy(), p_test, umbral)
    metricas.update({
        "modelo":nombre_modelo, "estrategia":estrategia,
        "n_train":len(train), "n_features_base":len(features),
        "prevalencia_train":train[TARGET].mean(), "prevalencia_test":test[TARGET].mean()
    })
    return {
        "pipeline":pipe_final, "features":features, "nums":nums, "cats":cats,
        "train":train, "test":test, "prob_test":p_test,
        "umbral":umbral, "tabla_umbral":tabla_umbral,
        "metricas":metricas
    }
```

```python
# =============================================================================
# 7.1 ENTRENAMIENTO DE LAS NUEVE COMBINACIONES
# =============================================================================
artefactos={}
metricas=[]
inicio=time.time()
for nombre_modelo, features in modelos.items():
    for estrategia in ["completos", "mediana", "regresion"]:
        clave=(nombre_modelo, estrategia)
        print("Ajustando", clave)
        art=ajustar_combinacion(nombre_modelo, features, estrategia)
        artefactos[clave]=art
        metricas.append(art["metricas"])

metricas_modelos=pd.DataFrame(metricas)
metricas_modelos["score_decision"] = (
    0.30*metricas_modelos["recall"] +
    0.20*metricas_modelos["f1"] +
    0.20*metricas_modelos["f2"] +
    0.15*metricas_modelos["pr_auc"] +
    0.10*metricas_modelos["roc_auc"] +
    0.05*metricas_modelos["specificity"]
)
metricas_modelos = metricas_modelos.sort_values("score_decision", ascending=False).reset_index(drop=True)
metricas_modelos.to_csv(RESULTS_DIR / "s3_metricas_9_combinaciones.csv", index=False, encoding="utf-8-sig")
print(f"Tiempo total: {time.time()-inicio:.1f} s")
display(metricas_modelos)
```

<h2 style="margin:0 0 10px 0;font-size:24px;">9. Coeficientes, odds ratios, intervalos tradicionales y VIF</h2>
<p style="margin:0 0 10px 0;">Los coeficientes numéricos se interpretan por una desviación estándar, debido al escalamiento. Para variables categóricas, cada OR compara una categoría con la referencia eliminada por <i>dummy coding</i>.</p>
<p style="margin:0;"><b>OR &gt; 1</b> aumenta las odds de mala calidad; <b>OR &lt; 1</b> las reduce, manteniendo constantes las demás variables.</p>
</div>

```python
# =============================================================================
# 9. MATRIZ TRANSFORMADA, COEFICIENTES, OR E IC WALD
# =============================================================================
def matriz_transformada(artefacto, frame=None):
    frame = artefacto["train"] if frame is None else frame
    X = artefacto["pipeline"].named_steps["prep"].transform(frame[artefacto["features"]])
    nombres = artefacto["pipeline"].named_steps["prep"].get_feature_names_out()
    return np.asarray(X, dtype=float), np.asarray(nombres, dtype=str)

def pesos_balanceados(y):
    y=np.asarray(y,dtype=int); n=len(y); n0=np.sum(y==0); n1=np.sum(y==1)
    w0=n/(2*n0); w1=n/(2*n1)
    return np.where(y==1,w1,w0)

def tabla_wald(artefacto):
    X,nombres=matriz_transformada(artefacto)
    y=artefacto["train"][TARGET].to_numpy(dtype=int)
    clf=artefacto["pipeline"].named_steps["clf"]
    beta=np.r_[clf.intercept_,clf.coef_.ravel()]
    Xb=np.column_stack([np.ones(len(X)),X])
    p=np.clip(expit(Xb@beta),1e-8,1-1e-8)
    sw=pesos_balanceados(y)
    w=sw*p*(1-p)
    hess=Xb.T@(Xb*w[:,None])
    cov=np.linalg.pinv(hess)
    se=np.sqrt(np.clip(np.diag(cov),0,np.inf))
    z=stats.norm.ppf(1-ALPHA/2)
    li=beta-z*se; ls=beta+z*se
    nombres_full=np.concatenate((["Intercepto"], nombres))
    out=pd.DataFrame({"parametro":nombres_full,"coeficiente":beta,"se_wald":se,"ic_wald_li":li,"ic_wald_ls":ls})
    out["odds_ratio"]=np.exp(out["coeficiente"])
    out["or_ic_wald_li"]=np.exp(out["ic_wald_li"])
    out["or_ic_wald_ls"]=np.exp(out["ic_wald_ls"])
    out["p_wald"]=2*(1-stats.norm.cdf(np.abs(out["coeficiente"]/out["se_wald"].replace(0,np.nan))))
    out["significativo_95"]=(out["ic_wald_li"]*out["ic_wald_ls"]>0)
    return out, X, y, nombres

coef_wald, X_final, y_train_final, nombres_final = tabla_wald(FINAL)
coef_wald["abs_coef"] = coef_wald["coeficiente"].abs()
coef_wald = coef_wald.sort_values("abs_coef", ascending=False)
coef_wald.to_csv(RESULTS_DIR / "s3_coeficientes_or_ic_tradicional.csv", index=False, encoding="utf-8-sig")
display(coef_wald.head(25))

# VIF sobre predictores transformados; se limita a columnas no constantes.
varianzas = X_final.var(axis=0)
idx_vif = np.flatnonzero(varianzas > 1e-12)
X_vif = X_final[:,idx_vif]
nombres_vif = nombres_final[idx_vif]
if X_vif.shape[1] >= 2:
    vif = pd.DataFrame({
        "variable":nombres_vif,
        "VIF":[variance_inflation_factor(X_vif,i) for i in range(X_vif.shape[1])]
    }).sort_values("VIF",ascending=False)
else:
    vif = pd.DataFrame({"variable":nombres_vif,"VIF":[1.0]*len(nombres_vif)})
vif["nivel"] = pd.cut(vif["VIF"], bins=[-np.inf,5,10,np.inf], labels=["controlado","revisar","alto"])
vif.to_csv(RESULTS_DIR / "s3_vif_modelo_final.csv", index=False, encoding="utf-8-sig")
display(vif.head(20))
```

<h2 style="margin:0 0 10px 0;font-size:24px;">10. Bootstrap estratificado del modelo final</h2>
<p style="margin:0 0 10px 0;">Se remuestrean observaciones de entrenamiento dentro de cada clase, preservando el desbalance. En cada remuestra se vuelve a estimar la regresión logística sobre la matriz ya transformada por el preprocesamiento final.</p>
<p style="margin:0;">Se calculan IC percentil al 95%, sesgo, dispersión, consistencia de signo y comparación con los IC Wald. Un parámetro se marca inestable si su IC bootstrap cruza cero o si mantiene el mismo signo en menos del 90% de las remuestras.</p>
</div>

```python
# =============================================================================
# 10. BOOTSTRAP ESTRATIFICADO DE COEFICIENTES DEL MODELO FINAL
# =============================================================================
def bootstrap_logistico(X, y, n_boot=N_BOOT_MODEL, seed=SEED):
    rngb=np.random.default_rng(seed)
    idx0=np.flatnonzero(y==0); idx1=np.flatnonzero(y==1)
    betas=[]; fallos=0
    for b in range(n_boot):
        ib=np.r_[rngb.choice(idx0,size=len(idx0),replace=True), rngb.choice(idx1,size=len(idx1),replace=True)]
        rngb.shuffle(ib)
        clf=LogisticRegression(penalty="l2",C=1e6,solver="lbfgs",max_iter=3000,class_weight="balanced",random_state=seed+b)
        try:
            clf.fit(X[ib],y[ib])
            betas.append(np.r_[clf.intercept_,clf.coef_.ravel()])
        except Exception:
            fallos += 1
    return np.asarray(betas), fallos

inicio=time.time()
betas_boot, fallos_boot = bootstrap_logistico(X_final,y_train_final)
if len(betas_boot) < max(50, int(0.8*N_BOOT_MODEL)):
    raise RuntimeError("Demasiadas remuestras bootstrap fallaron; revisar separación o colinealidad.")

parametros=np.concatenate((["Intercepto"], nombres_final))
coef_original = np.r_[FINAL["pipeline"].named_steps["clf"].intercept_, FINAL["pipeline"].named_steps["clf"].coef_.ravel()]
li_b=np.percentile(betas_boot,2.5,axis=0); ls_b=np.percentile(betas_boot,97.5,axis=0)
media_b=betas_boot.mean(axis=0); sd_b=betas_boot.std(axis=0,ddof=1)
signo_cons=np.mean(np.sign(betas_boot)==np.sign(coef_original),axis=0)

bootstrap_coef=pd.DataFrame({
    "parametro":parametros,"coef_original":coef_original,"media_bootstrap":media_b,
    "sesgo_bootstrap":media_b-coef_original,"sd_bootstrap":sd_b,
    "ic_boot_li":li_b,"ic_boot_ls":ls_b,"signo_consistente":signo_cons
})
bootstrap_coef["or_original"]=np.exp(bootstrap_coef["coef_original"])
bootstrap_coef["or_ic_boot_li"]=np.exp(bootstrap_coef["ic_boot_li"])
bootstrap_coef["or_ic_boot_ls"]=np.exp(bootstrap_coef["ic_boot_ls"])
bootstrap_coef["inestable"]=(bootstrap_coef["ic_boot_li"]<=0)&(bootstrap_coef["ic_boot_ls"]>=0) | (bootstrap_coef["signo_consistente"]<0.90)

comparacion_ic=bootstrap_coef.merge(coef_wald.drop(columns=["abs_coef"]),on="parametro",how="left")
comparacion_ic["ancho_ic_boot"]=comparacion_ic["ic_boot_ls"]-comparacion_ic["ic_boot_li"]
comparacion_ic["ancho_ic_wald"]=comparacion_ic["ic_wald_ls"]-comparacion_ic["ic_wald_li"]
comparacion_ic["ratio_ancho_boot_wald"]=comparacion_ic["ancho_ic_boot"]/comparacion_ic["ancho_ic_wald"].replace(0,np.nan)
comparacion_ic.to_csv(RESULTS_DIR / "s3_bootstrap_coeficientes_or.csv",index=False,encoding="utf-8-sig")
np.save(RESULTS_DIR / "s3_bootstrap_coeficientes.npy",betas_boot)

resumen_boot=pd.DataFrame([{
    "remuestras_solicitadas":N_BOOT_MODEL,"remuestras_exitosas":len(betas_boot),"fallos":fallos_boot,
    "tasa_exito":len(betas_boot)/N_BOOT_MODEL,"parametros_inestables":int(bootstrap_coef["inestable"].sum()),
    "tiempo_segundos":time.time()-inicio
}])
resumen_boot.to_csv(RESULTS_DIR / "s3_resumen_bootstrap.csv",index=False,encoding="utf-8-sig")
display(resumen_boot)
display(comparacion_ic.sort_values("inestable",ascending=False).head(30))
```

<p style="margin:0;">La comparación final no se limita a AUC. Se revisan retención, Recall, F1, PR-AUC, coeficientes estandarizados y amplitud de intervalos del mismo modelo integrado bajo casos completos, mediana y regresión.</p>

```python
# =============================================================================
# 11. COEFICIENTES E IC DEL MODELO INTEGRADO BAJO CADA ESTRATEGIA
# =============================================================================
modelo_comparacion="M3_Integrado_Sin_Fuga"
comparacion_coef=[]
for est in ["completos","mediana","regresion"]:
    art=artefactos[(modelo_comparacion,est)]
    tabla,_,_,_=tabla_wald(art)
    for _,r in tabla.iterrows():
        comparacion_coef.append({
            "estrategia":est,"parametro":r["parametro"],"coeficiente":r["coeficiente"],
            "odds_ratio":r["odds_ratio"],"ic_li":r["ic_wald_li"],"ic_ls":r["ic_wald_ls"],
            "ancho_ic":r["ic_wald_ls"]-r["ic_wald_li"]
        })
comparacion_coef=pd.DataFrame(comparacion_coef)
comparacion_coef.to_csv(RESULTS_DIR / "s3_impacto_imputacion_coeficientes_ic.csv",index=False,encoding="utf-8-sig")

metricas_imp = metricas_modelos.loc[metricas_modelos["modelo"]==modelo_comparacion,
    ["estrategia","n_train","n_test","recall","precision","f1","f2","roc_auc","pr_auc","brier","score_decision"]].copy()
if not resumen_global.empty:
    metricas_imp = metricas_imp.merge(resumen_global.rename(columns={"metodo":"estrategia"}),on="estrategia",how="left")
metricas_imp.to_csv(RESULTS_DIR / "s3_impacto_imputacion_metricas.csv",index=False,encoding="utf-8-sig")
display(metricas_imp)

# Dispersión de coeficientes comunes entre estrategias.
comunes=(comparacion_coef.groupby("parametro")["estrategia"].nunique())
comunes=comunes[comunes==3].index
coef_pivot=(comparacion_coef.loc[comparacion_coef["parametro"].isin(comunes)]
            .pivot(index="parametro",columns="estrategia",values="coeficiente"))
coef_pivot["rango_coef"] = coef_pivot.max(axis=1)-coef_pivot.min(axis=1)
coef_pivot=coef_pivot.sort_values("rango_coef",ascending=False)
display(coef_pivot.head(20))
coef_pivot.to_csv(RESULTS_DIR / "s3_sensibilidad_coeficientes_imputacion.csv",encoding="utf-8-sig")
```

```python
# =============================================================================
# 11.1 DECISIÓN FUNDAMENTADA DE IMPUTACIÓN
# =============================================================================
filas_decision=[]
for est in ["completos","mediana","regresion"]:
    m=metricas_imp.loc[metricas_imp["estrategia"]==est].iloc[0]
    ret=float(retencion.query("modelo==@modelo_comparacion and estrategia==@est and particion=='train'")["retencion"].iloc[0])
    rmse=np.nan
    if not resumen_global.empty and est in set(resumen_global["metodo"]):
        rmse=float(resumen_global.loc[resumen_global["metodo"]==est,"rmse_promedio"].iloc[0])
    filas_decision.append({
        "estrategia":est,"retencion_train":ret,"recall":m["recall"],"f1":m["f1"],"f2":m["f2"],
        "pr_auc":m["pr_auc"],"roc_auc":m["roc_auc"],"brier":m["brier"],"rmse_imputacion":rmse
    })
decision_imp=pd.DataFrame(filas_decision)
decision_imp["puntaje_compuesto"]=(
    .25*decision_imp["retencion_train"] + .25*decision_imp["recall"] + .15*decision_imp["f1"] +
    .15*decision_imp["f2"] + .10*decision_imp["pr_auc"] + .10*decision_imp["roc_auc"]
)
# La calidad de imputación se usa como criterio de desempate, no como sustituto de desempeño.
if decision_imp["rmse_imputacion"].notna().any():
    rmse_norm=(decision_imp["rmse_imputacion"]-decision_imp["rmse_imputacion"].min())/(decision_imp["rmse_imputacion"].max()-decision_imp["rmse_imputacion"].min()+1e-12)
    decision_imp["puntaje_compuesto"] -= .03*rmse_norm.fillna(0)
decision_imp=decision_imp.sort_values("puntaje_compuesto",ascending=False)
ESTRATEGIA_RECOMENDADA=decision_imp.iloc[0]["estrategia"]
decision_imp["recomendacion"] = np.where(decision_imp["estrategia"]==ESTRATEGIA_RECOMENDADA,"RECOMENDADA","alternativa/sensibilidad")
decision_imp.to_csv(RESULTS_DIR / "s3_decision_imputacion.csv",index=False,encoding="utf-8-sig")
display(decision_imp)

tarjeta(
    "📌 Decisión de imputación",
    f"La estrategia recomendada por retención, desempeño predictivo y calidad de imputación es <b>{ESTRATEGIA_RECOMENDADA}</b>. "
    "Casos completos se conserva como análisis de sensibilidad para cuantificar el sesgo potencial por exclusión.",
    COLORS["amber"]
)
```

```python
# =============================================================================
# 12. CALIBRACIÓN Y SENSIBILIDAD AL UMBRAL
# =============================================================================
frac_pos, mean_pred = calibration_curve(y_test,p_test,n_bins=10,strategy="quantile")
fig,ax=plt.subplots(figsize=(7.5,6))
ax.plot(mean_pred,frac_pos,marker="o",label="Modelo final")
ax.plot([0,1],[0,1],linestyle="--",label="Calibración perfecta")
ax.set_xlabel("Probabilidad predicha media"); ax.set_ylabel("Frecuencia observada")
ax.set_title("Curva de calibración · modelo final"); ax.legend()
guardar_figura("08_calibracion_modelo_final.png"); plt.show()

sens=[]
for t in np.linspace(.05,.80,76):
    pred=(p_test>=t).astype(int)
    sens.append({"umbral":t,"precision":precision_score(y_test,pred,zero_division=0),"recall":recall_score(y_test,pred,zero_division=0),"f1":f1_score(y_test,pred,zero_division=0),"f2":fbeta_score(y_test,pred,beta=2,zero_division=0)})
sens=pd.DataFrame(sens)
sens.to_csv(RESULTS_DIR / "s3_sensibilidad_umbral.csv",index=False,encoding="utf-8-sig")
fig,ax=plt.subplots(figsize=(9.5,6))
for c in ["precision","recall","f1","f2"]: ax.plot(sens["umbral"],sens[c],label=c.upper())
ax.axvline(FINAL["umbral"],linestyle="--",label=f"Umbral elegido={FINAL['umbral']:.2f}")
ax.set_xlabel("Umbral"); ax.set_ylabel("Métrica"); ax.set_ylim(0,1)
ax.set_title("Sensibilidad operacional al umbral"); ax.legend(ncol=3)
guardar_figura("09_sensibilidad_umbral.png"); plt.show()
```

```python
# =============================================================================
# 12.1 DESEMPEÑO POR SUBGRUPOS TERRITORIALES Y SOCIOECONÓMICOS
# =============================================================================
test_diag=FINAL["test"].copy()
test_diag["probabilidad"] = p_test
test_diag["prediccion"] = pred_test

def metricas_grupo(g):
    if g[TARGET].nunique()<2 or len(g)<20:
        return pd.Series({"n":len(g),"prevalencia":g[TARGET].mean(),"recall":np.nan,"precision":np.nan,"f1":np.nan,"roc_auc":np.nan})
    return pd.Series({
        "n":len(g),"prevalencia":g[TARGET].mean(),
        "recall":recall_score(g[TARGET],g["prediccion"],zero_division=0),
        "precision":precision_score(g[TARGET],g["prediccion"],zero_division=0),
        "f1":f1_score(g[TARGET],g["prediccion"],zero_division=0),
        "roc_auc":roc_auc_score(g[TARGET],g["probabilidad"])
    })

diagnosticos=[]
for col in ["zona_geografica","nivel_socioeconomico"]:
    if col in test_diag.columns:
        t=test_diag.groupby(col,observed=True).apply(metricas_grupo,include_groups=False).reset_index()
        t.insert(0,"dimension",col)
        t=t.rename(columns={col:"grupo"})
        diagnosticos.append(t)
diagnostico_grupos=pd.concat(diagnosticos,ignore_index=True) if diagnosticos else pd.DataFrame(columns=["dimension","grupo","n","prevalencia","recall","precision","f1","roc_auc"])
diagnostico_grupos.to_csv(RESULTS_DIR / "s3_diagnostico_subgrupos.csv",index=False,encoding="utf-8-sig")
display(diagnostico_grupos)
```
